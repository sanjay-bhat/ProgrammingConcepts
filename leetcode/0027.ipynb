# LeetCode #27: Remove Element

https://leetcode.com/problems/remove-element/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Two Pointers** ★ | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Scan for each occurrence of val and left-shift all subsequent elements to fill the gap. Each removal costs $O(n)$ for the shift, giving $O(n^2)$ total for up to n removals.

### Optimal: Two Pointers ★
Write pointer `k` starts at 0. Scan left-to-right; copy `nums[i]` to `nums[k]` only when `nums[i] != val`, then advance `k`. One pass, no shifting.

**Why this is better than Brute Force:** Replaces $O(n)$ per-removal shifting with a single $O(1)$ write, dropping total time from $O(n^2)$ to $O(n)$.

**Constraints:**
* 0 ≤ nums.length ≤ 100
* 0 ≤ nums[i] ≤ 50
* 0 ≤ val ≤ 100

## Solutions

### C#

```
public class Solution {
    public int RemoveElement(int[] nums, int val) {
        int k = 0;
        for (int i = 0; i < nums.Length; i++) {
            // Keep only elements that aren't the target value
            if (nums[i] != val)
                nums[k++] = nums[i];
        }
        return k;
    }
}
```

### Python

```
class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        k = 0
        for i in range(len(nums)):
            # Keep only elements that aren't the target value
            if nums[i] != val:
                nums[k] = nums[i]
                k += 1
        return k
```

### Rust

```
pub fn remove_element(nums: &mut Vec<i32>, val: i32) -> i32 {
    let mut k = 0usize;
    for i in 0..nums.len() {
        // Keep only elements that aren't the target value
        if nums[i] != val {
            nums[k] = nums[i];
            k += 1;
        }
    }
    k as i32
}
```

## Example Scenarios

**1. Common Case**
**Input:** `nums = [3,2,2,3]`, `val = 3`
A fast pointer scans every element; a slow pointer writes only the keepers. Elements equal to `val` are skipped, compacting the survivors to the front. **Output: k = 2, nums = [2,2,_,_]**

**2. Slightly Complex**
**Input:** `nums = [0,1,2,2,3,0,4,2]`, `val = 2`
The value to remove appears in scattered positions; each kept element is written at slow and slow advances, leaving the removed ones behind. **Output: k = 5, nums = [0,1,3,0,4,...]**

**3. Edge Case: Time Factor**
**Input:** `nums = [1,2,3]`, `val = 5`
No element matches, so every element is kept and written back to itself. Still a full $O(n)$ scan with slow tracking fast exactly. **Output: k = 3, nums = [1,2,3]**

**4. Edge Case: Space Factor**
**Input:** `nums = [4,4,4,4]`, `val = 4`
Every element matches and is dropped; slow never advances, so $k = 0$. All in place with $O(1)$ extra space. **Output: k = 0, nums = []**

**5. Almost-Impossible but Plausible**
**Input:** `nums = []`, `val = 0`
An empty array has nothing to scan; the loop body never runs and $k = 0$ is returned immediately. **Output: k = 0, nums = []**
