# LeetCode #27: Remove Element

https://leetcode.com/problems/remove-element/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Two Pointers** ★ | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Scan for each occurrence of val and left-shift all subsequent elements to fill the gap. Each removal costs $O(n)$ for the shift, giving $O(n^2)$ total for up to n removals.

### Optimal: Two Pointers ★
Write pointer `k` starts at 0. Scan left-to-right; copy `nums[i]` to `nums[k]` only when `nums[i] != val`, then advance `k`. One pass, no shifting.

**Why this is better than Brute Force:** Replaces $O(n)$ per-removal shifting with a single $O(1)$ write, dropping total time from $O(n^2)$ to $O(n)$.

**Constraints:**
* 0 ≤ nums.length ≤ 100
* 0 ≤ nums[i] ≤ 50
* 0 ≤ val ≤ 100

## Solutions

### C#

```
public class Solution {
    public int RemoveElement(int[] nums, int val) {
        int k = 0;
        for (int i = 0; i < nums.Length; i++) {
            // Keep only elements that aren't the target value
            if (nums[i] != val)
                nums[k++] = nums[i];
        }
        return k;
    }
}
```

### Python

```
class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        k = 0
        for i in range(len(nums)):
            # Keep only elements that aren't the target value
            if nums[i] != val:
                nums[k] = nums[i]
                k += 1
        return k
```

### Rust

```
pub fn remove_element(nums: &mut Vec<i32>, val: i32) -> i32 {
    let mut k = 0usize;
    for i in 0..nums.len() {
        // Keep only elements that aren't the target value
        if nums[i] != val {
            nums[k] = nums[i];
            k += 1;
        }
    }
    k as i32
}
```

## Example Scenarios

**1. Common Case**
**Input:** nums = [3, 2, 2, 3], val = 3
i=0: 3=val → skip. i=1: 2≠3 → nums[0]=2, k=1. i=2: 2≠3 → nums[1]=2, k=2. i=3: 3=val → skip. Array prefix: [2, 2]. Return **2**.

**2. Slightly Complex**
**Input:** nums = [0, 1, 2, 2, 3, 0, 4, 2], val = 2
Three 2s (indices 2, 3, 7) are skipped; 0, 1, 3, 0, 4 are written in order. Return **5** with prefix [0, 1, 3, 0, 4].

**3. Edge Case: Time Factor — val dominates, single survivor at the end**
**Input:** nums = [2, 2, …, 2, 1]  (n−1 twos then a 1), val = 2
Loop visits all n elements; only the last triggers a write (nums[0]=1, k→1). $O(n)$ reads, $O(1)$ writes — Brute Force would shift n−1 elements n−1 times for $O(n^2)$.

**4. Edge Case: Space Factor — val absent, no elements removed**
**Input:** nums = [1, 2, 3, 4, 5], val = 9
Every element passes the filter and is written back to its own index (self-assignment). k reaches 5. Return **5** — array unchanged, $O(1)$ extra space used at all times.

**5. Almost-Impossible but Plausible — empty array**
**Input:** nums = [], val = 0
Loop range 0..0 is empty; k stays 0. Return **0** — no iteration, no bounds access, correct immediately.
