# LeetCode #58: Length of Last Word

https://leetcode.com/problems/length-of-last-word/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Split-then-last** | $O(n)$ | $O(n)$ |
| **Reverse Scan ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Split-then-last
Split the string on whitespace to get an array of words and return the length of the last element. Concise, but allocates an $O(n)$ array to hold every word — unnecessary work when only the tail matters.

### Reverse Scan ★
Walk backward from the end of the string. First, skip any trailing spaces (there may be none). Then count consecutive non-space characters until another space or the beginning of the string is reached. The counter at that point is the answer.

**Why this is better:** No extra allocation. A single pass over at most $n$ characters suffices, and only two integer variables are used regardless of input size.

**Constraints:**
* $1 \leq$ `s.length` $\leq 10^4$
* `s` consists of English letters and spaces `' '`.
* There is at least one word in `s`.


## Solutions

### C#

```
public class Solution {
    public int LengthOfLastWord(string s) {
        int i = s.Length - 1;
        // Skip trailing spaces
        while (i >= 0 && s[i] == ' ') i--;
        // Count the last word
        int count = 0;
        while (i >= 0 && s[i] != ' ') {
            count++;
            i--;
        }
        return count;
    }
}
```

### Python

```
class Solution:
    def length_of_last_word(self, s: str) -> int:
        i = len(s) - 1
        # Skip trailing spaces
        while i >= 0 and s[i] == ' ':
            i -= 1
        # Count the last word
        count = 0
        while i >= 0 and s[i] != ' ':
            count += 1
            i -= 1
        return count
```

### Go

```
func lengthOfLastWord(s string) int {
    i := len(s) - 1
    // Skip trailing spaces
    for i >= 0 && s[i] == ' ' {
        i--
    }
    // Count the last word
    count := 0
    for i >= 0 && s[i] != ' ' {
        count++
        i--
    }
    return count
}
```

### Rust

```
impl Solution {
    pub fn length_of_last_word(s: String) -> i32 {
        let bytes = s.as_bytes();
        let mut i = bytes.len() as i32 - 1;
        // Skip trailing spaces
        while i >= 0 && bytes[i as usize] == b' ' {
            i -= 1;
        }
        // Count the last word
        let mut count = 0i32;
        while i >= 0 && bytes[i as usize] != b' ' {
            count += 1;
            i -= 1;
        }
        count
    }
}
```

## Example Scenarios

### 1. Common Case – Two Words
**Input:** `"Hello World"`
Scan backward from index 10 ('d'). No trailing spaces, so counting begins immediately: 'd','l','r','o','W' = 5 chars, then a space halts the loop. Returns **5**.

### 2. Trailing Spaces – Must Skip First
**Input:** `"   fly me   to   the moon   "`
Scanning from the end hits three trailing spaces first; the pointer skips them. Then 'n','o','o','m' are counted before hitting the space before "moon". Returns **4**.

### 3. Single Word – Entire String
**Input:** `"luffy"`
No spaces anywhere. The skip-spaces loop exits immediately; the count-word loop runs the full length of the string. Returns **5**.

### 4. Single Character Word
**Input:** `"a "`
One trailing space is skipped, then 'a' is counted and the start of the string is reached. Returns **1**.

### 5. Multiple Interior Spaces – Short Last Word
**Input:** `"day  by  day"`
No trailing spaces. Counting starts at the last 'd': 'y','a','d' = 3 chars, then the space between "by" and "day" halts counting. The double interior spaces between "day" and "by" are never visited. Returns **3**.
