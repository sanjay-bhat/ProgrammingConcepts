# LeetCode #70: Climbing Stairs

https://leetcode.com/problems/climbing-stairs/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^n)$ | $O(n)$ |
| **Optimal: Space-Optimized DP ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Recursively try taking 1 or 2 steps at each position without memoization. This recomputes the same sub-problems exponentially — $O(2^n)$ calls for $n$ steps.

### Optimal: Space-Optimized DP ★
The number of ways to reach step $n$ equals ways to reach step $n-1$ (take 1 step) plus ways to reach step $n-2$ (take 2 steps). This is the Fibonacci recurrence. By tracking only the previous two values we reduce space to $O(1)$.

**Constraints:**
* $1 \leq n \leq 45$


## Solutions

### C#

```
public class Solution {
    public int ClimbStairs(int n) {
        if (n <= 2) return n;

        // prev2 = ways to reach (i-2), prev1 = ways to reach (i-1)
        int prev2 = 1, prev1 = 2;

        for (int i = 3; i <= n; i++) {
            // Ways to reach step i = reach from one step below + two steps below
            int curr = prev1 + prev2;
            prev2 = prev1;
            prev1 = curr;
        }

        return prev1;
    }
}
```

### Python

```
class Solution:
    def climbStairs(self, n: int) -> int:
        if n <= 2:
            return n

        # prev2 = ways to reach (i-2), prev1 = ways to reach (i-1)
        prev2, prev1 = 1, 2

        for _ in range(3, n + 1):
            # Ways to reach step i = reach from one step below + two steps below
            prev2, prev1 = prev1, prev1 + prev2

        return prev1
```

### Go

```
func climbStairs(n int) int {
    if n <= 2 {
        return n
    }

    // prev2 = ways to reach (i-2), prev1 = ways to reach (i-1)
    prev2, prev1 := 1, 2

    for i := 3; i <= n; i++ {
        // Ways to reach step i = reach from one step below + two steps below
        prev2, prev1 = prev1, prev1+prev2
    }

    return prev1
}
```

### Rust

```
impl Solution {
    pub fn climb_stairs(n: i32) -> i32 {
        if n <= 2 { return n; }

        // prev2 = ways to reach (i-2), prev1 = ways to reach (i-1)
        let (mut prev2, mut prev1) = (1i32, 2i32);

        for _ in 3..=n {
            // Ways to reach step i = reach from one step below + two steps below
            let curr = prev1 + prev2;
            prev2 = prev1;
            prev1 = curr;
        }

        prev1
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n = 5`
The Fibonacci sequence gives $1, 2, 3, 5, 8$, so there are 8 distinct ways to climb 5 stairs. `prev1` advances through each value in one linear pass.

### 2. Slightly Complex
**Input:** `n = 10`
The answer is 89 (the 10th Fibonacci number, 1-indexed). The loop runs only 8 iterations, demonstrating $O(n)$ time with $O(1)$ space.

### 3. Edge Case: Time Factor
**Input:** `n = 45` (maximum constraint)
The answer is $1{,}836{,}311{,}903$, which fits inside a 32-bit signed integer. The loop runs 43 iterations — the longest execution path.

### 4. Edge Case: Space Factor
**Input:** `n = 1`
Only one way: take a single step. The base case returns immediately, using $O(1)$ space with no loop at all.

### 5. Almost-Impossible but Plausible
**Input:** `n = 44`
The answer is $1{,}134{,}903{,}170 \approx 1.13 \times 10^9$, just below the int32 ceiling of $\approx 2.1 \times 10^9$. One step further ($n=45$) nearly doubles it, still safely within bounds.
