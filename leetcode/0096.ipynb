# LeetCode #96: Unique Binary Search Trees

https://leetcode.com/problems/unique-binary-search-trees/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(4^n / n^{3/2})$ | $O(n)$ |
| **Optimal: Catalan Number DP ★** | $O(n^2)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Recursively enumerate all structurally unique BSTs by trying every node as root and counting left and right sub-tree combinations. Exponential overlap makes this infeasible for large $n$.

### Optimal: Catalan Number DP ★
Build `dp[i]` = number of unique BSTs with $i$ nodes. For each size $i$, sum over every possible root $k$: left subtree has $k-1$ nodes and right has $i-k$ nodes, giving $dp[i] = \sum_{k=1}^{i} dp[k-1] \cdot dp[i-k]$. The answer is the $(n)$-th Catalan number, computed bottom-up in $O(n^2)$.

**Constraints:**
* $1 \leq n \leq 19$


## Solutions

### C#

```
public class Solution {
    public int NumTrees(int n) {
        // dp[i] = number of unique BSTs that can be formed with i nodes
        int[] dp = new int[n + 1];
        dp[0] = 1; // Empty tree is one valid BST
        dp[1] = 1; // Single-node tree has exactly one shape

        for (int i = 2; i <= n; i++) {
            for (int k = 1; k <= i; k++) {
                // Root k splits: left subtree has k-1 nodes, right has i-k nodes
                dp[i] += dp[k - 1] * dp[i - k];
            }
        }

        return dp[n];
    }
}
```

### Rust

```
impl Solution {
    pub fn num_trees(n: i32) -> i32 {
        let n = n as usize;
        // dp[i] = number of unique BSTs that can be formed with i nodes
        let mut dp = vec![0i32; n + 1];
        dp[0] = 1; // Empty tree is one valid BST
        dp[1] = 1; // Single-node tree has exactly one shape

        for i in 2..=n {
            for k in 1..=i {
                // Root k splits: left subtree has k-1 nodes, right has i-k nodes
                dp[i] += dp[k - 1] * dp[i - k];
            }
        }

        dp[n]
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n = 3`
Five distinct BSTs are possible with 3 nodes: each node can be the root, and the subtree shapes vary. `dp[3] = dp[0]*dp[2] + dp[1]*dp[1] + dp[2]*dp[0] = 2+1+2 = 5`.

### 2. Slightly Complex
**Input:** `n = 5`
The 5th Catalan number is 42. The DP builds up through `dp[2]=2, dp[3]=5, dp[4]=14, dp[5]=42` with two nested loops.

### 3. Edge Case: Time Factor
**Input:** `n = 19` (maximum constraint)
The answer is $1{,}767{,}263{,}190$, within int32. The inner loop runs $\sum_{i=2}^{19} i = 189$ multiplications — the worst-case path through both loops.

### 4. Edge Case: Space Factor
**Input:** `n = 1`
Only one BST — a single root node. `dp[1] = 1` is returned immediately, having allocated an array of just 2 integers.

### 5. Almost-Impossible but Plausible
**Input:** `n = 18`
The answer is $\approx 4.77 \times 10^8$, and `n = 19` gives $\approx 1.77 \times 10^9$ — both fit in int32. At `n = 20` the Catalan number $\approx 6.56 \times 10^9$ would overflow, but the constraint caps at 19, precisely chosen to keep the result in 32-bit range.
