# LeetCode #97: Interleaving String

https://leetcode.com/problems/interleaving-string/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^{m+n})$ | $O(m+n)$ |
| **Optimal: 1D DP ★** | $O(m \times n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Recursively pick the next character from either `s1` or `s2` and check if it matches `s3`. Without memoization, the same `(i, j)` pair is revisited exponentially.

### Optimal: 1D DP ★
`dp[j]` = whether `s1[0..i-1]` and `s2[0..j-1]` together form an interleaving of `s3[0..i+j-1]`. Process `s1` row by row: `dp[j]` is true if it was true from above (used one char of `s1`) or `dp[j-1]` is true (used one char of `s2`), and the matching character of `s3` agrees. One 1D array replaces the full 2D table.

**Constraints:**
* $0 \leq m, n \leq 100$
* $m + n = s3.\text{length}$


## Solutions

### C#

```
public class Solution {
    public bool IsInterleave(string s1, string s2, string s3) {
        int m = s1.Length, n = s2.Length;
        if (m + n != s3.Length) return false;

        // dp[j] = true if s1[0..i-1] and s2[0..j-1] interleave to form s3[0..i+j-1]
        bool[] dp = new bool[n + 1];
        dp[0] = true;

        // Seed: using only s2 characters (no s1)
        for (int j = 1; j <= n; j++)
            dp[j] = dp[j - 1] && s2[j - 1] == s3[j - 1];

        for (int i = 1; i <= m; i++) {
            // Can we extend using s1[i-1]?
            dp[0] = dp[0] && s1[i - 1] == s3[i - 1];

            for (int j = 1; j <= n; j++) {
                // Extend from above (take s1[i-1]) or from the left (take s2[j-1])
                dp[j] = (dp[j] && s1[i - 1] == s3[i + j - 1])
                       || (dp[j - 1] && s2[j - 1] == s3[i + j - 1]);
            }
        }

        return dp[n];
    }
}
```

### Python

```
class Solution:
    def isInterleave(self, s1: str, s2: str, s3: str) -> bool:
        m, n = len(s1), len(s2)
        if m + n != len(s3):
            return False

        # dp[j] = True if s1[0..i-1] and s2[0..j-1] interleave to form s3[0..i+j-1]
        dp = [False] * (n + 1)
        dp[0] = True

        # Seed: using only s2 characters (no s1)
        for j in range(1, n + 1):
            dp[j] = dp[j - 1] and s2[j - 1] == s3[j - 1]

        for i in range(1, m + 1):
            # Can we extend using s1[i-1]?
            dp[0] = dp[0] and s1[i - 1] == s3[i - 1]

            for j in range(1, n + 1):
                # Extend from above (take s1[i-1]) or from the left (take s2[j-1])
                dp[j] = (dp[j] and s1[i - 1] == s3[i + j - 1]) \
                      or (dp[j - 1] and s2[j - 1] == s3[i + j - 1])

        return dp[n]
```

### Rust

```
impl Solution {
    pub fn is_interleave(s1: String, s2: String, s3: String) -> bool {
        let (m, n) = (s1.len(), s2.len());
        if m + n != s3.len() { return false; }

        let s1 = s1.as_bytes();
        let s2 = s2.as_bytes();
        let s3 = s3.as_bytes();

        // dp[j] = true if s1[0..i-1] and s2[0..j-1] interleave to form s3[0..i+j-1]
        let mut dp = vec![false; n + 1];
        dp[0] = true;

        // Seed: using only s2 characters (no s1)
        for j in 1..=n {
            dp[j] = dp[j - 1] && s2[j - 1] == s3[j - 1];
        }

        for i in 1..=m {
            // Can we extend using s1[i-1]?
            dp[0] = dp[0] && s1[i - 1] == s3[i - 1];

            for j in 1..=n {
                // Extend from above (take s1[i-1]) or from the left (take s2[j-1])
                dp[j] = (dp[j] && s1[i - 1] == s3[i + j - 1])
                     || (dp[j - 1] && s2[j - 1] == s3[i + j - 1]);
            }
        }

        dp[n]
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `s1 = "aabcc", s2 = "dbbca", s3 = "aadbbcbcac"`
A valid interleaving exists: take `a` from s1, `d` from s2, `a` from s1, `bb` from s2, `c` from s1, `ca` from s2, `c` from s1. The DP confirms `dp[5] = true`.

### 2. Slightly Complex
**Input:** `s1 = "aabcc", s2 = "dbbca", s3 = "aadbbbaccc"`
No valid interleaving — the character counts match but the ordering does not allow it. The DP propagates `false` at the conflicting position and returns `false`.

### 3. Edge Case: Time Factor
**Input:** `m = n = 100`, all characters identical
Every cell is reachable (trivially), filling all $100 \times 100 = 10{,}000$ DP cells — the worst-case time path. Result is `true`.

### 4. Edge Case: Space Factor
**Input:** `s1 = "", s2 = "abc", s3 = "abc"`
Empty `s1`: the entire interleaving must come from `s2`. The DP array has length 4 and seeds to `[T, T, T, T]` if all characters match — $O(n)$ minimal allocation with $m=0$.

### 5. Almost-Impossible but Plausible
**Input:** `s1 = "a", s2 = "b", s3 = "ba"`
The interleaving `b`(from s2) then `a`(from s1) is valid — result `true`. The DP with $m=n=1$ fills only 4 cells, showing correctness on the smallest non-trivial case.
