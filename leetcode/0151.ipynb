# LeetCode #151: Reverse Words in a String

https://leetcode.com/problems/reverse-words-in-a-string/

## Comparison of Approaches
| Approach | Time | Space |
|---|---|---|
| Split & Reverse | O(n) | O(n) |
| Two Pointers In-Place ★ | O(n) | O(1) |

## Understanding the Methods
### Split & Reverse
Split the string on whitespace, filter out empty strings, reverse the list, and join with a single space. Simple but uses extra space proportional to input length.

### Optimal: Two Pointers In-Place ★
Reverse the entire string in-place, then reverse each individual word. Finally, clean up extra spaces. This achieves O(1) extra space. In practice, most languages treat strings as immutable so we work with a char array or list of characters.

## Solutions
### C#

```
public class Solution {
    public string ReverseWords(string s) {
        var words = s.Trim().Split(' ', StringSplitOptions.RemoveEmptyEntries);
        Array.Reverse(words);
        return string.Join(' ', words);
    }
}
```

### Python

```
class Solution:
    def reverseWords(self, s: str) -> str:
        return ' '.join(s.split()[::-1])
```

### Go

```
package main

import (
    "fmt"
    "strings"
)

func reverseWords(s string) string {
    words := strings.Fields(s)
    for i, j := 0, len(words)-1; i < j; i, j = i+1, j-1 {
        words[i], words[j] = words[j], words[i]
    }
    return strings.Join(words, " ")
}

func main() {
    fmt.Println(reverseWords("the sky is blue")) // "blue is sky the"
}
```

### Rust

```
pub struct Solution;

impl Solution {
    pub fn reverse_words(s: String) -> String {
        s.split_whitespace()
            .rev()
            .collect::<Vec<&str>>()
            .join(" ")
    }
}
```

## Example Scenarios

### 1. Common Case — Clean Sentence
**Input:** `s = "the sky is blue"`

Split on spaces into `["the","sky","is","blue"]`, reverse the list, and join with single spaces. Result: `"blue is sky the"`.

### 2. Slightly Complex — Leading / Trailing Spaces
**Input:** `s = "  hello world  "`

Trimming drops the surrounding blanks; splitting yields `["hello","world"]`. Reversed and joined gives `"world hello"` with no stray spaces.

### 3. Edge Case: Time — Multiple Internal Spaces
**Input:** `s = "a good   example"`

Consecutive spaces collapse because empty tokens are discarded: `["a","good","example"]`. One $O(n)$ pass reverses to `"example good a"`.

### 4. Edge Case: Space — Heavy Padding, In-Place Goal
**Input:** `s = "  Bob    Loves  Alice   "`

Even with lots of filler whitespace the word count is 3. The two-pointer in-place method rebuilds the answer `"Alice Loves Bob"` using $O(1)$ extra space.

### 5. Almost-Impossible but Plausible — Single Word
**Input:** `s = "singleword"`

Splitting gives a one-element list; reversing a single element changes nothing, so the output is `"singleword"` unchanged.
