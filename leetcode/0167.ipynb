# LeetCode #167: Two Sum II - Input Array Is Sorted

https://leetcode.com/problems/two-sum-ii-input-array-is-sorted/

## Comparison of Approaches
| Approach | Time | Space |
|---|---|---|
| Hash Map | O(n) | O(n) |
| Two Pointers ★ | O(n) | O(1) |

## Understanding the Methods
### Hash Map
Use a hash map to find complements — same as Two Sum I. But this ignores the sorted property.

### Optimal: Two Pointers ★
With the array sorted, use left and right pointers. If sum < target, move left pointer right (increase sum). If sum > target, move right pointer left (decrease sum). If sum == target, return the 1-indexed positions. Guaranteed to find a solution per problem constraints.

## Solutions
### C#

```
public class Solution {
    public int[] TwoSum(int[] numbers, int target) {
        int left = 0, right = numbers.Length - 1;
        while (left < right) {
            int sum = numbers[left] + numbers[right];
            if (sum == target) return new[] { left + 1, right + 1 };
            if (sum < target) left++;
            else right--;
        }
        return new[] { -1, -1 }; // guaranteed to have solution
    }
}
```

### Python

```
from typing import List

class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        left, right = 0, len(numbers) - 1
        while left < right:
            s = numbers[left] + numbers[right]
            if s == target:
                return [left + 1, right + 1]
            elif s < target:
                left += 1
            else:
                right -= 1
        return []
```

### Go

```
package main

import "fmt"

func twoSum(numbers []int, target int) []int {
    left, right := 0, len(numbers)-1
    for left < right {
        s := numbers[left] + numbers[right]
        if s == target {
            return []int{left + 1, right + 1}
        } else if s < target {
            left++
        } else {
            right--
        }
    }
    return nil
}

func main() {
    fmt.Println(twoSum([]int{2, 7, 11, 15}, 9)) // [1 2]
}
```

### Rust

## Example Scenarios

### 1. Common Case — Converge in a Few Steps
**Input:** `numbers = [2,7,11,15]`, `target = 9`
Left at $2$, right at $15$: sum $17 > 9$, move right in. Now $2+11=13>9$, move right again. $2+7=9$ — return 1-indexed `[1,2]`.

### 2. Slightly Complex — Move Left Pointer Up
**Input:** `numbers = [2,3,4]`, `target = 6`
$2+4=6$ immediately matches, returning `[1,3]`. Because the array is sorted, a too-small sum would instead advance the left pointer.

### 3. Edge Case: Time Factor — Long Array, Late Match
**Input:** `numbers = [1,2,3,4,4,9,56,90]`, `target = 8`
Pointers sweep inward from both ends across eight elements; $4+4=8$ meets in the middle, giving `[4,5]`.

### 4. Edge Case: Space Factor — Negatives, O(1) Extra
**Input:** `numbers = [-1,0]`, `target = -1`
The two-pointer scan uses only two indices (no hash map). $-1+0=-1$ matches on the first check: `[1,2]`.

### 5. Almost-Impossible but Plausible — Only the Extremes Work
**Input:** `numbers = [5,25,75], target = 100`
Inner value $25$ can never pair to $100$ within range; $25+75=100$ succeeds after the left pointer steps once, returning `[2,3]`.
