# LeetCode #171: Excel Sheet Column Number

https://leetcode.com/problems/excel-sheet-column-number/

## Comparison of Approaches
| Approach | Time | Space |
|---|---|---|
| Base-26 Decode ★ | O(n) | O(1) |

## Understanding the Methods
### Brute Force
No simpler approach; this is a base-26 to decimal conversion.

### Optimal: Base-26 Decode ★
Process left to right. For each character, multiply the running result by 26 and add the column value of the current letter (A=1, B=2, ..., Z=26). This is the standard polynomial evaluation (Horner's method) for base conversion.

## Solutions
### C#

```
public class Solution {
    public int TitleToNumber(string columnTitle) {
        int result = 0;
        foreach (char c in columnTitle) {
            result = result * 26 + (c - 'A' + 1);
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def titleToNumber(self, columnTitle: str) -> int:
        result = 0
        for c in columnTitle:
            result = result * 26 + (ord(c) - ord('A') + 1)
        return result
```

### Go

```
package main

import "fmt"

func titleToNumber(columnTitle string) int {
    result := 0
    for _, c := range columnTitle {
        result = result*26 + int(c-'A'+1)
    }
    return result
}

func main() {
    fmt.Println(titleToNumber("A"))   // 1
    fmt.Println(titleToNumber("AB"))  // 28
    fmt.Println(titleToNumber("ZY"))  // 701
}
```

### Rust

```
pub struct Solution;

impl Solution {
    pub fn title_to_number(column_title: String) -> i32 {
        column_title.bytes().fold(0i32, |acc, c| {
            acc * 26 + (c - b'A' + 1) as i32
        })
    }
}
```

## Examples

### 1. Common Case — Single Letter
**Input:** `columnTitle = "A"`
$result = 0 \times 26 + 1 = 1$. One letter maps straight to its 1-based value. Output $1$.

### 2. Slightly Complex — Two Letters
**Input:** `columnTitle = "AB"`
Process left to right: $A \to 1$; then $1 \times 26 + 2 = 28$. Output $28$ — like reading a base-26 number.

### 3. Edge Case: Time Factor — Larger Column
**Input:** `columnTitle = "ZY"`
$Z \to 26$; then $26 \times 26 + 25 = 701$. Still $O(\text{len})$ work. Output $701$.

### 4. Edge Case: Space Factor — Single Max Letter, O(1)
**Input:** `columnTitle = "Z"`
Only a running integer accumulator is needed. $Z$ is the largest single digit: $26$.

### 5. Almost-Impossible but Plausible — INT_MAX
**Input:** `columnTitle = "FXSHRXW"`
Seven letters accumulate to $2147483647$, exactly $2^{31}-1$, so 32-bit arithmetic sits right at the limit.
