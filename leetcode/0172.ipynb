# LeetCode #172: Factorial Trailing Zeroes

https://leetcode.com/problems/factorial-trailing-zeroes/

## Comparison of Approaches
| Approach | Time | Space |
|---|---|---|
| Compute factorial | O(n) or worse | O(n) |
| Count factor 5s ★ | O(log₅ n) | O(1) |

## Understanding the Methods
### Brute Force
Compute n! and count trailing zeros. Impractical for large n due to huge numbers.

### Optimal: Count factor 5s ★
Trailing zeros come from factors of 10 = 2 × 5. Since 2s always outnumber 5s in a factorial, count factors of 5. Every multiple of 5 contributes one 5; every multiple of 25 contributes an extra 5, etc. Sum: n/5 + n/25 + n/125 + ... until the term is 0.

## Solutions
### C#

```
public class Solution {
    public int TrailingZeroes(int n) {
        int count = 0;
        while (n >= 5) {
            n /= 5;
            count += n;
        }
        return count;
    }
}
```

### Python

```
class Solution:
    def trailingZeroes(self, n: int) -> int:
        count = 0
        while n >= 5:
            n //= 5
            count += n
        return count
```

### Go

```
package main

import "fmt"

func trailingZeroes(n int) int {
    count := 0
    for n >= 5 {
        n /= 5
        count += n
    }
    return count
}

func main() {
    fmt.Println(trailingZeroes(3))   // 0
    fmt.Println(trailingZeroes(5))   // 1
    fmt.Println(trailingZeroes(25))  // 6
}
```

### Rust

```
pub struct Solution;

impl Solution {
    pub fn trailing_zeroes(mut n: i32) -> i32 {
        let mut count = 0;
        while n >= 5 {
            n /= 5;
            count += n;
        }
        count
    }
}
```

## Examples

### 1. Common Case — One Factor of Five
**Input:** `n = 5`
Trailing zeros come from pairs of $2$ and $5$; fives are scarcer, so just count them. $\lfloor 5/5 \rfloor = 1$. Output $1$.

### 2. Slightly Complex — Two Multiples of Five
**Input:** `n = 10`
$\lfloor 10/5 \rfloor = 2$ (from $5$ and $10$); $\lfloor 10/25 \rfloor = 0$. Total $2$.

### 3. Edge Case: Time Factor — Powers of Five Count Twice
**Input:** `n = 25`
$\lfloor 25/5 \rfloor = 5$ plus $\lfloor 25/25 \rfloor = 1$ (since $25 = 5^2$ contributes an extra five). Total $6$.

### 4. Edge Case: Space Factor — No Fives at All, O(1)
**Input:** `n = 4`
$4! = 24$ has no trailing zero: $\lfloor 4/5 \rfloor = 0$, loop stops immediately. Only a counter is stored. Output $0$.

### 5. Almost-Impossible but Plausible — Large n, Many Terms
**Input:** `n = 10000`
Sum $\lfloor 10000/5 \rfloor + \lfloor /25 \rfloor + \lfloor /125 \rfloor + \lfloor /625 \rfloor + \lfloor /3125 \rfloor = 2000+400+80+16+3 = 2499$.
