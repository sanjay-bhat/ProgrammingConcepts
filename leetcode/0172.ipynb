# LeetCode #172: Factorial Trailing Zeroes

https://leetcode.com/problems/factorial-trailing-zeroes/

## Comparison of Approaches
| Approach | Time | Space |
|---|---|---|
| Compute factorial | O(n) or worse | O(n) |
| Count factor 5s ★ | O(log₅ n) | O(1) |

## Understanding the Methods
### Brute Force
Compute n! and count trailing zeros. Impractical for large n due to huge numbers.

### Optimal: Count factor 5s ★
Trailing zeros come from factors of 10 = 2 × 5. Since 2s always outnumber 5s in a factorial, count factors of 5. Every multiple of 5 contributes one 5; every multiple of 25 contributes an extra 5, etc. Sum: n/5 + n/25 + n/125 + ... until the term is 0.

## Solutions
### C#

```
public class Solution {
    public int TrailingZeroes(int n) {
        int count = 0;
        while (n >= 5) {
            n /= 5;
            count += n;
        }
        return count;
    }
}
```

### Python

```
class Solution:
    def trailingZeroes(self, n: int) -> int:
        count = 0
        while n >= 5:
            n //= 5
            count += n
        return count
```

### Go

```
package main

import "fmt"

func trailingZeroes(n int) int {
    count := 0
    for n >= 5 {
        n /= 5
        count += n
    }
    return count
}

func main() {
    fmt.Println(trailingZeroes(3))   // 0
    fmt.Println(trailingZeroes(5))   // 1
    fmt.Println(trailingZeroes(25))  // 6
}
```

### Rust

```
pub struct Solution;

impl Solution {
    pub fn trailing_zeroes(mut n: i32) -> i32 {
        let mut count = 0;
        while n >= 5 {
            n /= 5;
            count += n;
        }
        count
    }
}
```

## Examples

**Common:** `n=5` → `1` (5! = 120, one trailing zero)

**Slightly Complex:** `n=10` → `2` (10! = 3628800, two trailing zeros)

**Edge Time:** `n=25` → `6` (25/5=5, 5/5=1, total=5+1=6)

**Edge Space:** `n=0` → `0` (0! = 1, no trailing zeros)

**Almost-Impossible:** `n=125` → `31` (125/5=25, 25/5=5, 5/5=1, total=25+5+1=31)
