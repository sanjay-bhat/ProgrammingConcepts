# 231. Power of Two
https://leetcode.com/problems/power-of-two/

## Complexity Comparison

| Approach | Time | Space | Notes |
|----------|------|-------|-------|
| Bit manipulation | $O(1)$ | $O(1)$ | Optimal — `n & (n-1) == 0` |
| Loop / divide by 2 | $O(\log n)$ | $O(1)$ | Repeatedly divide |

## Methodology

A power of two in binary has exactly one bit set (e.g., 1, 10, 100, 1000). The expression `n & (n-1)` clears the lowest set bit. If the result is 0, then n had exactly one set bit, meaning it is a power of two. We also check n > 0 since 0 and negatives are not powers of two.

## Solutions

### C#

```
public class Solution {
    public bool IsPowerOfTwo(int n) {
        return n > 0 && (n & (n - 1)) == 0;
    }
}
```

### Python

```
class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        return n > 0 and (n & (n - 1)) == 0
```

### Rust

## Example Scenarios

1. **Common Case** — A familiar power of two  
   **Input:** `n = 16`  
   $16 = 2^4 = \texttt{10000}_2$. Compute $n \& (n-1) = \texttt{10000} \& \texttt{01111} = 0$. Since $n > 0$ and the result is $0$, return `true`.

2. **Slightly Complex** — A non-power that is close to one  
   **Input:** `n = 6`  
   $6 = \texttt{110}_2$. Compute $n \& (n-1) = \texttt{110} \& \texttt{101} = \texttt{100} = 4 \ne 0$. Two bits are set, so return `false`. Catches the misconception that "even = power of two."

3. **Edge Case: Time Factor** — The largest 32-bit power of two  
   **Input:** `n = 1073741824`  
   $1073741824 = 2^{30}$, the largest power of two fitting in a signed 32-bit integer. $n \& (n-1) = 0$ in $O(1)$. No loop or log needed — the bit trick works in constant time regardless of magnitude.

4. **Edge Case: Space Factor** — Zero input  
   **Input:** `n = 0`  
   $0 = \texttt{00\ldots0}_2$ — no bits set at all. The first check $n > 0$ fails immediately, returning `false`. Zero is not a power of two despite $0 \& (0-1) = 0$. The guard clause prevents this false positive.

5. **Almost-Impossible but Plausible** — Maximum negative value  
   **Input:** `n = -2147483648`  
   $-2147483648 = -2^{31}$ in two's complement: $\texttt{10000000\ldots0}_2$. Technically has one bit set, so $n \& (n-1) = 0$. But $n > 0$ is false, so we correctly return `false`. Negative numbers are never powers of two.
