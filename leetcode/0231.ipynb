# 231. Power of Two
https://leetcode.com/problems/power-of-two/

## Complexity Comparison

| Approach | Time | Space | Notes |
|----------|------|-------|-------|
| Bit manipulation | $O(1)$ | $O(1)$ | Optimal — `n & (n-1) == 0` |
| Loop / divide by 2 | $O(\log n)$ | $O(1)$ | Repeatedly divide |

## Methodology

A power of two in binary has exactly one bit set (e.g., 1, 10, 100, 1000). The expression `n & (n-1)` clears the lowest set bit. If the result is 0, then n had exactly one set bit, meaning it is a power of two. We also check n > 0 since 0 and negatives are not powers of two.

## Solutions

### C#

```
public class Solution {
    public bool IsPowerOfTwo(int n) {
        return n > 0 && (n & (n - 1)) == 0;
    }
}
```

### Python

```
class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        return n > 0 and (n & (n - 1)) == 0
```

### Go

```
func isPowerOfTwo(n int) bool {
    return n > 0 && n&(n-1) == 0
}
```

### Rust

```
impl Solution {
    pub fn is_power_of_two(n: i32) -> bool {
        n > 0 && (n & (n - 1)) == 0
    }
}
```

## Example Scenarios

1. **Common Case** — A familiar power of two  
   **Input:** `n = 16`  
   $16 = 2^4 = \texttt{10000}_2$. Compute $n \mathbin{\&} (n-1) = \texttt{10000} \mathbin{\&} \texttt{01111} = 0$. Since $n > 0$ and the result is $0$, return `true`.

2. **Slightly Complex** — Large power of two  
   **Input:** `n = 1048576`  
   $1048576 = 2^{20}$. In binary this is a single $1$ followed by twenty $0$s. $n \mathbin{\&} (n-1)$ clears that lone bit to zero. Return `true`.

3. **Edge Case: Time Factor** — Not a power, but close  
   **Input:** `n = 6`  
   $6 = \texttt{110}_2$. $n \mathbin{\&} (n-1) = \texttt{110} \mathbin{\&} \texttt{101} = \texttt{100} \neq 0$. Multiple set bits means it is not a power of two. Return `false`. The bit trick catches this in $O(1)$.

4. **Edge Case: Space Factor** — Zero and negative  
   **Input:** `n = 0` then `n = -2147483648`  
   $n = 0$ fails the $n > 0$ guard immediately. $n = -2^{31}$ (\texttt{INT\_MIN}) also fails — it has one set bit in two's complement but is negative, so it is not a power of two.

5. **Almost-Impossible but Plausible** — Maximum 32-bit power  
   **Input:** `n = 2147483647` ($2^{31} - 1$, \texttt{INT\_MAX})  
   $2^{31}-1 = \texttt{0111\ldots1}_2$ (31 set bits). $n \mathbin{\&} (n-1) = \texttt{0111\ldots0} \neq 0$. Despite being the largest positive 32-bit integer, it is not a power of two. The largest valid power is $2^{30} = 1073741824$.
