# 231. Power of Two
https://leetcode.com/problems/power-of-two/

## Complexity Comparison

| Approach | Time | Space | Notes |
|----------|------|-------|-------|
| Bit manipulation | $O(1)$ | $O(1)$ | Optimal — `n & (n-1) == 0` |
| Loop / divide by 2 | $O(\log n)$ | $O(1)$ | Repeatedly divide |

## Methodology

A power of two in binary has exactly one bit set (e.g., 1, 10, 100, 1000). The expression `n & (n-1)` clears the lowest set bit. If the result is 0, then n had exactly one set bit, meaning it is a power of two. We also check n > 0 since 0 and negatives are not powers of two.

## Solutions

### C#

```
public class Solution {
    public bool IsPowerOfTwo(int n) {
        return n > 0 && (n & (n - 1)) == 0;
    }
}
```

### Python

```
class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        return n > 0 and (n & (n - 1)) == 0
```

## Example Scenarios

### 1. Common Case — A familiar power of two
**Input:** `n = 16`  
$16 = 2^4 = 10000_2$ has exactly one set bit. Compute $n \mathbin{\&} (n-1) = 10000_2 \mathbin{\&} 01111_2 = 0$. Since $n > 0$ and the AND is $0$, return `true`.

### 2. Slightly Complex — Larger power of two
**Input:** `n = 64`  
$64 = 2^6 = 1000000_2$. Subtracting one flips the lone high bit and sets every lower bit to give $0111111_2$, so $n \mathbin{\&} (n-1) = 0$. Return `true`.

### 3. Edge Case: Time Factor — A near miss
**Input:** `n = 6`  
$6 = 110_2$ has two set bits. $n \mathbin{\&} (n-1) = 110_2 \mathbin{\&} 101_2 = 100_2 = 4 \neq 0$, so $n$ is not a power of two — return `false`. The test is $O(1)$; no divide-by-two loop is ever needed.

### 4. Edge Case: Space Factor — The smallest power
**Input:** `n = 1`  
$1 = 2^0 = 1_2$. Here $n - 1 = 0$, so $n \mathbin{\&} (n-1) = 0$ using only a single machine word of scratch space. Return `true`.

### 5. Almost-Impossible but Plausible — One bit, but negative
**Input:** `n = -2147483648`  
As a 32-bit pattern this is $1000\ldots0_2$ with only the sign bit set, so $n \mathbin{\&} (n-1) = 0$ would pass the bit test alone. But the guard $n > 0$ fails, so return `false` — the sign check is what rescues the trick.
