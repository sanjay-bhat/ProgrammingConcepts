# 258. Add Digits
https://leetcode.com/problems/add-digits/

## Complexity Comparison

| Approach | Time | Space | Description |
|----------|------|-------|-------------|
| Simulation | O(log n) | O(1) | Repeatedly sum digits until single digit |
| **Digital Root (Optimal)** | **O(1)** | **O(1)** | Mathematical formula using modulo 9 |

## Methodology

The digital root of a number can be computed with a formula: for any positive number, the digital root is `1 + (num - 1) % 9`. This is because the digital root cycles through 1-9 for positive integers. For 0, the result is 0.

## Solutions

### C#

```
public class Solution {
    public int AddDigits(int num) {
        if (num == 0) return 0;
        return 1 + (num - 1) % 9;
    }
}
```

### Python

```
class Solution:
    def addDigits(self, num: int) -> int:
        if num == 0:
            return 0
        return 1 + (num - 1) % 9
```

### Go

```
func addDigits(num int) int {
    if num == 0 {
        return 0
    }
    return 1 + (num - 1) % 9
}
```

### Rust

```
impl Solution {
    pub fn add_digits(num: i32) -> i32 {
        if num == 0 { return 0; }
        1 + (num - 1) % 9
    }
}
```

## Example Scenarios

1. **Common Case** — `num = 38`
   **Input:** A two-digit number.
   $3 + 8 = 11$, then $1 + 1 = 2$. Using the digital root formula: $1 + (38-1) \bmod 9 = 1 + 1 = 2$.

2. **Slightly Complex** — `num = 9999`
   **Input:** A four-digit number with all 9s.
   $9+9+9+9=36$, then $3+6=9$. The digital root of any multiple of 9 (except 0) is always $9$.

3. **Edge Case: Time Factor** — `num = 2,147,483,647` ($2^{31}-1$, maximum 32-bit integer)
   **Input:** The largest possible input.
   The $O(1)$ formula computes $1 + (2{,}147{,}483{,}646 \bmod 9) = 1 + 0 = 1$ instantly — no digit-by-digit iteration needed.

4. **Edge Case: Space Factor** — `num = 0`
   **Input:** The smallest valid input (boundary).
   Zero is the special case: digital root is $0$. The formula handles it with the initial `if num == 0` check. No computation, $O(1)$ space.

5. **Almost-Impossible but Plausible** — `num = 999,999,999`
   **Input:** Nine 9s — the largest 9-digit number.
   Simulation would sum digits three times ($81 \to 9$). But the $O(1)$ formula gives $1 + (999{,}999{,}998 \bmod 9) = 1 + 8 = 9$ in one step — demonstrating why the math shortcut matters.
