# 258. Add Digits
https://leetcode.com/problems/add-digits/

## Complexity Comparison

| Approach | Time | Space | Description |
|----------|------|-------|-------------|
| Simulation | O(log n) | O(1) | Repeatedly sum digits until single digit |
| **Digital Root (Optimal)** | **O(1)** | **O(1)** | Mathematical formula using modulo 9 |

## Methodology

The digital root of a number can be computed with a formula: for any positive number, the digital root is `1 + (num - 1) % 9`. This is because the digital root cycles through 1-9 for positive integers. For 0, the result is 0.

## Solutions

### C#

```
public class Solution {
    public int AddDigits(int num) {
        if (num == 0) return 0;
        return 1 + (num - 1) % 9;
    }
}
```

### Python

```
class Solution:
    def addDigits(self, num: int) -> int:
        if num == 0:
            return 0
        return 1 + (num - 1) % 9
```

### Go

```
func addDigits(num int) int {
    if num == 0 {
        return 0
    }
    return 1 + (num - 1) % 9
}
```

### Rust

```
impl Solution {
    pub fn add_digits(num: i32) -> i32 {
        if num == 0 { return 0; }
        1 + (num - 1) % 9
    }
}
```

## Example Scenarios

### 1. Common Case — `num = 38`
**Input:** A two-digit number.
$3 + 8 = 11$, then $1 + 1 = 2$. Closed form: $1 + (38-1) \bmod 9 = 1 + 1 = 2$. Output: `2`.

### 2. Slightly Complex — `num = 12345`
**Input:** A five-digit number.
$1+2+3+4+5 = 15$, then $1+5 = 6$. Closed form: $1 + (12345-1)\bmod 9 = 6$. Output: `6`.

### 3. Edge Case: Time Factor — `num = 999999`
**Input:** All nines.
$9 \times 6 = 54 \to 5+4 = 9$. Since $num \bmod 9 = 0$ and $num \ne 0$, the digital root is $9$. The $O(1)$ formula avoids the loop entirely.

### 4. Edge Case: Space Factor — `num = 0`
**Input:** Zero.
Already a single digit, so the answer is `0`. This is the special case the formula $1+(num-1)\bmod 9$ must guard (it would otherwise give $9$).

### 5. Almost-Impossible but Plausible — `num = 2147483647` ($2^{31}-1$)
**Input:** The maximum 32-bit signed integer.
Digits sum to $46 \to 10 \to 1$. Closed form: $1 + (2147483647-1)\bmod 9 = 1$. Output: `1`.
