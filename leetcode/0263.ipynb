# 263. Ugly Number
https://leetcode.com/problems/ugly-number/

## Complexity Comparison

| Approach | Time | Space | Description |
|----------|------|-------|-------------|
| **Divide by factors (Optimal)** | **O(log n)** | **O(1)** | Repeatedly divide by 2, 3, 5 |

## Methodology

An ugly number has only prime factors 2, 3, and 5. We repeatedly divide the number by 2, 3, and 5 as long as it is divisible. If the result is 1, it is an ugly number. Numbers less than or equal to 0 are not ugly.

## Solutions

### C#

```
public class Solution {
    public bool IsUgly(int n) {
        if (n <= 0) return false;
        foreach (int factor in new[] { 2, 3, 5 }) {
            while (n % factor == 0) n /= factor;
        }
        return n == 1;
    }
}
```

### Python

```
class Solution:
    def isUgly(self, n: int) -> bool:
        if n <= 0:
            return False
        for factor in (2, 3, 5):
            while n % factor == 0:
                n //= factor
        return n == 1
```

### Go

```
func isUgly(n int) bool {
    if n <= 0 {
        return false
    }
    for _, factor := range []int{2, 3, 5} {
        for n%factor == 0 {
            n /= factor
        }
    }
    return n == 1
}
```

### Rust

```
impl Solution {
    pub fn is_ugly(mut n: i32) -> bool {
        if n <= 0 { return false; }
        for factor in [2, 3, 5] {
            while n % factor == 0 {
                n /= factor;
            }
        }
        n == 1
    }
}
```

## Example Scenarios

1. **Common Case** — `n = 6`
   **Input:** A small positive integer.
   $6 = 2 \times 3$. Divide by 2: $6 \to 3$. Divide by 3: $3 \to 1$. Result is 1, so it is ugly.

2. **Slightly Complex** — `n = 14`
   **Input:** A number with a prime factor other than 2, 3, 5.
   $14 = 2 \times 7$. Divide by 2: $14 \to 7$. Now 7 is not divisible by 2, 3, or 5, and $7 \neq 1$. Not ugly.

3. **Edge Case: Time Factor** — `n = 2^{30} = 1,073,741,824`
   **Input:** The largest power of 2 within 32-bit range.
   Divide by 2 thirty times: $2^{30} \to 2^{29} \to \cdots \to 1$. That is $O(\log n) = 30$ divisions — the most iterations the algorithm performs.

4. **Edge Case: Space Factor** — `n = 1`
   **Input:** The smallest ugly number (boundary).
   $1$ has no prime factors at all. The loop body never executes; we immediately return `True`. $O(1)$ time and space.

5. **Almost-Impossible but Plausible** — `n = -2,147,483,648` ($-2^{31}$, minimum 32-bit integer)
   **Input:** The most negative 32-bit value.
   The guard `n <= 0` returns `False` immediately. Negative numbers and zero are never ugly by definition.
