# 263. Ugly Number
https://leetcode.com/problems/ugly-number/

## Complexity Comparison

| Approach | Time | Space | Description |
|----------|------|-------|-------------|
| **Divide by factors (Optimal)** | **O(log n)** | **O(1)** | Repeatedly divide by 2, 3, 5 |

## Methodology

An ugly number has only prime factors 2, 3, and 5. We repeatedly divide the number by 2, 3, and 5 as long as it is divisible. If the result is 1, it is an ugly number. Numbers less than or equal to 0 are not ugly.

## Solutions

### C#

```
public class Solution {
    public bool IsUgly(int n) {
        if (n <= 0) return false;
        foreach (int factor in new[] { 2, 3, 5 }) {
            while (n % factor == 0) n /= factor;
        }
        return n == 1;
    }
}
```

### Python

```
class Solution:
    def isUgly(self, n: int) -> bool:
        if n <= 0:
            return False
        for factor in (2, 3, 5):
            while n % factor == 0:
                n //= factor
        return n == 1
```

### Go

```
func isUgly(n int) bool {
    if n <= 0 {
        return false
    }
    for _, factor := range []int{2, 3, 5} {
        for n%factor == 0 {
            n /= factor
        }
    }
    return n == 1
}
```

### Rust

```
impl Solution {
    pub fn is_ugly(mut n: i32) -> bool {
        if n <= 0 { return false; }
        for factor in [2, 3, 5] {
            while n % factor == 0 {
                n /= factor;
            }
        }
        n == 1
    }
}
```

## Example Scenarios

1. **`n = 6`** - 6 = 2 * 3, only factors 2 and 3. Returns `true`.

2. **`n = 1`** - 1 has no prime factors. Returns `true` (by convention).

3. **`n = 14`** - 14 = 2 * 7, has prime factor 7. Returns `false`.

4. **`n = 0`** - Not positive. Returns `false`.

5. **`n = -6`** - Negative numbers are not ugly. Returns `false`.
