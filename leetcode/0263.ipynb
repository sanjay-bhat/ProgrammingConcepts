# 263. Ugly Number
https://leetcode.com/problems/ugly-number/

## Complexity Comparison

| Approach | Time | Space | Description |
|----------|------|-------|-------------|
| **Divide by factors (Optimal)** | **O(log n)** | **O(1)** | Repeatedly divide by 2, 3, 5 |

## Methodology

An ugly number has only prime factors 2, 3, and 5. We repeatedly divide the number by 2, 3, and 5 as long as it is divisible. If the result is 1, it is an ugly number. Numbers less than or equal to 0 are not ugly.

## Solutions

### C#

```
public class Solution {
    public bool IsUgly(int n) {
        if (n <= 0) return false;
        foreach (int factor in new[] { 2, 3, 5 }) {
            while (n % factor == 0) n /= factor;
        }
        return n == 1;
    }
}
```

### Python

```
class Solution:
    def isUgly(self, n: int) -> bool:
        if n <= 0:
            return False
        for factor in (2, 3, 5):
            while n % factor == 0:
                n //= factor
        return n == 1
```

### Go

```
func isUgly(n int) bool {
    if n <= 0 {
        return false
    }
    for _, factor := range []int{2, 3, 5} {
        for n%factor == 0 {
            n /= factor
        }
    }
    return n == 1
}
```

### Rust

```
impl Solution {
    pub fn is_ugly(mut n: i32) -> bool {
        if n <= 0 { return false; }
        for factor in [2, 3, 5] {
            while n % factor == 0 {
                n /= factor;
            }
        }
        n == 1
    }
}
```

## Example Scenarios

### 1. Common Case — `n = 6`
**Input:** A small composite.
$6 = 2 \times 3$. Divide by 2 to get 3, then by 3 to get 1. Only factors in $\{2,3,5\}$ remain, so it is ugly. Output: `true`.

### 2. Slightly Complex — `n = 14`
**Input:** A number with a foreign prime.
$14 = 2 \times 7$. After dividing by 2 we get 7, which 2, 3, 5 cannot reduce. The value stalls at 7. Output: `false`.

### 3. Edge Case: Time Factor — `n = 1000000`
**Input:** $10^6 = 2^6 \times 5^6$.
Six divisions by 2, then six by 5, reach 1. The loop runs $O(\log n)$ times — at most about 20 divisions. Output: `true`.

### 4. Edge Case: Space Factor — `n = 1`
**Input:** The base case.
1 has no prime factors and is ugly by convention. No divisions, $O(1)$ space. Output: `true`.

### 5. Almost-Impossible but Plausible — `n = 2147483647`
**Input:** The maximum 32-bit signed integer, which is the Mersenne prime $2^{31}-1$.
It has no factor of 2, 3, or 5, so it never reduces to 1. Output: `false`. (Any $n \le 0$ also returns `false`.)
