# LeetCode #1016: Subarray Sums Divisible by K

https://leetcode.com/problems/subarray-sums-divisible-by-k/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Prefix Sum + Hash Map ★** | $O(n)$ | $O(k)$ |

---

## Understanding the Methods

### Brute Force
Check every subarray by computing its sum and testing divisibility by k. Two nested loops give $O(n^2)$ time with no extra space.

### Optimal: Prefix Sum + Hash Map ★
Track the running prefix sum modulo k. If the same remainder appears twice, the subarray between those two indices is divisible by k. Store remainder frequencies in a hash map of size k.

**Constraints:**
* $1 \leq nums.length \leq 3 \times 10^4$
* $-10^4 \leq nums[i] \leq 10^4$
* $2 \leq k \leq 10^4$

## Solutions

### C#

```
public class Solution {
    public int SubarraysDivByK(int[] nums, int k) {
        // Count how many times each prefix-sum remainder has appeared
        int[] remainderCount = new int[k];
        remainderCount[0] = 1; // empty prefix has remainder 0
        int prefixSum = 0, result = 0;

        foreach (int num in nums) {
            // Advance running sum and normalize remainder to [0, k)
            prefixSum += num;
            int rem = ((prefixSum % k) + k) % k;

            // Every prior index with the same remainder forms a valid subarray
            result += remainderCount[rem];
            remainderCount[rem]++;
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def subarrays_div_by_k(self, nums: list[int], k: int) -> int:
        # Count how many times each prefix-sum remainder has appeared
        remainder_count = {0: 1}  # empty prefix has remainder 0
        prefix_sum = 0
        result = 0

        for num in nums:
            # Advance running sum and normalize remainder to [0, k)
            prefix_sum += num
            rem = prefix_sum % k

            # Every prior index with the same remainder forms a valid subarray
            result += remainder_count.get(rem, 0)
            remainder_count[rem] = remainder_count.get(rem, 0) + 1

        return result
```

### Rust

```
impl Solution {
    pub fn subarrays_div_by_k(nums: Vec<i32>, k: i32) -> i32 {
        // Count how many times each prefix-sum remainder has appeared
        let mut remainder_count = vec![0i32; k as usize];
        remainder_count[0] = 1; // empty prefix has remainder 0
        let mut prefix_sum = 0i32;
        let mut result = 0i32;

        for num in nums {
            // Advance running sum and normalize remainder to [0, k)
            prefix_sum += num;
            let rem = (((prefix_sum % k) + k) % k) as usize;

            // Every prior index with the same remainder forms a valid subarray
            result += remainder_count[rem];
            remainder_count[rem] += 1;
        }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [4, 5, 0, -2, -3, 1]`, `k = 5`
The prefix sums mod 5 are `[4, 4, 4, 2, 4, 0]`. Remainder 4 appears three times (plus the initial 0-count), giving $7$ valid subarrays in total.

### 2. Slightly Complex
**Input:** `nums = [5]`, `k = 9`
The only subarray `[5]` has sum 5, which is not divisible by 9. The map never records a collision, so the answer is $0$.

### 3. Edge Case: Time Factor
**Input:** `nums = [2, 2, 2, ..., 2]` (30 000 twos), `k = 2`
Every prefix sum is even, so the remainder is always 0. The count array entry for 0 grows to 30 001, and each step adds the previous count — yielding the maximum possible answer in a single linear pass.

### 4. Edge Case: Space Factor
**Input:** `nums = [1, 2, 3, ..., k-1]`, `k` large
All remainders are distinct; the count array fills with exactly one entry per slot and no collisions occur, so the answer is $0$ and the array occupies $O(k)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [-10000, 10000, -10000, 10000]`, `k = 10000`
Alternating signs keep the prefix sum oscillating between $\pm 10^4$. After normalization, remainders alternate between 0 and 0 (since $10000 \bmod 10000 = 0$), producing multiple valid pairs despite the extreme values.
