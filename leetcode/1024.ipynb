# LeetCode #1024: Video Stitching

https://leetcode.com/problems/video-stitching/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n \cdot T)$ | $O(1)$ |
| **Optimal: Greedy Interval Cover ★** | $O(n \log n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
At each position, try every clip to see which one extends coverage. Repeat until coverage reaches T or no progress is possible — $O(n \cdot T)$ time.

### Optimal: Greedy Interval Cover ★
Sort clips by start time. At each step, among all clips whose start is within current coverage, pick the one that extends coverage the furthest. Count the number of clip selections needed.

**Constraints:**
* $1 \leq clips.length \leq 100$
* $0 \leq clips[i][0] \leq clips[i][1] \leq 100$
* $1 \leq T \leq 100$

## Solutions

### C#

```
public class Solution {
    public int VideoStitching(int[][] clips, int time) {
        // Sort clips by start time so we process them left-to-right
        System.Array.Sort(clips, (a, b) => a[0] - b[0]);
        int clipsUsed = 0, currentEnd = 0, farthestReach = 0;
        int i = 0;

        while (currentEnd < time) {
            // Among clips that start within our current coverage, pick the one reaching farthest
            while (i < clips.Length && clips[i][0] <= currentEnd) {
                farthestReach = Math.Max(farthestReach, clips[i][1]);
                i++;
            }
            // No clip could extend coverage — impossible to stitch
            if (farthestReach <= currentEnd) return -1;
            currentEnd = farthestReach;
            clipsUsed++;
        }
        return clipsUsed;
    }
}
```

### Python

```
class Solution:
    def video_stitching(self, clips: list[list[int]], time: int) -> int:
        # Sort clips by start time so we process them left-to-right
        clips.sort()
        clips_used = 0
        current_end = 0
        farthest_reach = 0
        i = 0

        while current_end < time:
            # Among clips that start within our current coverage, pick the one reaching farthest
            while i < len(clips) and clips[i][0] <= current_end:
                farthest_reach = max(farthest_reach, clips[i][1])
                i += 1
            # No clip could extend coverage — impossible to stitch
            if farthest_reach <= current_end:
                return -1
            current_end = farthest_reach
            clips_used += 1

        return clips_used
```

### Go

```
import "sort"

func videoStitching(clips [][]int, time int) int {
    // Sort clips by start time so we process them left-to-right
    sort.Slice(clips, func(i, j int) bool { return clips[i][0] < clips[j][0] })
    clipsUsed, currentEnd, farthestReach, i := 0, 0, 0, 0

    for currentEnd < time {
        // Among clips that start within our current coverage, pick the one reaching farthest
        for i < len(clips) && clips[i][0] <= currentEnd {
            if clips[i][1] > farthestReach {
                farthestReach = clips[i][1]
            }
            i++
        }
        // No clip could extend coverage — impossible to stitch
        if farthestReach <= currentEnd {
            return -1
        }
        currentEnd = farthestReach
        clipsUsed++
    }
    return clipsUsed
}
```

### Rust

```
impl Solution {
    pub fn video_stitching(mut clips: Vec<Vec<i32>>, time: i32) -> i32 {
        // Sort clips by start time so we process them left-to-right
        clips.sort_by_key(|c| c[0]);
        let (mut clips_used, mut current_end, mut farthest_reach, mut i) = (0, 0, 0, 0);

        while current_end < time {
            // Among clips that start within our current coverage, pick the one reaching farthest
            while i < clips.len() && clips[i][0] <= current_end {
                farthest_reach = farthest_reach.max(clips[i][1]);
                i += 1;
            }
            // No clip could extend coverage — impossible to stitch
            if farthest_reach <= current_end {
                return -1;
            }
            current_end = farthest_reach;
            clips_used += 1;
        }
        clips_used
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `clips = [[0,2],[4,6],[8,10],[1,9],[1,5],[5,9]]`, `time = 10`
Sort by start gives `[[0,2],[1,5],[1,9],[4,6],[5,9],[8,10]]`. From 0, the farthest reach is 9; from 9 another clip reaches 10 — two clips suffice.

### 2. Slightly Complex
**Input:** `clips = [[0,1],[1,2]]`, `time = 5`
Coverage reaches only 2; no clip starts between 2 and 5, so the function returns $-1$ — the interval cannot be fully stitched.

### 3. Edge Case: Time Factor
**Input:** `clips = [[0,1],[0,2],[0,3],...,[0,100]]`, `time = 100`
All clips start at 0; we scan every clip in the inner loop once on the very first outer iteration, selecting the one with end 100 — one clip, $O(n)$ inner work total.

### 4. Edge Case: Space Factor
**Input:** `clips = [[0,100]]`, `time = 100`
A single clip covers the entire range. The algorithm uses $O(1)$ extra space and returns $1$ immediately.

### 5. Almost-Impossible but Plausible
**Input:** `clips = [[0,0],[0,100]]`, `time = 100`
The zero-length clip `[0,0]` is dominated by `[0,100]`. The inner loop processes both but picks the 100-end clip, still returning $1$. Zero-length clips are harmless.
