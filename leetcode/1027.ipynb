# LeetCode #1027: Longest Arithmetic Subsequence

https://leetcode.com/problems/longest-arithmetic-subsequence/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^3)$ | $O(1)$ |
| **Optimal: DP with HashMap ★** | $O(n^2)$ | $O(n^2)$ |

---

## Understanding the Methods

### Brute Force
For every pair of indices $(i, j)$ treat them as the last two elements of an arithmetic subsequence, then scan all preceding elements to extend it. Repeating this for all pairs gives $O(n^3)$.

### Optimal: DP with HashMap ★
Maintain `dp[i]` as a dictionary mapping each arithmetic difference `d` to the length of the longest arithmetic subsequence ending at index `i` with that difference. For each pair $(j, i)$ where $j < i$, update `dp[i][d] = dp[j].get(d, 1) + 1`. A single two-level pass achieves $O(n^2)$ time.

**Constraints:**
* $2 \leq nums.length \leq 1000$
* $0 \leq nums[i] \leq 10^4$

## Solutions

### C#

```
using System.Collections.Generic;
public class Solution {
    public int LongestArithSeqLength(int[] nums) {
        int n = nums.Length;
        // dp[i] maps difference -> longest arithmetic subsequence length ending at i
        var dp = new Dictionary<int,int>[n];
        for (int i = 0; i < n; i++) dp[i] = new Dictionary<int,int>();
        int best = 2;
        for (int i = 1; i < n; i++) {
            for (int j = 0; j < i; j++) {
                int d = nums[i] - nums[j];
                // Extend the subsequence ending at j with difference d
                int prev = dp[j].ContainsKey(d) ? dp[j][d] : 1;
                int cur = prev + 1;
                if (!dp[i].ContainsKey(d) || dp[i][d] < cur)
                    dp[i][d] = cur;
                if (cur > best) best = cur;
            }
        }
        return best;
    }
}
```

### Python

```
class Solution:
    def longest_arith_seq_length(self, nums: list[int]) -> int:
        n = len(nums)
        # dp[i] maps difference -> longest arithmetic subsequence length ending at i
        dp = [{} for _ in range(n)]
        best = 2
        for i in range(1, n):
            for j in range(i):
                d = nums[i] - nums[j]
                # Extend the subsequence ending at j with difference d
                cur = dp[j].get(d, 1) + 1
                if cur > dp[i].get(d, 0):
                    dp[i][d] = cur
                if cur > best:
                    best = cur
        return best
```

### Go

```
func longestArithSeqLength(nums []int) int {
    n := len(nums)
    // dp[i] maps difference -> longest arithmetic subsequence length ending at i
    dp := make([]map[int]int, n)
    for i := range dp { dp[i] = map[int]int{} }
    best := 2
    for i := 1; i < n; i++ {
        for j := 0; j < i; j++ {
            d := nums[i] - nums[j]
            // Extend the subsequence ending at j with difference d
            prev, ok := dp[j][d]
            if !ok { prev = 1 }
            cur := prev + 1
            if cur > dp[i][d] { dp[i][d] = cur }
            if cur > best { best = cur }
        }
    }
    return best
}
```

### Rust

```
use std::collections::HashMap;
impl Solution {
    pub fn longest_arith_seq_length(nums: Vec<i32>) -> i32 {
        let n = nums.len();
        // dp[i] maps difference -> longest arithmetic subsequence length ending at i
        let mut dp: Vec<HashMap<i32,i32>> = vec![HashMap::new(); n];
        let mut best = 2i32;
        for i in 1..n {
            for j in 0..i {
                let d = nums[i] - nums[j];
                // Extend the subsequence ending at j with difference d
                let prev = *dp[j].get(&d).unwrap_or(&1);
                let cur = prev + 1;
                let e = dp[i].entry(d).or_insert(0);
                if cur > *e { *e = cur; }
                if cur > best { best = cur; }
            }
        }
        best
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [3, 6, 9, 12]`
Every consecutive pair has difference $3$, so the whole array is one arithmetic subsequence. The DP finds `dp[3][3] = 4` after extending through indices 0→1→2→3. Answer: **4**.

### 2. Slightly Complex
**Input:** `nums = [9, 4, 7, 2, 10]`
Multiple differences coexist. The pair $(4, 7)$ and $(7, 10)$ share difference $3$; the DP records `dp[4][3] = 3` at index 4 (value 10). Answer: **3**.

### 3. Edge Case: Time Factor
**Input:** `nums` of length 1000 with all distinct values and no arithmetic triple.
Every pair $(j, i)$ creates a fresh dictionary entry, so all $O(n^2) = 500{,}000$ entries are filled. Each lookup is $O(1)$ amortized, so the algorithm still runs in $\approx 10^6$ operations.

### 4. Edge Case: Space Factor
**Input:** `nums = [0, 1, 2, ..., 999]` (1000 distinct values).
Each index $i$ accumulates up to $i$ different differences in its HashMap. Total entries across all $n$ maps reach $\approx \frac{n^2}{2} = 500{,}000$, the worst-case $O(n^2)$ space scenario.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [10000, 0, 10000, 0, 10000]`
Differences alternate between $-10000$ and $+10000$. No arithmetic subsequence of length $> 2$ exists with a fixed difference. The DP explores all pairs but the best stays at $2$; none of the large-difference chains can connect three elements.
