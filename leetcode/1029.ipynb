# LeetCode #1029: Two City Scheduling

https://leetcode.com/problems/two-city-scheduling/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^{2n})$ | $O(1)$ |
| **Optimal: Greedy Sort ★** | $O(n \log n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Try every assignment of n people to city A and n people to city B, computing the total cost for each valid split. Exponential in the number of people.

### Optimal: Greedy Sort ★
Sort people by the "regret" of sending them to city B instead of A: `costA - costB`. The n people with the smallest (most negative) regret go to A, and the rest go to B. This minimises total cost in one sorting step.

**Constraints:**
* $2 \leq costs.length \leq 200$ (even length)
* $0 \leq costs[i][0], costs[i][1] \leq 1000$

## Solutions

### C#

```
public class Solution {
    public int TwoCitySchedCost(int[][] costs) {
        // Sort by how much cheaper it is to fly to city A versus city B
        // (negative = A is cheaper, positive = B is cheaper)
        System.Array.Sort(costs, (a, b) => (a[0] - a[1]) - (b[0] - b[1]));
        int n = costs.Length / 2;
        int total = 0;

        // First half (cheapest-A people) fly to A; second half fly to B
        for (int i = 0; i < n; i++)
            total += costs[i][0];
        for (int i = n; i < 2 * n; i++)
            total += costs[i][1];

        return total;
    }
}
```

### Python

```
class Solution:
    def two_city_sched_cost(self, costs: list[list[int]]) -> int:
        # Sort by how much cheaper it is to fly to city A versus city B
        costs.sort(key=lambda c: c[0] - c[1])
        n = len(costs) // 2
        total = 0

        # First half (cheapest-A people) fly to A; second half fly to B
        for i in range(n):
            total += costs[i][0]
        for i in range(n, 2 * n):
            total += costs[i][1]

        return total
```

### Go

```
import "sort"

func twoCitySchedCost(costs [][]int) int {
    // Sort by how much cheaper it is to fly to city A versus city B
    sort.Slice(costs, func(i, j int) bool {
        return (costs[i][0] - costs[i][1]) < (costs[j][0] - costs[j][1])
    })
    n := len(costs) / 2
    total := 0

    // First half (cheapest-A people) fly to A; second half fly to B
    for i := 0; i < n; i++ {
        total += costs[i][0]
    }
    for i := n; i < 2*n; i++ {
        total += costs[i][1]
    }
    return total
}
```

### Rust

```
impl Solution {
    pub fn two_city_sched_cost(mut costs: Vec<Vec<i32>>) -> i32 {
        // Sort by how much cheaper it is to fly to city A versus city B
        costs.sort_by_key(|c| c[0] - c[1]);
        let n = costs.len() / 2;
        let mut total = 0;

        // First half (cheapest-A people) fly to A; second half fly to B
        for i in 0..n {
            total += costs[i][0];
        }
        for i in n..2 * n {
            total += costs[i][1];
        }
        total
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `costs = [[10,20],[30,200],[400,50],[30,20]]`
After sorting by `costA - costB`: `[10,20], [30,20], [30,200], [400,50]`. Person 0 and 1 fly to A ($10+30$), persons 2 and 3 fly to B ($200+50$) — total $290$.

### 2. Slightly Complex
**Input:** `costs = [[259,770],[448,54],[926,667],[184,139],[840,118],[577,469]]`
Regret values push the two people who strongly prefer A into that city, and those who strongly prefer B into B — the greedy sort surfaces the optimal split without examining all $\binom{6}{3}=20$ assignments.

### 3. Edge Case: Time Factor
**Input:** `n = 100` people, all `costs[i] = [1000, 1000]`
Every regret is 0; the sort is a no-op. All $2n = 200$ people are processed in a single $O(n)$ loop — worst-case sort still $O(n \log n)$ overall.

### 4. Edge Case: Space Factor
**Input:** `costs = [[0,1000],[0,1000],...] ` ($n = 100$ pairs)
Every person should go to A ($costA = 0$). After sorting all regrets are equal; the first 100 go to A for free, the last 100 go to B for $1000$ each. No extra space beyond the input array.

### 5. Almost-Impossible but Plausible
**Input:** `costs = [[1000,0],[1000,0],...,[0,1000],[0,1000]]` (half each type, $n=100$)
The greedy places the zero-B-cost people in B and zero-A-cost people in A. Total cost is 0 — the perfect assignment is automatically found by sorting on regret.
