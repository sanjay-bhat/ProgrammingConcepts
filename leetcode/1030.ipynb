# LeetCode #1030: Matrix Cells in Distance Order

https://leetcode.com/problems/matrix-cells-in-distance-order/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (Sort)** | $O(R \cdot C \cdot \log(R \cdot C))$ | $O(R \cdot C)$ |
| **Optimal: BFS ★** | $O(R \cdot C)$ | $O(R \cdot C)$ |

---

## Understanding the Methods

### Brute Force (Sort)
Generate all $R \times C$ cells, compute their Manhattan distance from $(rCenter, cCenter)$, then sort — straightforward but $O(R \cdot C \cdot \log(R \cdot C))$ time.

### Optimal: BFS ★
Start a BFS from the center cell. Because BFS explores in order of hop count and hop count equals Manhattan distance on a grid, visited cells emerge in non-decreasing distance order with no sorting needed — $O(R \cdot C)$ total.

**Constraints:**
* $1 \leq R, C \leq 100$
* $0 \leq rCenter < R$, $0 \leq cCenter < C$

## Solutions

### C#

```
public class Solution {
    public int[][] AllCellsDistOrder(int rows, int cols, int rCenter, int cCenter) {
        bool[,] visited = new bool[rows, cols];
        int[][] result = new int[rows * cols][];
        Queue<int[]> queue = new Queue<int[]>();

        // Seed BFS from the center — distance 0 comes first automatically
        queue.Enqueue(new int[] { rCenter, cCenter });
        visited[rCenter, cCenter] = true;
        int idx = 0;
        int[] dr = { 0, 0, 1, -1 }, dc = { 1, -1, 0, 0 };

        while (queue.Count > 0) {
            int[] cell = queue.Dequeue();
            result[idx++] = cell;

            // Expand to all 4 neighbours, each one step farther from the center
            for (int d = 0; d < 4; d++) {
                int nr = cell[0] + dr[d], nc = cell[1] + dc[d];
                if (nr >= 0 && nr < rows && nc >= 0 && nc < cols && !visited[nr, nc]) {
                    visited[nr, nc] = true;
                    queue.Enqueue(new int[] { nr, nc });
                }
            }
        }
        return result;
    }
}
```

### Python

```
from collections import deque

class Solution:
    def all_cells_dist_order(self, rows: int, cols: int, r_center: int, c_center: int) -> list[list[int]]:
        visited = [[False] * cols for _ in range(rows)]
        result = []
        queue = deque([[r_center, c_center]])
        visited[r_center][c_center] = True

        # BFS expands by Manhattan distance — no sort needed
        while queue:
            r, c = queue.popleft()
            result.append([r, c])
            for dr, dc in ((0,1),(0,-1),(1,0),(-1,0)):
                nr, nc = r + dr, c + dc
                if 0 <= nr < rows and 0 <= nc < cols and not visited[nr][nc]:
                    visited[nr][nc] = True
                    queue.append([nr, nc])

        return result
```

### Go

```
func allCellsDistOrder(rows int, cols int, rCenter int, cCenter int) [][]int {
    visited := make([][]bool, rows)
    for i := range visited {
        visited[i] = make([]bool, cols)
    }
    result := make([][]int, 0, rows*cols)
    queue := [][]int{{rCenter, cCenter}}
    visited[rCenter][cCenter] = true
    dirs := [][2]int{{0,1},{0,-1},{1,0},{-1,0}}

    // BFS expands by Manhattan distance — no sort needed
    for len(queue) > 0 {
        cell := queue[0]
        queue = queue[1:]
        result = append(result, cell)
        for _, d := range dirs {
            nr, nc := cell[0]+d[0], cell[1]+d[1]
            if nr >= 0 && nr < rows && nc >= 0 && nc < cols && !visited[nr][nc] {
                visited[nr][nc] = true
                queue = append(queue, []int{nr, nc})
            }
        }
    }
    return result
}
```

### Rust

```
use std::collections::VecDeque;

impl Solution {
    pub fn all_cells_dist_order(rows: i32, cols: i32, r_center: i32, c_center: i32) -> Vec<Vec<i32>> {
        let (rows, cols) = (rows as usize, cols as usize);
        let mut visited = vec![vec![false; cols]; rows];
        let mut result = Vec::with_capacity(rows * cols);
        let mut queue = VecDeque::new();
        queue.push_back((r_center as usize, c_center as usize));
        visited[r_center as usize][c_center as usize] = true;
        let dirs: [(i32, i32); 4] = [(0,1),(0,-1),(1,0),(-1,0)];

        // BFS expands by Manhattan distance — no sort needed
        while let Some((r, c)) = queue.pop_front() {
            result.push(vec![r as i32, c as i32]);
            for (dr, dc) in dirs {
                let (nr, nc) = (r as i32 + dr, c as i32 + dc);
                if nr >= 0 && nr < rows as i32 && nc >= 0 && nc < cols as i32 {
                    let (nr, nc) = (nr as usize, nc as usize);
                    if !visited[nr][nc] {
                        visited[nr][nc] = true;
                        queue.push_back((nr, nc));
                    }
                }
            }
        }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `rows=2, cols=2, rCenter=0, cCenter=1`
BFS from $(0,1)$ visits cells at distance 0, 1, 2 in order: $(0,1) \to (0,0),(1,1) \to (1,0)$. Result: `[[0,1],[0,0],[1,1],[1,0]]`.

### 2. Slightly Complex
**Input:** `rows=2, cols=3, rCenter=1, cCenter=2`
Center is a corner of the 2×3 grid. BFS fans outward; cells at distance 1 are $(0,2),(1,1)$, distance 2 are $(0,1),(1,0)$, distance 3 is $(0,0)$.

### 3. Edge Case: Time Factor
**Input:** `rows=100, cols=100, rCenter=0, cCenter=0`
Center is a corner; the farthest cell $(99,99)$ is at distance $198$. All $10{,}000$ cells are visited in $O(R \cdot C)$ time — the BFS never revisits any cell.

### 4. Edge Case: Space Factor
**Input:** `rows=100, cols=100, rCenter=50, cCenter=50`
Centered grid means the BFS queue peaks in size when the expanding diamond front spans the widest row — up to $\approx 200$ cells in the queue at once. Total output is $10{,}000$ cells.

### 5. Almost-Impossible but Plausible
**Input:** `rows=1, cols=1, rCenter=0, cCenter=0`
A $1 \times 1$ grid. BFS immediately dequeues the single cell and returns `[[0,0]]` — no neighbours to explore.
