# LeetCode #1034: Coloring A Border

https://leetcode.com/problems/coloring-a-border/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(mn)$ | $O(mn)$ |
| **Optimal: BFS Connected-Component ★** | $O(mn)$ | $O(mn)$ |

---

## Understanding the Methods

### Brute Force
DFS/BFS naively revisiting cells without a clean visited array can lead to redundant checks; identifying border cells requires re-scanning neighbours each time.

### Optimal: BFS Connected-Component ★
BFS from the source cell to discover all cells belonging to its original-colour component. A cell is a **border cell** if any of its four neighbours is out-of-bounds or has a different original colour. After discovery, colour only border cells to avoid altering interior cells.

**Constraints:**
* `m == grid.length`, `n == grid[0].length`
* `1 <= m, n <= 50`
* `1 <= grid[i][j] <= 1000`
* `1 <= r0 < m`, `1 <= c0 < n`
* `1 <= color <= 1000`

## Solutions

### C#

```
public class Solution {
    public int[][] ColorBorder(int[][] grid, int r0, int c0, int color) {
        int m = grid.Length, n = grid[0].Length;
        int original = grid[r0][c0];
        bool[][] visited = new bool[m][];
        for (int i = 0; i < m; i++) visited[i] = new bool[n];

        // BFS to find the connected component
        var queue = new Queue<(int r, int c)>();
        queue.Enqueue((r0, c0));
        visited[r0][c0] = true;
        var border = new List<(int r, int c)>();

        int[] dr = {-1, 1, 0, 0};
        int[] dc = {0, 0, -1, 1};

        while (queue.Count > 0) {
            var (r, c) = queue.Dequeue();
            bool isBorder = false;

            for (int d = 0; d < 4; d++) {
                int nr = r + dr[d], nc = c + dc[d];
                if (nr < 0 || nr >= m || nc < 0 || nc >= n || grid[nr][nc] != original) {
                    // Out-of-bounds or different colour means this cell is on the border
                    isBorder = true;
                } else if (!visited[nr][nc]) {
                    visited[nr][nc] = true;
                    queue.Enqueue((nr, nc));
                }
            }
            if (isBorder) border.Add((r, c));
        }

        // Paint only border cells
        foreach (var (r, c) in border)
            grid[r][c] = color;

        return grid;
    }
}
```

### Python

```
from collections import deque

class Solution:
    def colorBorder(self, grid: list[list[int]], r0: int, c0: int, color: int) -> list[list[int]]:
        m, n = len(grid), len(grid[0])
        original = grid[r0][c0]
        visited = [[False] * n for _ in range(m)]
        queue = deque([(r0, c0)])
        visited[r0][c0] = True
        border = []

        dirs = [(-1, 0), (1, 0), (0, -1), (0, 1)]
        while queue:
            r, c = queue.popleft()
            is_border = False
            for dr, dc in dirs:
                nr, nc = r + dr, c + dc
                if not (0 <= nr < m and 0 <= nc < n) or grid[nr][nc] != original:
                    # Neighbour out-of-bounds or different colour → border cell
                    is_border = True
                elif not visited[nr][nc]:
                    visited[nr][nc] = True
                    queue.append((nr, nc))
            if is_border:
                border.append((r, c))

        for r, c in border:
            grid[r][c] = color
        return grid
```

### Go

```
func colorBorder(grid [][]int, r0 int, c0 int, color int) [][]int {
    m, n := len(grid), len(grid[0])
    original := grid[r0][c0]
    visited := make([][]bool, m)
    for i := range visited {
        visited[i] = make([]bool, n)
    }

    type point struct{ r, c int }
    queue := []point{{r0, c0}}
    visited[r0][c0] = true
    var border []point

    dirs := [][2]int{{-1, 0}, {1, 0}, {0, -1}, {0, 1}}
    for len(queue) > 0 {
        cur := queue[0]
        queue = queue[1:]
        isBorder := false
        for _, d := range dirs {
            nr, nc := cur.r+d[0], cur.c+d[1]
            if nr < 0 || nr >= m || nc < 0 || nc >= n || grid[nr][nc] != original {
                // Neighbour absent or different colour → border cell
                isBorder = true
            } else if !visited[nr][nc] {
                visited[nr][nc] = true
                queue = append(queue, point{nr, nc})
            }
        }
        if isBorder {
            border = append(border, cur)
        }
    }

    // Paint only border cells to preserve interior
    for _, p := range border {
        grid[p.r][p.c] = color
    }
    return grid
}
```

### Rust

```
use std::collections::VecDeque;

impl Solution {
    pub fn color_border(mut grid: Vec<Vec<i32>>, r0: i32, c0: i32, color: i32) -> Vec<Vec<i32>> {
        let m = grid.len() as i32;
        let n = grid[0].len() as i32;
        let original = grid[r0 as usize][c0 as usize];
        let mut visited = vec![vec![false; n as usize]; m as usize];
        let mut queue = VecDeque::new();
        queue.push_back((r0, c0));
        visited[r0 as usize][c0 as usize] = true;
        let mut border = Vec::new();
        let dirs: [(i32, i32); 4] = [(-1, 0), (1, 0), (0, -1), (0, 1)];

        while let Some((r, c)) = queue.pop_front() {
            let mut is_border = false;
            for (dr, dc) in dirs {
                let nr = r + dr;
                let nc = c + dc;
                if nr < 0 || nr >= m || nc < 0 || nc >= n
                    || grid[nr as usize][nc as usize] != original {
                    // Out-of-bounds or different colour marks this as a border cell
                    is_border = true;
                } else if !visited[nr as usize][nc as usize] {
                    visited[nr as usize][nc as usize] = true;
                    queue.push_back((nr, nc));
                }
            }
            if is_border { border.push((r, c)); }
        }

        for (r, c) in border {
            grid[r as usize][c as usize] = color;
        }
        grid
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `grid = [[1,1],[1,2]], r0=0, c0=0, color=3`
Component of colour 1 occupies all three top-left cells; each borders a cell of colour 2 or the grid edge, so all three become 3. Result: `[[3,3],[3,2]]`.

### 2. Slightly Complex
**Input:** `grid = [[1,2,2],[2,3,2]], r0=0, c0=1, color=4`
Cell (0,1) has colour 2; BFS finds its connected component {(0,1),(0,2),(1,0)→no,(1,2)}. Interior cells (if any) stay untouched; only border cells are repainted 4.

### 3. Edge Case: Time Factor
**Input:** `50×50` grid all colour 1, `r0=25, c0=25, color=2`.
BFS visits all 2500 cells but only cells on the outer ring (196 cells) qualify as border. The $O(mn)$ traversal processes every cell exactly once.

### 4. Edge Case: Space Factor
**Input:** `50×50` grid all colour 1, BFS queue may hold up to $mn$ entries; visited array is $50 \times 50 = 2500$ booleans — $O(mn)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `grid = [[1,1,1],[1,1,1],[1,1,1]], r0=1, c0=1, color=1`
Target colour equals original colour; all border cells are repainted 1 (no visual change), but the algorithm still correctly identifies and processes them without corrupting interior cells.
