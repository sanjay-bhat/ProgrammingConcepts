# LeetCode #1035: Uncrossed Lines

https://leetcode.com/problems/uncrossed-lines/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^{\min(m,n)})$ | $O(m+n)$ |
| **Optimal: LCS DP ★** | $O(m \times n)$ | $O(m \times n)$ |

---

## Understanding the Methods

### Brute Force
Recursively match each element of `nums1` to some element of `nums2` while ensuring order is preserved, exploring all possibilities without memoization. Exponential branches make this infeasible beyond small inputs.

### Optimal: LCS DP ★
A connecting line between `nums1[i]` and `nums2[j]` never crosses another line iff both arrays are traversed in order — exactly the condition for a common subsequence. The maximum number of non-crossing lines equals the Longest Common Subsequence (LCS) of the two arrays, solvable in $O(m \times n)$ DP.

**Constraints:**
* $1 \leq nums1.length, nums2.length \leq 500$
* $1 \leq nums1[i], nums2[j] \leq 2000$

## Solutions

### C#

```
public class Solution {
    public int MaxUncrossedLines(int[] nums1, int[] nums2) {
        int m = nums1.Length, n = nums2.Length;
        // dp[i][j] = max non-crossing lines using nums1[0..i) and nums2[0..j)
        var dp = new int[m + 1, n + 1];
        for (int i = 1; i <= m; i++) {
            for (int j = 1; j <= n; j++) {
                if (nums1[i-1] == nums2[j-1])
                    // Matching values let us extend the best pairing found so far
                    dp[i, j] = dp[i-1, j-1] + 1;
                else
                    // Carry forward the best achievable without this pair
                    dp[i, j] = Math.Max(dp[i-1, j], dp[i, j-1]);
            }
        }
        return dp[m, n];
    }
}
```

### Python

```
class Solution:
    def max_uncrossed_lines(self, nums1: list[int], nums2: list[int]) -> int:
        m, n = len(nums1), len(nums2)
        # dp[i][j] = max non-crossing lines using nums1[0..i) and nums2[0..j)
        dp = [[0] * (n + 1) for _ in range(m + 1)]
        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if nums1[i-1] == nums2[j-1]:
                    # Matching values let us extend the best pairing found so far
                    dp[i][j] = dp[i-1][j-1] + 1
                else:
                    # Carry forward the best achievable without this pair
                    dp[i][j] = max(dp[i-1][j], dp[i][j-1])
        return dp[m][n]
```

### Go

```
func maxUncrossedLines(nums1 []int, nums2 []int) int {
    m, n := len(nums1), len(nums2)
    // dp[i][j] = max non-crossing lines using nums1[0..i) and nums2[0..j)
    dp := make([][]int, m+1)
    for i := range dp { dp[i] = make([]int, n+1) }
    for i := 1; i <= m; i++ {
        for j := 1; j <= n; j++ {
            if nums1[i-1] == nums2[j-1] {
                // Matching values let us extend the best pairing found so far
                dp[i][j] = dp[i-1][j-1] + 1
            } else {
                // Carry forward the best achievable without this pair
                if dp[i-1][j] > dp[i][j-1] { dp[i][j] = dp[i-1][j] } else { dp[i][j] = dp[i][j-1] }
            }
        }
    }
    return dp[m][n]
}
```

### Rust

```
impl Solution {
    pub fn max_uncrossed_lines(nums1: Vec<i32>, nums2: Vec<i32>) -> i32 {
        let (m, n) = (nums1.len(), nums2.len());
        // dp[i][j] = max non-crossing lines using nums1[0..i) and nums2[0..j)
        let mut dp = vec![vec![0i32; n + 1]; m + 1];
        for i in 1..=m {
            for j in 1..=n {
                if nums1[i-1] == nums2[j-1] {
                    // Matching values let us extend the best pairing found so far
                    dp[i][j] = dp[i-1][j-1] + 1;
                } else {
                    // Carry forward the best achievable without this pair
                    dp[i][j] = dp[i-1][j].max(dp[i][j-1]);
                }
            }
        }
        dp[m][n]
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums1 = [1,4,2]`, `nums2 = [1,2,4]`
The LCS is `[1, 4]` or `[1, 2]` — both length 2. DP fills the table and returns **2** non-crossing lines.

### 2. Slightly Complex
**Input:** `nums1 = [2,5,1,2,5]`, `nums2 = [10,5,2,1,5,2]`
Multiple shared values with valid orderings. The DP traces the longest common subsequence `[5,1,5]` or `[2,5,2]`, each of length 3. Answer: **3**.

### 3. Edge Case: Time Factor
**Input:** `nums1` and `nums2` both of length 500 with no common elements.
Every cell uses only the `max` branch — no match shortcut. All $500 \times 500 = 250{,}000$ cells are evaluated, exercising the worst-case time of $O(m \times n)$.

### 4. Edge Case: Space Factor
**Input:** `nums1 = [1,2,...,500]`, `nums2 = [1,2,...,500]` (identical arrays).
The full $501 \times 501$ DP table fills with values from 0 to 500 along the diagonal, consuming the maximum $O(m \times n)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `nums1 = [2000,2000,...,2000]` (500 twos), `nums2 = [2000,2000,...,2000]` (500 twos).
Every pair matches, so `dp[i][j] = min(i,j)`. The answer is 500, but the DP must still fill all 250,000 cells because no early exit applies — every cell takes the match branch.
