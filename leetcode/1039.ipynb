# LeetCode #1039: Minimum Score Triangulation of Polygon

https://leetcode.com/problems/minimum-score-triangulation-of-polygon/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n!)$ | $O(n)$ |
| **Optimal: Interval DP ★** | $O(n^3)$ | $O(n^2)$ |

---

## Understanding the Methods

### Brute Force
Enumerate all possible triangulations recursively. The number of triangulations of an n-gon is the Catalan number $C_{n-2}$, which grows exponentially.

### Optimal: Interval DP ★
`dp[i][j]` = minimum triangulation score for the sub-polygon from vertex `i` to vertex `j`. For each pair $(i, j)$, try every vertex $k$ between them as the third vertex of a triangle that spans the full interval, then recurse on the two sub-polygons.

**Constraints:**
* $3 \leq values.length \leq 50$
* $1 \leq values[i] \leq 100$

## Solutions

### C#

```
public class Solution {
    public int MinScoreTriangulation(int[] values) {
        int n = values.Length;
        int[,] dp = new int[n, n];

        // Fill by increasing interval length (need at least 3 vertices for a triangle)
        for (int len = 2; len < n; len++) {
            for (int i = 0; i + len < n; i++) {
                int j = i + len;
                dp[i, j] = int.MaxValue;
                // Try every interior vertex k as the apex of the spanning triangle
                for (int k = i + 1; k < j; k++) {
                    dp[i, j] = Math.Min(dp[i, j],
                        dp[i, k] + values[i] * values[k] * values[j] + dp[k, j]);
                }
            }
        }
        return dp[0, n - 1];
    }
}
```

### Python

```
class Solution:
    def min_score_triangulation(self, values: list[int]) -> int:
        n = len(values)
        dp = [[0] * n for _ in range(n)]

        # Fill by increasing interval length
        for length in range(2, n):
            for i in range(n - length):
                j = i + length
                dp[i][j] = float('inf')
                # Try every interior vertex k as the apex of the spanning triangle
                for k in range(i + 1, j):
                    dp[i][j] = min(dp[i][j],
                        dp[i][k] + values[i] * values[k] * values[j] + dp[k][j])

        return dp[0][n - 1]
```

### Go

```
func minScoreTriangulation(values []int) int {
    n := len(values)
    dp := make([][]int, n)
    for i := range dp {
        dp[i] = make([]int, n)
    }

    // Fill by increasing interval length
    for length := 2; length < n; length++ {
        for i := 0; i+length < n; i++ {
            j := i + length
            dp[i][j] = 1<<31 - 1
            // Try every interior vertex k as the apex of the spanning triangle
            for k := i + 1; k < j; k++ {
                cost := dp[i][k] + values[i]*values[k]*values[j] + dp[k][j]
                if cost < dp[i][j] {
                    dp[i][j] = cost
                }
            }
        }
    }
    return dp[0][n-1]
}
```

### Rust

```
impl Solution {
    pub fn min_score_triangulation(values: Vec<i32>) -> i32 {
        let n = values.len();
        let mut dp = vec![vec![0i32; n]; n];

        // Fill by increasing interval length
        for length in 2..n {
            for i in 0..n - length {
                let j = i + length;
                dp[i][j] = i32::MAX;
                // Try every interior vertex k as the apex of the spanning triangle
                for k in i + 1..j {
                    let cost = dp[i][k] + values[i] * values[k] * values[j] + dp[k][j];
                    if cost < dp[i][j] {
                        dp[i][j] = cost;
                    }
                }
            }
        }
        dp[0][n - 1]
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `values = [1,2,3]`
Only one triangle is possible: $1 \times 2 \times 3 = 6$. `dp[0][2] = 6`.

### 2. Slightly Complex
**Input:** `values = [3,7,4,5]`
Two triangulations: `(0,1,3)+(1,2,3) = 3*7*5 + 7*4*5 = 105+140 = 245` vs `(0,1,2)+(0,2,3) = 3*7*4 + 3*4*5 = 84+60 = 144`. Minimum is $144$.

### 3. Edge Case: Time Factor
**Input:** `values` has 50 elements, all equal to 100
The triple-nested loop runs $O(n^3) = 125{,}000$ iterations. With uniform values the product $100^3 = 10^6$ fits in an int; the DP fills in $O(n^2) = 2500$ cells.

### 4. Edge Case: Space Factor
**Input:** `values` has 50 elements
The DP table requires $50 \times 50 = 2500$ integers — small, but it represents the maximum allocation for this problem size.

### 5. Almost-Impossible but Plausible
**Input:** `values = [1,100,1,100,1,100,...,1]` (alternating)
High-value vertices are unavoidable in every triangle; the DP minimises their frequency by routing triangles to avoid pairing two 100-value vertices, but the score is still large due to the constraint structure.
