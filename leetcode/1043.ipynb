# LeetCode #1043: Partition Array for Maximum Sum

https://leetcode.com/problems/partition-array-for-maximum-sum/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(k^n)$ | $O(n)$ |
| **Optimal: 1-D DP ★** | $O(n \times k)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Try every possible way to partition the array into segments of length at most $k$ by recursion. The exponential branching factor makes this infeasible for large inputs.

### Optimal: 1-D DP ★
`dp[i]` stores the maximum sum achievable for the first `i` elements. For each position, try extending a window of width 1 through $k$ backward, taking the segment maximum times its length. Each cell depends only on earlier cells, so a single left-to-right pass suffices.

**Constraints:**
* $1 \leq \text{arr.length} \leq 500$
* $0 \leq \text{arr}[i] \leq 10^9$
* $1 \leq k \leq \text{arr.length}$

## Solutions

### C#

```
public class Solution {
    public int MaxSumAfterPartitioning(int[] arr, int k) {
        int n = arr.Length;
        int[] dp = new int[n + 1];
        for (int i = 1; i <= n; i++) {
            int maxVal = 0;
            // Extend the current window backward up to k elements
            for (int j = 1; j <= k && i - j >= 0; j++) {
                // Track the maximum element seen in this window
                maxVal = Math.Max(maxVal, arr[i - j]);
                // Fill the whole window with that max and add the best prefix before it
                dp[i] = Math.Max(dp[i], dp[i - j] + maxVal * j);
            }
        }
        return dp[n];
    }
}
```

### Python

```
class Solution:
    def max_sum_after_partitioning(self, arr: list[int], k: int) -> int:
        n = len(arr)
        dp = [0] * (n + 1)
        for i in range(1, n + 1):
            max_val = 0
            # Extend the current window backward up to k elements
            for j in range(1, min(k, i) + 1):
                # Track the maximum element seen in this window
                max_val = max(max_val, arr[i - j])
                # Fill the whole window with that max and add the best prefix before it
                dp[i] = max(dp[i], dp[i - j] + max_val * j)
        return dp[n]
```

### Go

```
func maxSumAfterPartitioning(arr []int, k int) int {
    n := len(arr)
    dp := make([]int, n+1)
    for i := 1; i <= n; i++ {
        maxVal := 0
        // Extend the current window backward up to k elements
        for j := 1; j <= k && i-j >= 0; j++ {
            // Track the maximum element seen in this window
            if arr[i-j] > maxVal { maxVal = arr[i-j] }
            // Fill the whole window with that max and add the best prefix before it
            candidate := dp[i-j] + maxVal*j
            if candidate > dp[i] { dp[i] = candidate }
        }
    }
    return dp[n]
}
```

### Rust

```
impl Solution {
    pub fn max_sum_after_partitioning(arr: Vec<i32>, k: i32) -> i32 {
        let n = arr.len();
        let k = k as usize;
        let mut dp = vec![0i32; n + 1];
        for i in 1..=n {
            let mut max_val = 0i32;
            // Extend the current window backward up to k elements
            for j in 1..=k.min(i) {
                // Track the maximum element seen in this window
                max_val = max_val.max(arr[i - j]);
                // Fill the whole window with that max and add the best prefix before it
                dp[i] = dp[i].max(dp[i - j] + max_val * j as i32);
            }
        }
        dp[n]
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [1, 15, 7, 9, 2, 5, 10]`, `k = 3`
The optimal partition is `[1, 15, 7] | [9, 2, 5] | [10]`, yielding $15 \times 3 + 9 \times 3 + 10 = 84$. Each segment is filled with its maximum, which the DP captures by tracking the running max as the window expands.

### 2. Slightly Complex
**Input:** `arr = [1, 4, 1, 5, 7, 3, 6, 1, 9, 9, 3]`, `k = 4`
Mixed highs and lows require the DP to weigh whether to merge a high-value element into a larger window or isolate it. Final answer is $83$.

### 3. Edge Case: Time Factor
**Input:** `arr = [1] * 500`, `k = 500`
Every inner loop runs the full $k$ iterations at every position, hitting the worst-case $O(n \times k) = O(500 \times 500) = 250{,}000$ operations. All values equal 1, so every partition yields the same total.

### 4. Edge Case: Space Factor
**Input:** `arr = [10^9]`, `k = 1`
Single element; the DP table has size 2, and only one window of width 1 is tried. Returns $10^9$ — maximum possible value for a single element.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [10^9, 1, 1, ..., 1]` (500 elements), `k = 500`
Placing $10^9$ into a window of 500 gives $10^9 \times 500 = 5 \times 10^{11}$, which exceeds a 32-bit integer. The problem constraints keep individual element values at most $10^9$ and array length at most $500$, so the return type must accommodate values up to $\approx 5 \times 10^{11}$ — confirming that a 64-bit integer is needed in practice even though the signature uses int.
