# LeetCode #1048: Longest String Chain

https://leetcode.com/problems/longest-string-chain/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2 \cdot L)$ | $O(n)$ |
| **Optimal: DP with Sort + Hash Map ★** | $O(n \cdot L^2)$ | $O(n \cdot L)$ |

---

## Understanding the Methods

### Brute Force
Sort words by length, then for each word check every shorter word to see if it can be a predecessor. Checking takes $O(L)$ per pair.

### Optimal: DP with Sort + Hash Map ★
Sort words by length. For each word, generate all possible predecessors by removing one character at a time ($L$ attempts, each $O(L)$), look them up in a hash map, and take the maximum chain length. Store each word's chain length in the map.

**Constraints:**
* $1 \leq words.length \leq 1000$
* $1 \leq words[i].length \leq 16$
* $words[i]$ consists of lowercase English letters

## Solutions

### C#

```
public class Solution {
    public int LongestStrChain(string[] words) {
        // Sort by length so shorter predecessors are always processed first
        System.Array.Sort(words, (a, b) => a.Length - b.Length);
        var chainLen = new Dictionary<string, int>();
        int result = 1;

        foreach (string word in words) {
            chainLen[word] = 1;
            // Try every word formed by deleting one character — if it exists, extend its chain
            for (int i = 0; i < word.Length; i++) {
                string predecessor = word.Substring(0, i) + word.Substring(i + 1);
                if (chainLen.TryGetValue(predecessor, out int prevLen)) {
                    chainLen[word] = Math.Max(chainLen[word], prevLen + 1);
                }
            }
            result = Math.Max(result, chainLen[word]);
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def longest_str_chain(self, words: list[str]) -> int:
        # Sort by length so shorter predecessors are always processed first
        words.sort(key=len)
        chain_len: dict[str, int] = {}
        result = 1

        for word in words:
            chain_len[word] = 1
            # Try every word formed by deleting one character — if it exists, extend its chain
            for i in range(len(word)):
                predecessor = word[:i] + word[i+1:]
                if predecessor in chain_len:
                    chain_len[word] = max(chain_len[word], chain_len[predecessor] + 1)
            result = max(result, chain_len[word])

        return result
```

### Go

```
import "sort"

func longestStrChain(words []string) int {
    // Sort by length so shorter predecessors are always processed first
    sort.Slice(words, func(i, j int) bool { return len(words[i]) < len(words[j]) })
    chainLen := make(map[string]int)
    result := 1

    for _, word := range words {
        chainLen[word] = 1
        // Try every word formed by deleting one character — if it exists, extend its chain
        for i := 0; i < len(word); i++ {
            predecessor := word[:i] + word[i+1:]
            if prev, ok := chainLen[predecessor]; ok && prev+1 > chainLen[word] {
                chainLen[word] = prev + 1
            }
        }
        if chainLen[word] > result {
            result = chainLen[word]
        }
    }
    return result
}
```

### Rust

```
use std::collections::HashMap;

impl Solution {
    pub fn longest_str_chain(mut words: Vec<String>) -> i32 {
        // Sort by length so shorter predecessors are always processed first
        words.sort_by_key(|w| w.len());
        let mut chain_len: HashMap<String, i32> = HashMap::new();
        let mut result = 1;

        for word in &words {
            let mut best = 1i32;
            // Try every word formed by deleting one character — if it exists, extend its chain
            for i in 0..word.len() {
                let predecessor = format!("{}{}", &word[..i], &word[i+1..]);
                if let Some(&prev) = chain_len.get(&predecessor) {
                    best = best.max(prev + 1);
                }
            }
            chain_len.insert(word.clone(), best);
            result = result.max(best);
        }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `words = ["a","b","ba","bca","bda","bdca"]`
Sorted by length: `a, b, ba, bca, bda, bdca`. Chain `a→ba→bda→bdca` has length 4, which is the answer.

### 2. Slightly Complex
**Input:** `words = ["xbc","pcxbcf","xb","cxbc","pcxbc"]`
The longest chain is `xb→xbc→cxbc→pcxbc→pcxbcf` (length 5). Each step inserts exactly one character.

### 3. Edge Case: Time Factor
**Input:** 1000 words, each of length 16
For each word, $L = 16$ predecessors are generated and each construction is $O(L) = O(16)$. Total: $1000 \times 16 \times 16 = 256{,}000$ operations — fast, but approaches the bound.

### 4. Edge Case: Space Factor
**Input:** All 1000 words are distinct and none is a predecessor of another
The hash map stores 1000 entries of average length $\leq 16$. No chain grows beyond 1; the map uses $O(n \cdot L)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `words = ["a","aa","aaa",...,"aaaa...a"]` (16 words, lengths 1–16)
A perfect chain of length 16. Each word adds exactly one 'a' to the previous. The DP builds the chain length incrementally: word of length $k$ gets chain length $k$.
