# LeetCode #1049: Last Stone Weight II

https://leetcode.com/problems/last-stone-weight-ii/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^n)$ | $O(n)$ |
| **Optimal: 0/1 Knapsack DP ★** | $O(n \cdot S)$ | $O(S)$ |

---

## Understanding the Methods

### Brute Force
Assign each stone a sign ($+$ or $-$) and enumerate all $2^n$ combinations. The answer is the minimum absolute sum — too slow for large inputs.

### Optimal: 0/1 Knapsack DP ★
The problem reduces to: partition stones into two groups minimising $|S_1 - S_2|$, which equals $|total - 2 \cdot S_1|$. Maximise $S_1 \leq total/2$ using a boolean knapsack DP bit-set. The answer is $total - 2 \cdot \max(S_1)$.

**Constraints:**
* $1 \leq stones.length \leq 30$
* $1 \leq stones[i] \leq 100$

## Solutions

### C#

```
public class Solution {
    public int LastStoneWeightII(int[] stones) {
        int total = 0;
        foreach (int s in stones) total += s;
        int target = total / 2;

        // dp[w] = true if we can form subset sum exactly w
        bool[] dp = new bool[target + 1];
        dp[0] = true;

        foreach (int stone in stones) {
            // Traverse backwards to avoid using the same stone twice
            for (int w = target; w >= stone; w--) {
                dp[w] = dp[w] || dp[w - stone];
            }
        }

        // Find the largest achievable sum <= total/2
        for (int w = target; w >= 0; w--) {
            if (dp[w]) return total - 2 * w;
        }
        return 0; // unreachable
    }
}
```

### Python

```
class Solution:
    def last_stone_weight_ii(self, stones: list[int]) -> int:
        total = sum(stones)
        target = total // 2

        # dp[w] = True if subset sum w is achievable
        dp = [False] * (target + 1)
        dp[0] = True

        for stone in stones:
            # Traverse backwards to avoid using the same stone twice
            for w in range(target, stone - 1, -1):
                dp[w] = dp[w] or dp[w - stone]

        # Largest achievable sum <= total//2
        for w in range(target, -1, -1):
            if dp[w]:
                return total - 2 * w
        return 0
```

### Go

### Rust

```
impl Solution {
    pub fn last_stone_weight_ii(stones: Vec<i32>) -> i32 {
        let total: i32 = stones.iter().sum();
        let target = (total / 2) as usize;

        // dp[w] = true if subset sum w is achievable
        let mut dp = vec![false; target + 1];
        dp[0] = true;

        for &stone in &stones {
            // Traverse backwards to avoid using the same stone twice
            for w in (stone as usize..=target).rev() {
                dp[w] = dp[w] || dp[w - stone as usize];
            }
        }

        // Largest achievable sum <= total/2
        for w in (0..=target).rev() {
            if dp[w] { return total - 2 * w as i32; }
        }
        0
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `stones = [2,7,4,1,8,1]`
Total = 23, target = 11. The knapsack finds subset sum 11 is achievable ($7+4$ or $7+3$... actually $7+4=11$). Result: $23 - 2 \times 11 = 1$.

### 2. Slightly Complex
**Input:** `stones = [31,26,33,21,40]`
Total = 151, target = 75. The DP finds the largest achievable sum $\leq 75$, say 73 ($40+33$). Result: $151 - 2 \times 73 = 5$.

### 3. Edge Case: Time Factor
**Input:** 30 stones each of value 100, target = 1500
The DP iterates $30 \times 1500 = 45{,}000$ steps — the maximum workload. Every prefix sum is a multiple of 100, so the dp array is sparse but the loop still runs fully.

### 4. Edge Case: Space Factor
**Input:** `stones = [100,100,...,100]` (30 stones), target = 1500
The boolean dp array has $1501$ entries. All values divisible by 100 are reachable; the minimum difference is 0 (perfectly balanced groups of 15).

### 5. Almost-Impossible but Plausible
**Input:** `stones = [1,2,3,...,30]`
Total = 465, target = 232. The knapsack explores a rich space of subset sums. The best partition achieves difference 1 (since 465 is odd, the minimum possible is $\geq 1$).
