# LeetCode #1053: Previous Permutation With One Swap

https://leetcode.com/problems/previous-permutation-with-one-swap/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Right-to-Left Linear Scan ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Generate every possible single-swap permutation, filter those smaller than the original, and return the largest. $O(n^2)$ swaps to enumerate, $O(n)$ space to track the best candidate.

### Optimal: Right-to-Left Linear Scan ★
Scan from the right to find the rightmost position `i` where `arr[i] > arr[i+1]` (first descent). Then, among all positions `j > i` where `arr[j] < arr[i]`, pick the largest value (`best`); among ties, take the leftmost `j` so that the larger displaced element lands as early as possible. One swap at `(i, j)` yields the largest valid predecessor. If no descent exists the array is already the minimum permutation.

**Constraints:**
* $1 \leq \text{arr.length} \leq 10^4$
* $1 \leq \text{arr}[i] \leq 10^4$


## Solutions

### C#

```
public class Solution {
    public int[] PrevPermOpt1(int[] arr) {
        int n = arr.Length;
        // Scan from right to find the rightmost descent — the best position to decrease
        for (int i = n - 2; i >= 0; i--) {
            if (arr[i] <= arr[i + 1]) continue;
            // Find the largest value in the suffix that is strictly less than arr[i]
            int best = 0;
            for (int k = i + 1; k < n; k++)
                if (arr[k] < arr[i] && arr[k] > best) best = arr[k];
            // Take the leftmost occurrence so the displaced larger value lands earliest
            for (int j = i + 1; j < n; j++) {
                if (arr[j] == best) {
                    (arr[i], arr[j]) = (arr[j], arr[i]);
                    return arr;
                }
            }
        }
        return arr; // Already the lexicographically smallest permutation
    }
}
```

### Python

```
class Solution:
    def prevPermOpt1(self, arr: list[int]) -> list[int]:
        n = len(arr)
        # Scan from right to find the rightmost descent — the best position to decrease
        for i in range(n - 2, -1, -1):
            if arr[i] <= arr[i + 1]:
                continue
            # Find the largest value in the suffix that is strictly less than arr[i]
            best = max(x for x in arr[i + 1:] if x < arr[i])
            # Take the leftmost occurrence so the displaced larger value lands earliest
            for j in range(i + 1, n):
                if arr[j] == best:
                    arr[i], arr[j] = arr[j], arr[i]
                    return arr
        return arr  # Already the lexicographically smallest permutation
```

### Rust

```
impl Solution {
    pub fn prev_perm_opt1(mut arr: Vec<i32>) -> Vec<i32> {
        let n = arr.len();
        // Scan from right to find the rightmost descent — the best position to decrease
        let mut i = n as i32 - 2;
        while i >= 0 {
            let idx = i as usize;
            if arr[idx] > arr[idx + 1] {
                // Find the largest value in the suffix that is strictly less than arr[i]
                let best = arr[idx + 1..].iter().filter(|&&x| x < arr[idx]).copied().max().unwrap();
                // Take the leftmost occurrence so the displaced larger value lands earliest
                for j in idx + 1..n {
                    if arr[j] == best {
                        arr.swap(idx, j);
                        return arr;
                    }
                }
            }
            i -= 1;
        }
        arr // Already the lexicographically smallest permutation
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [3, 2, 1]`
Scan from right: descent at `i=1` (`arr[1]=2 > arr[2]=1`). Best value in suffix $< 2$ is $1$, leftmost at `j=2`. Swap $\Rightarrow$ `[3, 1, 2]` — one position shifted, suffix maximised.

### 2. Slightly Complex
**Input:** `arr = [1, 1, 5]`
Scan from right: `arr[1]=1 < arr[2]=5`, no descent at `i=1`; `arr[0]=1 = arr[1]=1`, no descent at `i=0`. No valid swap exists — return `[1, 1, 5]` unchanged, confirming duplicate-prefix detection.

### 3. Edge Case: Time Factor
**Input:** `arr = [2, 1, 3, 4, 5]` (descent buried at far left)
Scan from right: positions $i=3,2,1$ have ascending pairs. Descent finally at `i=0` (`arr[0]=2 > arr[1]=1`). Best $= 1$, `j=1`. Swap $\Rightarrow$ `[1, 2, 3, 4, 5]`. Full $O(n)$ scan required — worst-case traversal.

### 4. Edge Case: Space Factor
**Input:** `arr = [1, 2, 3, 4, 5]` (sorted ascending)
No descent found anywhere — the array is already the lexicographic minimum. Returns as-is with $O(1)$ extra space; no candidate array is ever materialised.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [3, 1, 1, 3]`
Descent at `i=0` (`arr[0]=3 > arr[1]=1`). Both `arr[1]` and `arr[2]` equal 1 (best value). Taking the **leftmost** `j=1` swaps to `[1, 3, 1, 3]`, which is strictly larger than `[1, 1, 3, 3]` (the result of taking rightmost `j=2`). Duplicate handling is the crux.
