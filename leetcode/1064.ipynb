# LeetCode #1064: Fixed Point

https://leetcode.com/problems/fixed-point/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (Linear Scan)** | $O(n)$ | $O(1)$ |
| **Optimal: Binary Search ★** | $O(\log n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force (Linear Scan)
Iterate from left to right and return the first index `i` where `arr[i] == i`. Works but misses the sorted-distinct structure.

### Optimal: Binary Search ★
Because `arr` is sorted with distinct integers, `arr[i] - i` is strictly increasing. A fixed point exists iff `arr[i] - i == 0`. Binary search on this monotone function finds the smallest fixed-point index in $O(\log n)$.

**Constraints:**
* $1 \leq arr.length \leq 1000$
* $-10^9 \leq arr[i] \leq 10^9$
* `arr` is sorted in strictly increasing order.

## Solutions

### C#

```
public class Solution {
    public int FixedPoint(int[] arr) {
        int lo = 0, hi = arr.Length - 1, ans = -1;
        while (lo <= hi) {
            int mid = lo + (hi - lo) / 2;
            if (arr[mid] == mid) {
                // Found a fixed point; search left for the smallest one
                ans = mid; hi = mid - 1;
            } else if (arr[mid] < mid) {
                // arr[mid]-mid < 0: fixed points can only exist to the right
                lo = mid + 1;
            } else {
                // arr[mid]-mid > 0: no fixed point to the right of mid (values grow too fast)
                hi = mid - 1;
            }
        }
        return ans;
    }
}
```

### Rust

```
impl Solution {
    pub fn fixed_point(arr: Vec<i32>) -> i32 {
        let (mut lo, mut hi) = (0i32, arr.len() as i32 - 1);
        let mut ans = -1i32;
        while lo <= hi {
            let mid = lo + (hi - lo) / 2;
            if arr[mid as usize] == mid {
                // Found a fixed point; search left for the smallest one
                ans = mid; hi = mid - 1;
            } else if arr[mid as usize] < mid {
                // arr[mid]-mid < 0: fixed points can only exist to the right
                lo = mid + 1;
            } else {
                // arr[mid]-mid > 0: no fixed point exists to the right
                hi = mid - 1;
            }
        }
        ans
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [-10, -5, 0, 3, 7]`
At index 3, `arr[3] = 3` — a fixed point. Binary search narrows to it in $\lceil \log_2 5 \rceil = 3$ steps. Answer: **3**.

### 2. Slightly Complex
**Input:** `arr = [0, 2, 5, 8, 17]`
Index 0 has `arr[0] = 0`, a fixed point. The first midpoint check may land at index 2 (value 5, `arr[mid] > mid`), driving the search left until it finds 0. Answer: **0**.

### 3. Edge Case: Time Factor
**Input:** `arr` of length 1000 with no fixed point (`arr[i] = i + 1` for all `i`).
`arr[mid] - mid = 1 > 0` at every midpoint, so `hi` is halved each step. Binary search exhausts all $\lceil \log_2 1000 \rceil = 10$ iterations without finding a match. Answer: **-1**.

### 4. Edge Case: Space Factor
**Input:** Single-element array `arr = [0]`.
Only one comparison needed: `arr[0] = 0 = 0`. $O(1)$ space and one step. Answer: **0**.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [-10^9, -10^9+1, ..., 0, ..., 10^9-2]` — 1000 elements where `arr[500] = 500`.
The binary search must handle $i$ as a 32-bit int and `arr[i]` as a potentially large negative or positive value; `arr[mid] - mid` could overflow in languages without 64-bit arithmetic. The implementation avoids subtraction by comparing `arr[mid]` directly to `mid`, sidestepping any overflow.
