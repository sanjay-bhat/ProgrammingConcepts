# LeetCode #1067: Digit Count in Range

https://leetcode.com/problems/digit-count-in-range/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O((high - low) \times \log high)$ | $O(1)$ |
| **Optimal: Digit DP (Place-Value Counting) ★** | $O(\log high)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Iterate every integer from $low$ to $high$, convert each to a string, and count occurrences of digit $d$. Works for small ranges but fails for ranges up to $10^9$.

### Optimal: Digit DP (Place-Value Counting) ★
Count occurrences of digit $d$ in $[0..n]$ using a mathematical place-value formula, then apply the prefix technique: $f(high) - f(low - 1)$. For each decimal position (ones, tens, hundreds, ...) count how many complete cycles of that digit appear plus the partial cycle contribution.

**Why this is better than Brute Force:** Runs in $O(\log n)$ digit steps regardless of the numeric range, making it feasible for inputs up to $10^9$.

**Constraints:**
* $0 \le d \le 9$
* $1 \le low \le high \le 2 \times 10^8$

## Solutions

### C#

```
public class Solution {
    public int DigitCountInRange(int d, int low, int high) {
        return Count(d, high) - Count(d, low - 1);
    }
    // Count occurrences of digit d in [0..n]
    private int Count(int d, int n) {
        if (n < 0) return 0;
        long result = 0, factor = 1;
        while (factor <= n) {
            long hi = n / factor;          // digits to the left of current position
            long lo = n % factor;          // digits to the right of current position
            long digit = hi % 10;          // the digit at the current position
            hi /= 10;
            if (d == 0) {
                // Leading-zero correction: subtract cases where the number has fewer digits
                result += (hi - 1) * factor + (lo + 1);
            } else if (digit > d) {
                // All complete cycles plus one extra full block
                result += (hi + 1) * factor;
            } else if (digit == d) {
                // Partial block for the current cycle
                result += hi * factor + lo + 1;
            } else {
                // Only complete cycles contribute
                result += hi * factor;
            }
            factor *= 10;
        }
        return (int)result;
    }
}
```

### Python

```
class Solution:
    def digit_count_in_range(self, d: int, low: int, high: int) -> int:
        def count(n: int) -> int:
            if n < 0:
                return 0
            result, factor = 0, 1
            while factor <= n:
                hi, lo = divmod(n, factor)
                digit = hi % 10
                hi //= 10
                if d == 0:
                    # Leading-zero correction: subtract cases where the number has fewer digits
                    result += (hi - 1) * factor + (lo + 1)
                elif digit > d:
                    # All complete cycles plus one extra full block
                    result += (hi + 1) * factor
                elif digit == d:
                    # Partial block for the current cycle
                    result += hi * factor + lo + 1
                else:
                    # Only complete cycles contribute
                    result += hi * factor
                factor *= 10
            return result
        return count(high) - count(low - 1)
```

### Go

```
func digitCountInRange(d int, low int, high int) int {
    var count func(n int) int
    count = func(n int) int {
        if n < 0 { return 0 }
        result, factor := 0, 1
        for factor <= n {
            hi, lo := n/factor, n%factor
            digit := hi % 10
            hi /= 10
            if d == 0 {
                // Leading-zero correction: subtract cases where number has fewer digits
                result += (hi-1)*factor + (lo + 1)
            } else if digit > d {
                // All complete cycles plus one extra full block
                result += (hi + 1) * factor
            } else if digit == d {
                // Partial block for the current cycle
                result += hi*factor + lo + 1
            } else {
                // Only complete cycles contribute
                result += hi * factor
            }
            factor *= 10
        }
        return result
    }
    return count(high) - count(low-1)
}
```

### Rust

```
impl Solution {
    pub fn digit_count_in_range(d: i32, low: i32, high: i32) -> i32 {
        fn count(d: i64, n: i64) -> i64 {
            if n < 0 { return 0; }
            let (mut result, mut factor) = (0i64, 1i64);
            while factor <= n {
                let (hi_full, lo) = (n / factor, n % factor);
                let digit = hi_full % 10;
                let hi = hi_full / 10;
                if d == 0 {
                    // Leading-zero correction: subtract cases where number has fewer digits
                    result += (hi - 1) * factor + (lo + 1);
                } else if digit > d {
                    // All complete cycles plus one extra full block
                    result += (hi + 1) * factor;
                } else if digit == d {
                    // Partial block for the current cycle
                    result += hi * factor + lo + 1;
                } else {
                    // Only complete cycles contribute
                    result += hi * factor;
                }
                factor *= 10;
            }
            result
        }
        (count(d as i64, high as i64) - count(d as i64, low as i64 - 1)) as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `d = 1, low = 1, high = 13`
Digit 1 appears in: 1, 10, 11 (twice), 12, 13 — total **6**. The formula computes $f(13) - f(0)$ in two place-value steps.

### 2. Slightly Complex
**Input:** `d = 0, low = 10, high = 99`
Leading-zero correction prevents counting "0" in 01, 02, … as separate numbers. Only the tens-zero numbers (10, 20, …, 90) contribute one zero each via the units column — total **9**.

### 3. Edge Case: Time Factor
**Input:** `d = 9, low = 1, high = 200000000`
Even for $high \approx 2 \times 10^8$, the while-loop iterates only 9 times (one per decimal digit). The place-value decomposition makes runtime truly $O(\log_{10} high)$ regardless of range size.

### 4. Edge Case: Space Factor
**Input:** `d = 5, low = 1, high = 2 \times 10^8$`
No auxiliary arrays or hash maps are used — only integer arithmetic with three local variables per iteration. Space is $O(1)$ regardless of input magnitude.

### 5. Almost-Impossible but Plausible
**Input:** `d = 0, low = 1, high = 1`
Range contains only the number 1, which has no zero digit. Leading-zero correction makes $f(1) - f(0) = 0$. Answer: **0**.
