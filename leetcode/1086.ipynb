# LeetCode #1086: High Five

https://leetcode.com/problems/high-five/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Sort + Group ★** | $O(n \log n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each unique student ID, scan all items to collect their scores, sort them, and take the top-5 average. Rescanning for each student costs $O(n)$ per student and $O(n^2)$ overall.

### Optimal: Sort + Group ★
Sort all items by `(id, score DESC)`. Group-by then takes the first 5 scores per ID in a single linear pass, making the sort the bottleneck at $O(n \log n)$.

**Constraints:**
* $1 \leq items.length \leq 1000$
* `items[i].length == 2`
* $1 \leq items[i][0] \leq 1000$ (student ID)
* $0 \leq items[i][1] \leq 100$ (score)
* For each student, there are at least 5 scores.

## Solutions

### C#

```
using System.Collections.Generic;
using System.Linq;
public class Solution {
    public int[][] HighFive(int[][] items) {
        // Sort by id ascending, then by score descending to surface top scores first
        var sorted = items.OrderBy(x => x[0]).ThenByDescending(x => x[1]);
        var result = new List<int[]>();
        int curId = -1, count = 0, sum = 0;
        foreach (var item in sorted) {
            if (item[0] != curId) {
                // Flush the completed student's average before starting the next
                if (curId != -1) result.Add(new int[] { curId, sum / 5 });
                curId = item[0]; count = 0; sum = 0;
            }
            if (count < 5) { sum += item[1]; count++; }
        }
        if (curId != -1) result.Add(new int[] { curId, sum / 5 });
        return result.ToArray();
    }
}
```

### Python

```
class Solution:
    def high_five(self, items: list[list[int]]) -> list[list[int]]:
        # Sort by id ascending, then by score descending to surface top scores first
        items.sort(key=lambda x: (x[0], -x[1]))
        result = []
        cur_id, count, total = -1, 0, 0
        for student_id, score in items:
            if student_id != cur_id:
                # Flush the completed student's average before starting the next
                if cur_id != -1:
                    result.append([cur_id, total // 5])
                cur_id, count, total = student_id, 0, 0
            if count < 5:
                total += score; count += 1
        if cur_id != -1:
            result.append([cur_id, total // 5])
        return result
```

### Go

```
import "sort"
func highFive(items [][]int) [][]int {
    // Sort by id ascending, then by score descending to surface top scores first
    sort.Slice(items, func(i, j int) bool {
        if items[i][0] != items[j][0] { return items[i][0] < items[j][0] }
        return items[i][1] > items[j][1]
    })
    var result [][]int
    curID, count, sum := -1, 0, 0
    for _, item := range items {
        if item[0] != curID {
            // Flush the completed student's average before starting the next
            if curID != -1 { result = append(result, []int{curID, sum / 5}) }
            curID, count, sum = item[0], 0, 0
        }
        if count < 5 { sum += item[1]; count++ }
    }
    if curID != -1 { result = append(result, []int{curID, sum / 5}) }
    return result
}
```

### Rust

```
impl Solution {
    pub fn high_five(mut items: Vec<Vec<i32>>) -> Vec<Vec<i32>> {
        // Sort by id ascending, then by score descending to surface top scores first
        items.sort_by(|a, b| a[0].cmp(&b[0]).then(b[1].cmp(&a[1])));
        let mut result: Vec<Vec<i32>> = Vec::new();
        let (mut cur_id, mut count, mut sum) = (-1i32, 0i32, 0i32);
        for item in &items {
            if item[0] != cur_id {
                // Flush the completed student's average before starting the next
                if cur_id != -1 { result.push(vec![cur_id, sum / 5]); }
                cur_id = item[0]; count = 0; sum = 0;
            }
            if count < 5 { sum += item[1]; count += 1; }
        }
        if cur_id != -1 { result.push(vec![cur_id, sum / 5]); }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `items = [[1,91],[1,92],[2,93],[2,97],[1,60],[2,77],[1,65],[2,100],[1,87],[2,100]]`
Student 1's top-5 scores: 92, 91, 87, 65, 60 → sum 395 → average 79. Student 2's top-5: 100, 100, 97, 93, 77 → sum 467 → average 93. Answer: `[[1,87],[2,93]]`.

### 2. Slightly Complex
**Input:** Student has exactly 5 scores and they are all the same value (e.g., 80).
After sorting, all 5 scores are consumed. `sum/5 = 80`. No partial-count edge case; the flush-on-next-id path executes cleanly.

### 3. Edge Case: Time Factor
**Input:** 1000 items split across 200 students, each with exactly 5 scores.
The sort processes all 1000 items in $O(1000 \log 1000) \approx 10{,}000$ comparisons. The linear pass then reads exactly $200 \times 5 = 1000$ items with no skips.

### 4. Edge Case: Space Factor
**Input:** 1000 items for a single student (only 5 of which count).
After sorting, the first 5 items are consumed and the remaining 995 are skipped. The result list holds one entry: $O(1)$ output despite $O(n)$ input.

### 5. Almost-Impossible but Plausible
**Input:** Items interleaved with IDs 1 and 1000 alternating, all scores equal to 100.
The sort separates them cleanly by ID. Both students' top-5 sums are 500, averages are 100. The boundary between the two groups is detected correctly because $1000 \neq 1$ at the flush check.
