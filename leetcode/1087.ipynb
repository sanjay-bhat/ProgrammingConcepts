# LeetCode #1087: Design Tic-Tac-Toe

https://leetcode.com/problems/design-tic-tac-toe/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n)$ per move | $O(n^2)$ |
| **Optimal: Counter Arrays ★** | $O(1)$ per move | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Store the full $n \times n$ board. After each move, scan the affected row, column, and both diagonals to check for a winner. $O(n)$ per move, $O(n^2)$ space.

### Optimal: Counter Arrays ★
Instead of tracking individual cells, maintain an integer counter per row, per column, and one each for the two diagonals. Player 1 adds $+1$; player 2 adds $-1$. A win is detected instantly when any counter reaches $+n$ (player 1) or $-n$ (player 2). $O(1)$ per move, $O(n)$ total space.

**Why this is better than Brute Force:** No board traversal is ever needed — a single counter comparison replaces the $O(n)$ scan, cutting per-move cost from linear to constant.

**Constraints:**
* $2 \leq n \leq 100$
* Player is either 1 or 2
* $1 \leq \text{row}, \text{col} \leq n$
* $(\text{row}, \text{col})$ is guaranteed to be an empty cell
* At most $n^2$ calls will be made to `move`


## Solutions

### C#

```
public class TicTacToe {
    private int[] rows, cols;
    private int diag, antiDiag, n;

    public TicTacToe(int n) {
        this.n = n;
        // One counter per row and column tracks the running player balance
        rows = new int[n];
        cols = new int[n];
    }

    public int Move(int row, int col, int player) {
        // Map player 1 to +1 and player 2 to -1 to fold both into one set of counters
        int delta = player == 1 ? 1 : -1;
        int win   = player == 1 ? n : -n;
        rows[row] += delta;
        cols[col] += delta;
        if (row == col)         diag     += delta; // top-left to bottom-right diagonal
        if (row + col == n - 1) antiDiag += delta; // top-right to bottom-left diagonal
        if (rows[row] == win || cols[col] == win ||
            diag == win || antiDiag == win)
            return player;
        return 0; // No winner yet
    }
}
```

### Python

```
class TicTacToe:
    def __init__(self, n: int):
        self.n = n
        # One counter per row and column tracks the running player balance
        self.rows = [0] * n
        self.cols = [0] * n
        self.diag = self.anti_diag = 0

    def move(self, row: int, col: int, player: int) -> int:
        # Map player 1 to +1 and player 2 to -1 to fold both into one set of counters
        delta = 1 if player == 1 else -1
        win = self.n if player == 1 else -self.n
        self.rows[row] += delta
        self.cols[col] += delta
        if row == col:                self.diag      += delta  # top-left to bottom-right
        if row + col == self.n - 1:   self.anti_diag += delta  # top-right to bottom-left
        if (self.rows[row] == win or self.cols[col] == win or
                self.diag == win or self.anti_diag == win):
            return player
        return 0  # No winner yet
```

### Go

```
type TicTacToe struct {
    rows, cols       []int
    diag, antiDiag  int
    n                int
}

func Constructor(n int) TicTacToe {
    // One counter per row and column tracks the running player balance
    return TicTacToe{rows: make([]int, n), cols: make([]int, n), n: n}
}

func (t *TicTacToe) Move(row, col, player int) int {
    // Map player 1 to +1 and player 2 to -1 to fold both into one set of counters
    delta, win := 1, t.n
    if player == 2 { delta, win = -1, -t.n }
    t.rows[row] += delta
    t.cols[col] += delta
    if row == col        { t.diag     += delta } // top-left to bottom-right diagonal
    if row+col == t.n-1 { t.antiDiag += delta } // top-right to bottom-left diagonal
    if t.rows[row] == win || t.cols[col] == win ||
        t.diag == win || t.antiDiag == win {
        return player
    }
    return 0 // No winner yet
}
```

### Rust

```
struct TicTacToe {
    rows: Vec<i32>,
    cols: Vec<i32>,
    diag: i32,
    anti_diag: i32,
    n: i32,
}

impl TicTacToe {
    fn new(n: i32) -> Self {
        // One counter per row and column tracks the running player balance
        TicTacToe {
            rows: vec![0; n as usize],
            cols: vec![0; n as usize],
            diag: 0,
            anti_diag: 0,
            n,
        }
    }

    fn make_move(&mut self, row: i32, col: i32, player: i32) -> i32 {
        // Map player 1 to +1 and player 2 to -1 to fold both into one set of counters
        let (delta, win) = if player == 1 { (1, self.n) } else { (-1, -self.n) };
        self.rows[row as usize] += delta;
        self.cols[col as usize] += delta;
        if row == col            { self.diag      += delta; } // top-left to bottom-right
        if row + col == self.n-1 { self.anti_diag += delta; } // top-right to bottom-left
        if self.rows[row as usize] == win || self.cols[col as usize] == win ||
           self.diag == win || self.anti_diag == win {
            return player;
        }
        0 // No winner yet
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n=3`, moves: `(0,0,1)`, `(0,1,2)`, `(0,2,1)`, `(1,0,2)`, `(1,1,1)`, `(2,2,1)`
Player 1 completes the main diagonal (`rows[0]=1`, `rows[1]=1`, `diag=3` after move 6). `diag == n` signals the win — no board scan needed.

### 2. Slightly Complex
**Input:** `n=3`, moves interleaved so player 2 fills column 1: `(0,1,2)`, `(1,1,2)`, `(2,1,2)`
After the third move `cols[1] == -3 == -n`, returning player 2 immediately. Counter arrays catch column wins just as fast as row wins.

### 3. Edge Case: Time Factor
**Input:** `n=100`, all moves go through the center anti-diagonal cells until the last slot
Every move is $O(1)$; after exactly $100$ moves on the anti-diagonal `antiDiag == \pm n`. No matter how many cells were filled elsewhere, detection is instant — the counter approach stays constant regardless of board size.

### 4. Edge Case: Space Factor
**Input:** `n=100`, both players alternate without either winning; all $10{,}000$ cells filled
Total extra space is $O(n) = 200$ integers for rows and columns plus 2 scalars — a $50\times$ reduction versus storing the full $100 \times 100$ board.

### 5. Almost-Impossible but Plausible
**Input:** `n=3`, player 2 fills anti-diagonal exactly as player 1 fills main diagonal simultaneously
Both diagonals hit $\pm n$ in the same move sequence. The check fires as soon as either reaches the threshold, so whichever player completes a line first is returned correctly — no tie scenario is possible since only one player moves at a time.
