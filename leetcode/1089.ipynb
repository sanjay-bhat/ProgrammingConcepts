# LeetCode #1089: Duplicate Zeros

https://leetcode.com/problems/duplicate-zeros/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n)$ | $O(n)$ |
| **Optimal: Two-Pass In-Place ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Allocate a second buffer, iterate forward appending each non-zero element once and each zero twice, then copy the buffer back into `arr`. Correct but requires $O(n)$ extra space.

### Optimal: Two-Pass In-Place ★
First pass counts zeros to compute the total right-shift. Second pass fills from right to left using a virtual extended index: each element is written at its new position, and zeros write an additional duplicate one slot below. No extra allocation needed.

**Constraints:**
* $1 \leq \text{arr.length} \leq 10000$
* $0 \leq \text{arr}[i] \leq 9$


## Solutions

### C#

```
public class Solution {
    public void DuplicateZeros(int[] arr) {
        int n = arr.Length;
        // Count zeros to determine total right-shift magnitude
        int zeros = 0;
        foreach (int x in arr) if (x == 0) zeros++;

        // Two-pointer fill from right; virtual index j accounts for duplicates
        int i = n - 1, j = n + zeros - 1;
        while (i >= 0) {
            // Write the element at its shifted position if it falls within bounds
            if (j < n) arr[j] = arr[i];
            // A zero occupies two consecutive slots; write the duplicate below
            if (arr[i] == 0) {
                j--;
                if (j < n) arr[j] = 0;
            }
            i--;
            j--;
        }
    }
}
```

### Go

```
func duplicateZeros(arr []int) {
    n := len(arr)
    // Count zeros to determine total right-shift magnitude
    zeros := 0
    for _, x := range arr {
        if x == 0 {
            zeros++
        }
    }

    // Two-pointer fill from right; virtual index j accounts for duplicates
    i, j := n-1, n+zeros-1
    for i >= 0 {
        // Write the element at its shifted position if it falls within bounds
        if j < n {
            arr[j] = arr[i]
        }
        // A zero occupies two consecutive slots; write the duplicate below
        if arr[i] == 0 {
            j--
            if j < n {
                arr[j] = 0
            }
        }
        i--
        j--
    }
}
```

### Rust

```
impl Solution {
    pub fn duplicate_zeros(arr: &mut Vec<i32>) {
        let n = arr.len();
        // Count zeros to determine total right-shift magnitude
        let zeros = arr.iter().filter(|&&x| x == 0).count();

        // Two-pointer fill from right; virtual index j accounts for duplicates
        let (mut i, mut j) = (n as i64 - 1, (n + zeros) as i64 - 1);
        while i >= 0 {
            // Write the element at its shifted position if it falls within bounds
            if j < n as i64 {
                arr[j as usize] = arr[i as usize];
            }
            // A zero occupies two consecutive slots; write the duplicate below
            if arr[i as usize] == 0 {
                j -= 1;
                if j < n as i64 {
                    arr[j as usize] = 0;
                }
            }
            i -= 1;
            j -= 1;
        }
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [1, 0, 2, 3, 0, 4, 5, 0]`
The two interior zeros shift right-hand elements out; the backward fill produces `[1, 0, 0, 2, 3, 0, 0, 4]` — the original trailing `5, 0` are displaced past the array boundary.

### 2. Slightly Complex
**Input:** `arr = [0, 0, 0, 0, 0]`
All five elements are zeros so the virtual length is 10; every slot in the real array is overwritten with 0, giving an identical-looking `[0, 0, 0, 0, 0]` — but five extra duplicates were silently discarded.

### 3. Edge Case: Time Factor
**Input:** `arr = [1, 2, 3, 4, 9, 9, 9, 0]`
The single zero sits at the very end; the backward pass must scan all $n = 8$ positions before it encounters it. The duplicate falls past index 7 and is dropped, so the array is unchanged.

### 4. Edge Case: Space Factor
**Input:** `arr = [1, 2, 3]`
No zeros at all: `zeros = 0`, so `j = i` throughout the loop. Every write is a no-op self-assignment. Confirms $O(1)$ extra space — nothing is allocated regardless of array size.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [0, 0, 8, 0, 8, 9, 8, 0]` ($n = 8$, four zeros)
Virtual length is 12; only positions 0–7 survive. The backward fill correctly handles all four zeros and their duplicates — even the first zero, whose duplicate lands at index 1 — without an off-by-one. Result: `[0, 0, 0, 0, 8, 0, 0, 8]`.
