# LeetCode #1090: Largest Element in an Array

https://leetcode.com/problems/largest-element-in-an-array/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n \log n)$ | $O(1)$ |
| **Optimal: Single Pass ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Sort the array, then return the last element. Correct but wasteful — sorting rearranges data and costs $O(n \log n)$ when all that is needed is one scan.

### Optimal: Single Pass ★
Initialize the maximum to the first element, then scan once from left to right updating the maximum whenever a larger value is found. Touches each element exactly once for $O(n)$ time and uses only one extra variable.

**Constraints:**
* $1 \leq \text{arr.length} \leq 10^5$
* $-10^9 \leq \text{arr}[i] \leq 10^9$


## Solutions

### C#

```
public class Solution {
    public int LargestElement(int[] arr) {
        // Initialize to the first element so comparisons are always valid
        int max = arr[0];
        // Update the running maximum whenever a strictly larger value is found
        for (int i = 1; i < arr.Length; i++)
            if (arr[i] > max) max = arr[i];
        return max;
    }
}
```

### Python

```
class Solution:
    def largest_element(self, arr: list[int]) -> int:
        # Python's built-in max performs a single linear scan internally
        return max(arr)
```

### Go

```
func largestElement(arr []int) int {
    // Initialize to the first element so comparisons are always valid
    max := arr[0]
    // Update the running maximum whenever a strictly larger value is found
    for _, x := range arr[1:] {
        if x > max {
            max = x
        }
    }
    return max
}
```

### Rust

```
impl Solution {
    pub fn largest_element(arr: Vec<i32>) -> i32 {
        // Iterator max performs a single linear scan and returns the greatest value
        *arr.iter().max().unwrap()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [3, 1, 4, 1, 5, 9, 2, 6]`
The running max updates at 4, 5, then 9; returns 9 after one full pass over 8 elements.

### 2. Slightly Complex
**Input:** `arr = [-5, -3, -8, -1, -4]`
All negative; max initializes to $-5$ and updates at $-3$ then $-1$. Correctly returns $-1$ — no special zero handling required.

### 3. Edge Case: Time Factor
**Input:** `arr = [1, 2, 3, 4, 5, 6, 7, 100]`
The maximum is the very last element; every comparison in the loop triggers an update, forcing the full $n - 1$ iterations. Worst case for time.

### 4. Edge Case: Space Factor
**Input:** `arr = [42]`
Single element; the loop body never executes. Returns 42 using only the one extra variable `max`, confirming $O(1)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [1000000000, 999999999, 1000000000]`
Two elements tie at $10^9$; the first occurrence sets the maximum and the second comparison ($= \text{max}$, not $> \text{max}$) is ignored. Returns $10^9$ correctly without overflow since $10^9 < 2^{31} - 1 \approx 2.1 \times 10^9$.
