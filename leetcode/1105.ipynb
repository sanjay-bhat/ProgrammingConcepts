# LeetCode #1105: Filling Bookcase Shelves

https://leetcode.com/problems/filling-bookcase-shelves/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^n)$ | $O(n)$ |
| **Optimal: 1-D DP ★** | $O(n^2)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Try every subset of books to place on a new shelf, recursing on the remainder. Exponential in the number of books.

### Optimal: 1-D DP ★
`dp[i]` = minimum height of shelves needed to place the first `i` books. For each book `i`, try every contiguous suffix of books ending at `i` that fits on one shelf (width constraint). The height of that shelf is the maximum thickness in the suffix.

**Constraints:**
* $1 \leq books.length \leq 1000$
* $1 \leq books[i][0] \leq shelfWidth \leq 1000$
* $1 \leq books[i][1] \leq 1000$

## Solutions

### C#

```
public class Solution {
    public int MinHeightShelves(int[][] books, int shelfWidth) {
        int n = books.Length;
        int[] dp = new int[n + 1];
        dp[0] = 0;
        for (int i = 1; i <= n; i++) dp[i] = int.MaxValue;

        for (int i = 1; i <= n; i++) {
            int width = 0, maxHeight = 0;
            // Try placing books[j-1..i-1] on the same shelf (scanning backwards)
            for (int j = i; j >= 1; j--) {
                width += books[j - 1][0];
                if (width > shelfWidth) break; // shelf is full — stop extending
                // Track the tallest book on this shelf
                maxHeight = Math.Max(maxHeight, books[j - 1][1]);
                // Best option: books before j on previous shelves + this shelf's height
                dp[i] = Math.Min(dp[i], dp[j - 1] + maxHeight);
            }
        }
        return dp[n];
    }
}
```

### Python

```
class Solution:
    def min_height_shelves(self, books: list[list[int]], shelf_width: int) -> int:
        n = len(books)
        dp = [float('inf')] * (n + 1)
        dp[0] = 0

        for i in range(1, n + 1):
            width = 0
            max_height = 0
            # Try placing books[j-1..i-1] on the same shelf (scanning backwards)
            for j in range(i, 0, -1):
                width += books[j - 1][0]
                if width > shelf_width:
                    break  # shelf is full — stop extending
                # Track the tallest book on this shelf
                max_height = max(max_height, books[j - 1][1])
                # Best option: books before j on previous shelves + this shelf's height
                dp[i] = min(dp[i], dp[j - 1] + max_height)

        return dp[n]
```

### Go

```
func minHeightShelves(books [][]int, shelfWidth int) int {
    n := len(books)
    dp := make([]int, n+1)
    for i := 1; i <= n; i++ { dp[i] = 1<<31 - 1 }

    for i := 1; i <= n; i++ {
        width, maxHeight := 0, 0
        // Try placing books[j-1..i-1] on the same shelf (scanning backwards)
        for j := i; j >= 1; j-- {
            width += books[j-1][0]
            if width > shelfWidth { break }
            // Track the tallest book on this shelf
            if books[j-1][1] > maxHeight { maxHeight = books[j-1][1] }
            // Best option: books before j on previous shelves + this shelf's height
            if dp[j-1]+maxHeight < dp[i] { dp[i] = dp[j-1] + maxHeight }
        }
    }
    return dp[n]
}
```

### Rust

```
impl Solution {
    pub fn min_height_shelves(books: Vec<Vec<i32>>, shelf_width: i32) -> i32 {
        let n = books.len();
        let mut dp = vec![i32::MAX; n + 1];
        dp[0] = 0;

        for i in 1..=n {
            let (mut width, mut max_height) = (0, 0);
            // Try placing books[j-1..i-1] on the same shelf (scanning backwards)
            for j in (1..=i).rev() {
                width += books[j - 1][0];
                if width > shelf_width { break; }
                // Track the tallest book on this shelf
                max_height = max_height.max(books[j - 1][1]);
                // Best option: books before j on previous shelves + this shelf's height
                if dp[j - 1] != i32::MAX {
                    dp[i] = dp[i].min(dp[j - 1] + max_height);
                }
            }
        }
        dp[n]
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `books = [[1,1],[2,3],[2,3],[1,1],[1,1],[1,1],[1,2]]`, `shelfWidth = 4`
Optimal packing: shelf 1 holds books 1–3 (width 5 → too wide; shelf 1 holds 1,2: width=3, height=3); continuing with the DP finds the minimum total height of $6$.

### 2. Slightly Complex
**Input:** `books = [[1,3],[2,4],[3,2]]`, `shelfWidth = 6`
Placing all three on one shelf: width $1+2+3=6 \leq 6$, height $= \max(3,4,2) = 4$. Splitting book 3 alone: $3 + 2 = 5 > 4$. All together is optimal at $4$.

### 3. Edge Case: Time Factor
**Input:** 1000 books, each width 1, `shelfWidth = 1000`
All 1000 books fit on one shelf. The inner loop for the last book scans all 1000 predecessors — $O(n^2) = 10^6$ iterations total, the worst case.

### 4. Edge Case: Space Factor
**Input:** 1000 books with varying widths
The dp array has 1001 entries. No additional data structures are used — $O(n)$ space is the only allocation.

### 5. Almost-Impossible but Plausible
**Input:** `books = [[1000,1000]]`, `shelfWidth = 1000`
A single book exactly fills the shelf. The inner loop runs once: width = 1000, max_height = 1000, dp[1] = dp[0] + 1000 = 1000.
