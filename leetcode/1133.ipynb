# LeetCode #1133: Largest Unique Number

https://leetcode.com/problems/largest-unique-number/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Frequency Count ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each element, scan the rest of the array to check for duplicates, then track the largest singleton found. $O(n^2)$ comparisons make this impractical for large inputs.

### Optimal: Frequency Count ★
Count each number's occurrences in a single pass using a dictionary (or fixed-size array since values are bounded). Then scan all keys descending to return the first with count 1.

**Why this is better than Brute Force:** Reduces from $O(n^2)$ to $O(n)$ by trading $O(n)$ space for a linear frequency table instead of nested scanning.

**Constraints:**
* $1 \leq n \leq 2000$
* $0 \leq \text{nums}[i] \leq 1000$


## Solutions

### C#

```
public class Solution {
    public int LargestUniqueNumber(int[] nums) {
        // Count how many times each value appears
        var freq = new Dictionary<int, int>();
        foreach (int n in nums)
            freq[n] = freq.GetValueOrDefault(n, 0) + 1;

        // Scan for the largest value with exactly one occurrence
        int best = -1;
        foreach (var kv in freq)
            if (kv.Value == 1 && kv.Key > best)
                best = kv.Key;

        return best;
    }
}
```

### Python

```
from collections import Counter

class Solution:
    def largestUniqueNumber(self, nums: list[int]) -> int:
        # Build a frequency map in a single pass
        freq = Counter(nums)

        # Return the largest key that appears exactly once, or -1 if none exists
        return max((n for n, cnt in freq.items() if cnt == 1), default=-1)
```

### Go

```
func largestUniqueNumber(nums []int) int {
    // Tally occurrences of each value
    freq := map[int]int{}
    for _, n := range nums {
        freq[n]++
    }

    // Find the largest singleton
    best := -1
    for val, cnt := range freq {
        if cnt == 1 && val > best {
            best = val
        }
    }
    return best
}
```

### Rust

```
use std::collections::HashMap;

impl Solution {
    pub fn largest_unique_number(nums: Vec<i32>) -> i32 {
        // Count occurrences of each number in one pass
        let mut freq: HashMap<i32, i32> = HashMap::new();
        for &n in &nums {
            *freq.entry(n).or_insert(0) += 1;
        }

        // Find the largest key whose count is exactly one
        freq.into_iter()
            .filter(|&(_, cnt)| cnt == 1)
            .map(|(val, _)| val)
            .max()
            .unwrap_or(-1)
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [5, 7, 3, 9, 4, 9, 8, 3, 1]`
Frequency map: `{5:1, 7:1, 3:2, 9:2, 4:1, 8:1, 1:1}`. Singletons: 5, 7, 4, 8, 1. Largest = **8**.

### 2. Slightly Complex
**Input:** `nums = [9, 9, 8, 8]`
All values appear twice. No singletons exist → return **-1**. The max-filter over an empty sequence returns the default.

### 3. Edge Case: Time Factor
**Input:** `nums = [0, 1, 2, ..., 999, 1000]` ($n = 1001$, all distinct)
Every element appears once; the frequency map has 1001 entries. Scanning all entries to find max takes $O(n)$. Largest = **1000**.

### 4. Edge Case: Space Factor
**Input:** `nums = [0, 0, 0, ..., 0]` ($n = 2000$ zeros)
Frequency map has exactly one entry `{0: 2000}`. Count is not 1, so result = **-1**. Map size is $O(1)$ (one key), well within $O(n)$ bound.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [1000, 1000, 999]` (near-max values, one singleton buried)
Frequencies: `{1000:2, 999:1}`. Only 999 qualifies → return **999**. Confirms the algorithm correctly ignores the larger but non-unique value.
