# LeetCode #1137: N-th Tribonacci Number

https://leetcode.com/problems/nth-tribonacci-number/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(3^n)$ | $O(n)$ |
| **Optimal: Iterative Rolling Variables ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Recursively compute $T(n) = T(n-1) + T(n-2) + T(n-3)$ without memoization. Each call spawns three sub-calls, leading to exponential work and massive redundant recomputation.

### Optimal: Iterative Rolling Variables ★
Keep only the last three values in three variables and slide them forward one step at a time. No array or recursion stack needed — constant space, linear time.

**Constraints:**
* $0 \leq n \leq 37$
* The answer is guaranteed to fit within a 32-bit integer.

## Solutions

### C#

```
public class Solution {
    public int Tribonacci(int n) {
        // Seed the three base cases directly
        if (n == 0) return 0;
        if (n <= 2) return 1;
        int a = 0, b = 1, c = 1;
        // Slide the window forward, discarding the oldest value each step
        for (int i = 3; i <= n; i++) {
            int next = a + b + c;
            a = b; b = c; c = next;
        }
        return c;
    }
}
```

### Python

```
class Solution:
    def tribonacci(self, n: int) -> int:
        # Seed the three base cases directly
        if n == 0: return 0
        if n <= 2: return 1
        a, b, c = 0, 1, 1
        # Slide the window forward, discarding the oldest value each step
        for _ in range(3, n + 1):
            a, b, c = b, c, a + b + c
        return c
```

### Go

```
func tribonacci(n int) int {
    // Seed the three base cases directly
    if n == 0 { return 0 }
    if n <= 2 { return 1 }
    a, b, c := 0, 1, 1
    // Slide the window forward, discarding the oldest value each step
    for i := 3; i <= n; i++ {
        a, b, c = b, c, a+b+c
    }
    return c
}
```

### Rust

```
impl Solution {
    pub fn tribonacci(n: i32) -> i32 {
        // Seed the three base cases directly
        if n == 0 { return 0; }
        if n <= 2 { return 1; }
        let (mut a, mut b, mut c) = (0i32, 1i32, 1i32);
        // Slide the window forward, discarding the oldest value each step
        for _ in 3..=n {
            let next = a + b + c;
            a = b; b = c; c = next;
        }
        c
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n = 4`
$T(4) = T(3) + T(2) + T(1) = 2 + 1 + 1 = 4$. The three variables slide four steps from the base seeds, accumulating the next value at each turn.

### 2. Slightly Complex
**Input:** `n = 25`
After 25 iterations the three-variable window accumulates to $1389537$. No intermediate array is allocated — each step simply overwrites the oldest slot.

### 3. Edge Case: Time Factor
**Input:** `n = 37`
Maximum input; the loop executes exactly 35 iterations (indices 3–37). The result $2082876103$ fits within a 32-bit signed integer ($< 2^{31} - 1 \approx 2.1 \times 10^9$), confirming the constraint guarantee.

### 4. Edge Case: Space Factor
**Input:** `n = 0`
Returns $0$ before the loop is entered, using exactly three local integer registers — the minimum possible space for any $n$.

### 5. Almost-Impossible but Plausible
**Input:** `n = 36`
Result is $1149851745$; adding the two prior terms gives $T(37) = 2082876103$. A naïve sum of $T(36) + T(35) + T(34)$ would be $\approx 2.08 \times 10^9$, just below the 32-bit overflow boundary — confirming the constraint that $n \leq 37$ is intentionally the last safe value.
