# LeetCode #1144: Decrease Elements To Make Array Zigzag

https://leetcode.com/problems/decrease-elements-to-make-array-zigzag/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Greedy Per-Parity ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Try all possible values for each element or simulate every valid zigzag pattern, recomputing costs. Quadratic comparisons make this inefficient.

### Optimal: Greedy Per-Parity ★
A zigzag can have either even-indexed elements as valleys or odd-indexed elements as valleys. For each parity choice, the only valid operation is to decrease valley candidates — we never need to increase. For each valley position `i`, the required decrement is `max(0, nums[i] - min(neighbors) + 1)`. Sum these for both parities and return the minimum.

**Constraints:**
* $1 \leq \text{nums.length} \leq 1000$
* $1 \leq \text{nums}[i] \leq 1000$

## Solutions

### C#

```
public class Solution {
    public int MovesToMakeZigzag(int[] nums) {
        int n = nums.Length;
        int[] ops = new int[2];
        for (int i = 0; i < n; i++) {
            // Find the smallest neighbor to determine how much nums[i] must shrink
            int left  = i > 0     ? nums[i - 1] : int.MaxValue;
            int right = i + 1 < n ? nums[i + 1] : int.MaxValue;
            int minNeighbor = Math.Min(left, right);
            // Valley must be strictly less than both neighbors; decrement only, never increase
            ops[i % 2] += Math.Max(0, nums[i] - minNeighbor + 1);
        }
        return Math.Min(ops[0], ops[1]);
    }
}
```

### Python

```
class Solution:
    def moves_to_make_zigzag(self, nums: list[int]) -> int:
        n = len(nums)
        ops = [0, 0]
        for i in range(n):
            # Find the smallest neighbor to determine how much nums[i] must shrink
            left  = nums[i - 1] if i > 0 else float('inf')
            right = nums[i + 1] if i + 1 < n else float('inf')
            min_neighbor = min(left, right)
            # Valley must be strictly less than both neighbors; decrement only, never increase
            ops[i % 2] += max(0, nums[i] - min_neighbor + 1)
        return min(ops)
```

### Go

```
func movesToMakeZigzag(nums []int) int {
    n := len(nums)
    ops := [2]int{}
    for i := 0; i < n; i++ {
        // Find the smallest neighbor to determine how much nums[i] must shrink
        left, right := 1<<31-1, 1<<31-1
        if i > 0 { left = nums[i-1] }
        if i+1 < n { right = nums[i+1] }
        minN := left; if right < minN { minN = right }
        // Valley must be strictly less than both neighbors; decrement only, never increase
        if d := nums[i] - minN + 1; d > 0 { ops[i%2] += d }
    }
    if ops[0] < ops[1] { return ops[0] }
    return ops[1]
}
```

### Rust

```
impl Solution {
    pub fn moves_to_make_zigzag(nums: Vec<i32>) -> i32 {
        let n = nums.len();
        let mut ops = [0i32; 2];
        for i in 0..n {
            // Find the smallest neighbor to determine how much nums[i] must shrink
            let left  = if i > 0     { nums[i-1] } else { i32::MAX };
            let right = if i+1 < n   { nums[i+1] } else { i32::MAX };
            let min_n = left.min(right);
            // Valley must be strictly less than both neighbors; decrement only, never increase
            ops[i % 2] += 0.max(nums[i] - min_n + 1);
        }
        ops[0].min(ops[1])
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [1, 2, 3]`
Making odd-indexed elements valleys: decrease $2$ by $\max(0, 2-1+1)=2$ → cost $2$. Making even-indexed elements valleys: decrease $1$ by $\max(0,1-2+1)=0$, decrease $3$ by $\max(0,3-2+1)=2$ → cost $2$. Both parities cost $2$; answer is $2$.

### 2. Slightly Complex
**Input:** `nums = [9, 6, 1, 6, 2]`
Even-parity valleys (indices 0,2,4): index 0 costs $\max(0,9-6+1)=4$, index 2 already less than both neighbors, index 4 is $\min(6) \to 0$. Odd-parity (indices 1,3): index 1 costs $\max(0,6-1+1)=6$, index 3 costs $\max(0,6-1+1)=6$ → $12$. Answer is $4$.

### 3. Edge Case: Time Factor
**Input:** `nums = [1000] * 1000`
All elements equal; one parity pass costs $999 \times 1 = 999$ decrements (each valley must go to $999$), the other costs the same. Loop runs $n=1000$ iterations — linear scan, no shortcuts.

### 4. Edge Case: Space Factor
**Input:** `nums = [1]`
Single element; both ops remain 0. No neighbors exist, so the decrement formula always evaluates to $\max(0, 1 - \infty + 1) = 0$. Answer is $0$ in constant space.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [1, 1000, 1, 1000, 1, 1000]` — already zigzag (peak-even)
Odd-parity valleys (peaks already high): cost $0$. Even-parity valleys would require decreasing each $1$ below its $1000$ neighbors — already satisfied. Both parities return $0$, confirming the greedy handles pre-existing zigzag arrays correctly.
