# LeetCode #1155: Number of Dice Rolls With Target Sum

https://leetcode.com/problems/number-of-dice-rolls-with-target-sum/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (Recursion)** | $O(k^n)$ | $O(n)$ |
| **Optimal: DP with Rolling Array ★** | $O(n \times target \times k)$ | $O(target)$ |

---

## Understanding the Methods

### Brute Force (Recursion)
Enumerate every combination of $n$ dice faces (up to $k$ each), summing all outcomes that equal `target`. With $k^n$ possibilities this is completely infeasible for the given constraints.

### Optimal: DP with Rolling Array ★
Let `dp[s]` = number of ways to achieve sum `s` using dice rolled so far. After each of the $n$ dice, update the table by distributing counts across the $k$ face additions. Using a rolling array (two 1-D arrays) keeps space at $O(target)$ rather than $O(n \times target)$.

**Constraints:**
* $1 \leq n, k \leq 30$
* $1 \leq target \leq 1000$

## Solutions

### C#

```
public class Solution {
    const int MOD = 1_000_000_007;
    public int NumRollsToTarget(int n, int k, int target) {
        // dp[s] = number of ways to reach sum s with dice rolled so far
        var dp = new long[target + 1];
        dp[0] = 1; // zero dice, zero sum: exactly one way
        for (int d = 0; d < n; d++) {
            var next = new long[target + 1];
            for (int s = 0; s <= target; s++) {
                if (dp[s] == 0) continue;
                // Distribute current ways across each face value 1..k
                for (int f = 1; f <= k && s + f <= target; f++)
                    next[s + f] = (next[s + f] + dp[s]) % MOD;
            }
            dp = next;
        }
        return (int)dp[target];
    }
}
```

### Python

```
class Solution:
    def num_rolls_to_target(self, n: int, k: int, target: int) -> int:
        MOD = 10**9 + 7
        # dp[s] = number of ways to reach sum s with dice rolled so far
        dp = [0] * (target + 1)
        dp[0] = 1  # zero dice, zero sum: exactly one way
        for _ in range(n):
            nxt = [0] * (target + 1)
            for s in range(target + 1):
                if not dp[s]: continue
                # Distribute current ways across each face value 1..k
                for f in range(1, k + 1):
                    if s + f <= target:
                        nxt[s + f] = (nxt[s + f] + dp[s]) % MOD
            dp = nxt
        return dp[target]
```

### Rust

```
impl Solution {
    pub fn num_rolls_to_target(n: i32, k: i32, target: i32) -> i32 {
        const MOD: i64 = 1_000_000_007;
        let t = target as usize;
        // dp[s] = number of ways to reach sum s with dice rolled so far
        let mut dp = vec![0i64; t + 1];
        dp[0] = 1; // zero dice, zero sum: exactly one way
        for _ in 0..n {
            let mut next = vec![0i64; t + 1];
            for s in 0..=t {
                if dp[s] == 0 { continue; }
                // Distribute current ways across each face value 1..k
                for f in 1..=k as usize {
                    if s + f <= t { next[s + f] = (next[s + f] + dp[s]) % MOD; }
                }
            }
            dp = next;
        }
        dp[t] as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n = 2`, `k = 6`, `target = 7`
Six ways to roll 7 with two standard dice: (1,6),(2,5),(3,4),(4,3),(5,2),(6,1). DP confirms `dp[7] = 6` after two iterations. Answer: **6**.

### 2. Slightly Complex
**Input:** `n = 2`, `k = 5`, `target = 10`
Only one way: both dice show 5. DP propagates `1 → 5` on first die, then `5 → 10` on second. Answer: **1**.

### 3. Edge Case: Time Factor
**Input:** `n = 30`, `k = 30`, `target = 500`.
Each of the 30 dice iterations scans up to $500$ sums and 30 face values: $30 \times 500 \times 30 = 450{,}000$ operations. Modular arithmetic prevents overflow throughout.

### 4. Edge Case: Space Factor
**Input:** `target = 1000`, `n = 1`, `k = 30`.
Only sums 1–30 can be reached; `dp[1000] = 0`. The rolling array allocates $1001$ slots regardless — $O(target)$ space at its maximum constraint value.

### 5. Almost-Impossible but Plausible
**Input:** `n = 30`, `k = 1`, `target = 30`.
Each die must show exactly 1; there is exactly one way. The DP propagates a single chain: $dp[0] \to dp[1] \to \ldots \to dp[30]$, always carrying value 1 (mod $10^9+7$). Answer: **1**.
