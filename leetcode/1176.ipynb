# LeetCode #1176: Diet Plan Performance

https://leetcode.com/problems/diet-plan-performance/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n \times k)$ | $O(1)$ |
| **Optimal: Sliding Window ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
For each of the $n - k + 1$ windows, recompute the sum from scratch with an inner loop of length $k$. Total cost $O(n \times k)$, which degrades badly when $k \approx n$.

### Optimal: Sliding Window ★
Compute the first window sum once, then slide: add the incoming element and subtract the outgoing element. Each window is evaluated in $O(1)$, giving $O(n)$ total for all windows.

**Why this is better than Brute Force:** Avoids redundant addition of shared elements, cutting the inner loop entirely.

**Constraints:**
* $1 \leq k \leq n \leq 10^5$
* $0 \leq \text{calories}[i] \leq 20{,}000$
* $0 \leq \text{lower} \leq \text{upper}$


## Solutions

### C#

```
public class Solution {
    public int DietPlanPerformance(int[] calories, int k, int lower, int upper) {
        int score = 0;
        int windowSum = 0;

        // Build the first window of length k
        for (int i = 0; i < k; i++) windowSum += calories[i];
        score += Evaluate(windowSum, lower, upper);

        // Slide the window one position at a time across the array
        for (int i = k; i < calories.Length; i++) {
            windowSum += calories[i] - calories[i - k]; // add incoming, drop outgoing
            score += Evaluate(windowSum, lower, upper);
        }

        return score;
    }

    private int Evaluate(int sum, int lower, int upper) =>
        sum < lower ? -1 : sum > upper ? 1 : 0;
}
```

### Python

```
class Solution:
    def dietPlanPerformance(self, calories: list[int], k: int, lower: int, upper: int) -> int:
        score = 0
        window_sum = sum(calories[:k])  # compute the first window once

        def evaluate(s: int) -> int:
            # Assign the point value for this window's total
            return -1 if s < lower else 1 if s > upper else 0

        score += evaluate(window_sum)

        # Slide: subtract the element leaving the left, add the one entering the right
        for i in range(k, len(calories)):
            window_sum += calories[i] - calories[i - k]
            score += evaluate(window_sum)

        return score
```

### Go

```
func dietPlanPerformance(calories []int, k int, lower int, upper int) int {
    score, windowSum := 0, 0

    // Seed the first window
    for i := 0; i < k; i++ {
        windowSum += calories[i]
    }
    score += evaluate(windowSum, lower, upper)

    // Slide across remaining positions
    for i := k; i < len(calories); i++ {
        windowSum += calories[i] - calories[i-k] // add new, drop old
        score += evaluate(windowSum, lower, upper)
    }
    return score
}

func evaluate(sum, lower, upper int) int {
    if sum < lower {
        return -1
    }
    if sum > upper {
        return 1
    }
    return 0
}
```

### Rust

```
impl Solution {
    pub fn diet_plan_performance(calories: Vec<i32>, k: i32, lower: i32, upper: i32) -> i32 {
        let k = k as usize;
        let mut score = 0i32;
        let mut window_sum: i32 = calories[..k].iter().sum();
        score += Self::evaluate(window_sum, lower, upper);

        // Slide the window: one element in, one out
        for i in k..calories.len() {
            window_sum += calories[i] - calories[i - k];
            score += Self::evaluate(window_sum, lower, upper);
        }
        score
    }

    fn evaluate(sum: i32, lower: i32, upper: i32) -> i32 {
        if sum < lower { -1 } else if sum > upper { 1 } else { 0 }
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `calories = [1,2,3,4,5]`, `k = 1`, `lower = 3`, `upper = 3`
Each window is a single element. Sums: 1(-1), 2(-1), 3(0), 4(+1), 5(+1). Score = **0**.

### 2. Slightly Complex
**Input:** `calories = [3,2]`, `k = 2`, `lower = 0`, `upper = 1`
Only one window: sum = 5 > 1, score = **1** (bad day).

### 3. Edge Case: Time Factor
**Input:** $n = 10^5$, $k = 1$, all elements = 10{,}000
Sliding window runs $n - 1$ iterations, each in $O(1)$. Every window sum equals 10,000; since 10,000 > upper (assume upper = 0), score = $n$. Confirms $O(n)$ at maximum array size.

### 4. Edge Case: Space Factor
**Input:** $n = 10^5$, $k = n$ (one single window covering the full array)
No sliding iterations occur — only the initial sum step and one evaluation. Two integers (`score`, `window_sum`) are the only allocation, confirming $O(1)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `calories = [0, 20000, 0, 20000]`, `k = 2`, `lower = 10000`, `upper = 15000`
Windows: `[0+20000=20000](+1)`, `[20000+0=20000](+1)`, `[0+20000=20000](+1)`. Score = **3**. Tests that alternating extremes don't accumulate from the wrong window due to off-by-one in the slide.
