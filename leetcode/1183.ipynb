# LeetCode #1183: Maximum Number of Ones

https://leetcode.com/problems/maximum-number-of-ones/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(m \cdot n \cdot sideLength^2)$ | $O(1)$ |
| **Optimal: Type Count + Greedy ★** | $O(m \cdot n)$ | $O(sideLength^2)$ |

---

## Understanding the Methods

### Brute Force
For each cell, check every `sideLength x sideLength` submatrix it belongs to and count potential 1s. This is extremely slow due to redundant work across overlapping submatrices.

### Optimal: Type Count + Greedy ★
Every position $(r, c)$ has a canonical "type" $(r \bmod sideLength, \ c \bmod sideLength)$ that determines which cell in the repeating tile it corresponds to. All positions of the same type compete for the same slots. Count how many cells of each type exist, sort descending, and assign a 1 to the top `maxOnes` types — each 1 in the tile propagates to all positions of that type.

**Constraints:**
* $1 \leq width, height \leq 100$
* $1 \leq sideLength \leq width$
* $0 \leq maxOnes \leq sideLength^2$


## Solutions

### C#


```
public int MaxOnes(int width, int height, int sideLength, int maxOnes) {
    // Count how many times each tile position (r%side, c%side) appears in the full matrix
    int[] freq = new int[sideLength * sideLength];
    for (int r = 0; r < sideLength; r++) {
        for (int c = 0; c < sideLength; c++) {
            // Full tiles plus any partial remainder in each dimension
            int rows = height / sideLength + (r < height % sideLength ? 1 : 0);
            int cols = width  / sideLength + (c < width  % sideLength ? 1 : 0);
            freq[r * sideLength + c] = rows * cols;
        }
    }
    // Greedily assign 1s to the most-frequent tile positions
    Array.Sort(freq);
    int total = 0;
    for (int i = freq.Length - 1; i >= freq.Length - maxOnes; i--)
        total += freq[i];
    return total;
}
```

### Python


```
def max_ones(width: int, height: int, side_length: int, max_ones: int) -> int:
    freq = []
    # Each tile position (r, c) appears in rows * cols cells of the full matrix
    for r in range(side_length):
        for c in range(side_length):
            rows = height // side_length + (1 if r < height % side_length else 0)
            cols = width  // side_length + (1 if c < width  % side_length else 0)
            freq.append(rows * cols)
    # Pick the max_ones most-frequent positions — they yield the most 1s
    freq.sort(reverse=True)
    return sum(freq[:max_ones])
```

### Go


```
import "sort"

func maxOnes(width int, height int, sideLength int, maxOnes int) int {
    freq := make([]int, sideLength*sideLength)
    // Compute repetition count for every position in the repeating tile
    for r := 0; r < sideLength; r++ {
        for c := 0; c < sideLength; c++ {
            rows := height/sideLength
            if r < height%sideLength { rows++ }
            cols := width/sideLength
            if c < width%sideLength { cols++ }
            freq[r*sideLength+c] = rows * cols
        }
    }
    sort.Sort(sort.Reverse(sort.IntSlice(freq)))
    total := 0
    for i := 0; i < maxOnes; i++ {
        total += freq[i]
    }
    return total
}
```

### Rust


```
pub fn max_ones(width: i32, height: i32, side_length: i32, max_ones: i32) -> i32 {
    let sl = side_length as usize;
    let mut freq: Vec<i32> = Vec::with_capacity(sl * sl);
    // Each tile cell (r,c) maps to rows*cols matrix positions
    for r in 0..side_length {
        for c in 0..side_length {
            let rows = height / side_length + if r < height % side_length { 1 } else { 0 };
            let cols = width  / side_length + if c < width  % side_length { 1 } else { 0 };
            freq.push(rows * cols);
        }
    }
    // Greedily place 1s at the positions that appear most often in the full matrix
    freq.sort_unstable_by(|a, b| b.cmp(a));
    freq.iter().take(max_ones as usize).sum()
}
```

## Example Scenarios

### 1. Common Case — One 1 Allowed per 2 by 2 Window
**Input:** `width = 3, height = 3, sideLength = 2, maxOnes = 1`

Every cell $(r, c)$ belongs to tile position $(r \bmod 2,\ c \bmod 2)$, and all cells sharing a position are forced to the same value. With a $3 \times 3$ matrix the four positions occur 4, 2, 2 and 1 times, so putting the single allowed 1 at position $(0,0)$ lights up 4 cells. Answer **4**.

### 2. Slightly Complex — Non-Square Matrix, Two 1s
**Input:** `width = 4, height = 3, sideLength = 3, maxOnes = 2`

Rows divide as $3 = 1 \cdot 3 + 0$ and columns as $4 = 1 \cdot 3 + 1$, so the position frequencies are not symmetric: column 0 of the tile occurs twice as often as columns 1 and 2. Sorted frequencies are $[2, 2, 2, 1, 1, 1, 1, 1, 1]$ and the top two give **4**. The remainder in each dimension is what breaks the symmetry — $(r < h \bmod s)$ and $(c < w \bmod s)$ in the code.

### 3. Edge Case: Time Factor — Every Tile Position Used
**Input:** `width = 10, height = 10, sideLength = 5, maxOnes = 25`

`maxOnes` equals $sideLength^2$, so every position in the tile may hold a 1 and the whole matrix fills: $10 \times 10 = $ **100**. The work is unaffected — the algorithm always builds all $sideLength^2$ frequencies and sorts them, $O(s^2 \log s)$, no matter how small `maxOnes` is. At the constraint maximum that is $10^4$ entries.

### 4. Edge Case: Space Factor — The Tile Is the Whole Matrix
**Input:** `width = 6, height = 6, sideLength = 6, maxOnes = 20`

With $sideLength = width = height$ there is exactly one window, so every tile position occurs once and the answer is simply `maxOnes` itself: **20**. This is the memory worst case in shape — the frequency list holds $sideLength^2$ entries, which reaches $100^2 = 10^4$ integers at the constraint limit, independent of how large `width` and `height` are.

### 5. Almost-Impossible but Plausible — Large Matrix, Almost No 1s
**Input:** `width = 99, height = 99, sideLength = 10, maxOnes = 3`

$99 = 9 \cdot 10 + 9$, so tile rows $0 \ldots 8$ occur 10 times and row 9 occurs 9 times; the same holds for columns. That makes the 81 positions with $r, c \le 8$ all occur $10 \times 10 = 100$ times, so the three best are 100, 100 and 100 and the answer is **300** — not 280, which is what you get by wrongly assuming only $(0,0)$ reaches 100. Three 1s out of $99 \times 99 = 9801$ cells still cover 3% of the matrix, because each one is replicated across every window.
