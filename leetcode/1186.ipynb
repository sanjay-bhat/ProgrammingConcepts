# LeetCode #1186: Maximum Subarray Sum with One Deletion

https://leetcode.com/problems/maximum-subarray-sum-with-one-deletion/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Dual Kadane DP ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
For each element try deleting it, then run Kadane's algorithm on the modified array. Each deletion + Kadane scan takes $O(n)$, and there are $n$ deletions, giving $O(n^2)$ total.

### Optimal: Dual Kadane DP ★
Maintain two running sums simultaneously: `noDelete` = max subarray ending here with no deletion; `oneDelete` = max subarray ending here with exactly one deletion already used. At each position, `oneDelete` can extend by adding `arr[i]` (paying for a prior deletion) or delete the current element (`noDelete + 0`). Both values are updated in one pass with $O(1)$ space.

**Constraints:**
* $1 \leq \text{arr.length} \leq 10^5$
* $-10^4 \leq \text{arr}[i] \leq 10^4$


## Solutions

### C#

```
public class Solution {
    public int MaximumSum(int[] arr) {
        // noDelete = max subarray sum ending here using no deletion
        // oneDelete = max subarray sum ending here using exactly one deletion
        int noDelete = arr[0], oneDelete = 0;
        int result = arr[0];

        for (int i = 1; i < arr.Length; i++) {
            // Using a deletion: either extend prior one-delete subarray, or delete current element
            oneDelete = Math.Max(oneDelete + arr[i], noDelete);
            // No deletion: extend or start fresh
            noDelete = Math.Max(noDelete + arr[i], arr[i]);

            result = Math.Max(result, Math.Max(noDelete, oneDelete));
        }

        return result;
    }
}
```

### Python

```
class Solution:
    def maximumSum(self, arr: list[int]) -> int:
        # no_delete = max subarray sum ending here using no deletion
        # one_delete = max subarray sum ending here using exactly one deletion
        no_delete = arr[0]
        one_delete = 0
        result = arr[0]

        for i in range(1, len(arr)):
            # Using a deletion: either extend prior one-delete subarray, or delete current element
            one_delete = max(one_delete + arr[i], no_delete)
            # No deletion: extend or start fresh
            no_delete = max(no_delete + arr[i], arr[i])

            result = max(result, no_delete, one_delete)

        return result
```

### Go

```
func maximumSum(arr []int) int {
    // noDelete = max subarray sum ending here using no deletion
    // oneDelete = max subarray sum ending here using exactly one deletion
    noDelete := arr[0]
    oneDelete := 0
    result := arr[0]

    for i := 1; i < len(arr); i++ {
        // Using a deletion: either extend prior one-delete subarray, or delete current element
        if noDelete > oneDelete+arr[i] {
            oneDelete = noDelete
        } else {
            oneDelete = oneDelete + arr[i]
        }
        // No deletion: extend or start fresh
        if noDelete+arr[i] > arr[i] {
            noDelete = noDelete + arr[i]
        } else {
            noDelete = arr[i]
        }

        if noDelete > result { result = noDelete }
        if oneDelete > result { result = oneDelete }
    }

    return result
}
```

### Rust

```
impl Solution {
    pub fn maximum_sum(arr: Vec<i32>) -> i32 {
        // no_delete = max subarray sum ending here using no deletion
        // one_delete = max subarray sum ending here using exactly one deletion
        let mut no_delete = arr[0];
        let mut one_delete = 0i32;
        let mut result = arr[0];

        for &num in &arr[1..] {
            // Using a deletion: either extend prior one-delete subarray, or delete current element
            one_delete = (one_delete + num).max(no_delete);
            // No deletion: extend or start fresh
            no_delete = (no_delete + num).max(num);

            result = result.max(no_delete).max(one_delete);
        }

        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [1, -2, 0, 3]`
Without deletion: Kadane gives `[0, 3] = 3`. Deleting `-2` enables `[1, 0, 3] = 4`. The `oneDelete` tracker captures this: at index 3, `oneDelete = 3 + noDelete_prev = 4`.

### 2. Slightly Complex
**Input:** `arr = [1, -2, -2, 3]`
Deleting either `-2` still leaves the other — the best is `[3] = 3` or `[1, -2, 3] = 2` with one delete. Deleting the first `-2` gives `[1, -2, 3] = 2`; deleting the second gives `[1, -2, 3] = 2`. Best no-delete: `[3]`. Result = 3.

### 3. Edge Case: Time Factor
**Input:** `arr = [-1, -1, \ldots, -1, 10^4]` ($10^5$ elements, positive at end)
The best subarray is just `[10^4]` with or without deletion. The loop runs $10^5 - 1$ iterations — the worst-case time path — before `noDelete` finally becomes positive at the last element.

### 4. Edge Case: Space Factor
**Input:** `arr = [10^4]` (single element)
Immediately returns `arr[0]` — no loop runs. The algorithm uses just 3 integer variables regardless of `n`, confirming $O(1)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [-10^4, 10^4, -10^4, 10^4]`
Without deletion: alternating signs limit the best subarray to a single $10^4$ element. With one deletion (e.g., delete the middle $-10^4$): `[10^4, 10^4] = 2 \times 10^4$. `oneDelete` captures this bridge across the deleted element in one pass.
