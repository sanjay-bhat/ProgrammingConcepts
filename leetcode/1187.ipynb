# LeetCode #1187: Make Two Arrays Equal by Reversing Subarrays

https://leetcode.com/problems/make-two-arrays-equal-by-reversing-subarrays/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n! \cdot n)$ | $O(n)$ |
| **Optimal: Sort and Compare ★** | $O(n \log n)$ | $O(1)$ extra |

---

## Understanding the Methods

### Brute Force
Try every possible sequence of subarray reversals on `arr` and check whether any sequence transforms it into `target`. The search space is factorial — completely impractical for large inputs.

### Optimal: Sort and Compare ★
Key insight: any sequence of subarray reversals can achieve any permutation of the array. Therefore `arr` can be transformed into `target` if and only if they are anagrams — they contain the same multiset of elements. Sorting both and comparing element-by-element verifies this in $O(n \log n)$.

**Constraints:**
* $1 \leq \text{target.length} \leq 1000$
* $\text{target.length} == \text{arr.length}$
* $1 \leq \text{target}[i], \text{arr}[i] \leq 1000$


## Solutions

### C#

```
public class Solution {
    public bool CanBeEqual(int[] target, int[] arr) {
        // Any permutation is reachable via reversals, so equality reduces to same multiset
        Array.Sort(target);
        Array.Sort(arr);
        return target.SequenceEqual(arr);
    }
}
```

### Python

```
class Solution:
    def can_be_equal(self, target: list[int], arr: list[int]) -> bool:
        # Any permutation is reachable via reversals, so equality reduces to same multiset
        return sorted(target) == sorted(arr)
```

### Go

```
import "sort"

func canBeEqual(target []int, arr []int) bool {
    // Any permutation is reachable via reversals, so equality reduces to same multiset
    sort.Ints(target)
    sort.Ints(arr)
    for i := range target {
        if target[i] != arr[i] {
            return false
        }
    }
    return true
}
```

### Rust

```
impl Solution {
    pub fn can_be_equal(mut target: Vec<i32>, mut arr: Vec<i32>) -> bool {
        // Any permutation is reachable via reversals, so equality reduces to same multiset
        target.sort_unstable();
        arr.sort_unstable();
        target == arr
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `target = [1,2,3,4]`, `arr = [2,4,1,3]`
Both sort to `[1,2,3,4]`. Since the multisets match, any sequence of reversals can reorder `arr` into `target` — return `true`.

### 2. Slightly Complex
**Input:** `target = [7]`, `arr = [7]`
Single-element arrays; sorted forms are identical. Trivially true — a zero-reversal sequence suffices.

### 3. Edge Case: Time Factor
**Input:** `target = [1,2,...,1000]`, `arr = [1000,999,...,1]` (reversed order)
Sorting both takes $O(n \log n) \approx 10{,}000$ comparisons for $n=1000$. After sorting both are `[1,...,1000]` — equal. Full sort is the worst-case path.

### 4. Edge Case: Space Factor
**Input:** `target = [1,2,3]`, `arr = [1,2,4]`
After sorting: `[1,2,3]` vs `[1,2,4]`. Last element differs — return `false`. In-place sort uses $O(1)$ extra space; no hash map or frequency array required.

### 5. Almost-Impossible but Plausible
**Input:** `target = [1,1,1,1000]`, `arr = [1000,1,1,1]`
Duplicates everywhere. Sorted: `[1,1,1,1000]` vs `[1,1,1,1000]` — equal. Confirms that duplicate handling falls naturally out of the sort-and-compare strategy with no special casing.
