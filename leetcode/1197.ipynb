# LeetCode #1197: Minimum Knight Moves

https://leetcode.com/problems/minimum-knight-moves/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (DFS + Memo)** | $O(|x| \cdot |y|)$ | $O(|x| \cdot |y|)$ |
| **Optimal: BFS (Symmetry-Reduced) ★** | $O(|x| \cdot |y|)$ | $O(|x| \cdot |y|)$ |

---

## Understanding the Methods

### Brute Force (DFS + Memo)
Recursively explore all 8 knight moves from the target back toward (0,0) with memoization. Terminates but explores from all quadrants.

### Optimal: BFS (Symmetry-Reduced) ★
Exploit 8-fold symmetry: reflect $(x, y)$ into the first quadrant and run BFS from $(0,0)$. BFS guarantees the first time the target is reached is the minimum distance. Restricting search to the first quadrant and a small buffer reduces visited states significantly.

**Why this is better than Brute Force:** BFS explores level by level so the first hit is optimal without needing memoization or distance tracking; symmetry reduction cuts the search space by up to $8\times$.

**Constraints:**
* $-300 \le x, y \le 300$

## Solutions

### C#

```
using System.Collections.Generic;
public class Solution {
    public int MinKnightMoves(int x, int y) {
        // Reflect into first quadrant — the board is 8-fold symmetric
        x = Math.Abs(x); y = Math.Abs(y);
        var queue = new Queue<(int, int, int)>();
        queue.Enqueue((0, 0, 0));
        var visited = new HashSet<(int, int)> { (0, 0) };
        int[] dr = { 1, 2, 2, 1, -1, -2, -2, -1 };
        int[] dc = { 2, 1, -1, -2, -2, -1, 1, 2 };
        while (queue.Count > 0) {
            var (r, c, moves) = queue.Dequeue();
            if (r == x && c == y) return moves;
            for (int d = 0; d < 8; d++) {
                int nr = r + dr[d], nc = c + dc[d];
                // Stay within a small buffer of the first quadrant to avoid blowup
                if (nr >= -2 && nc >= -2 && !visited.Contains((nr, nc))) {
                    visited.Add((nr, nc));
                    queue.Enqueue((nr, nc, moves + 1));
                }
            }
        }
        return -1;
    }
}
```

### Python

```
from collections import deque
class Solution:
    def min_knight_moves(self, x: int, y: int) -> int:
        # Reflect into first quadrant — the board is 8-fold symmetric
        x, y = abs(x), abs(y)
        queue = deque([(0, 0, 0)])
        visited = {(0, 0)}
        moves_8 = [(1,2),(2,1),(-1,2),(-2,1),(1,-2),(2,-1),(-1,-2),(-2,-1)]
        while queue:
            r, c, moves = queue.popleft()
            if r == x and c == y:
                return moves
            for dr, dc in moves_8:
                nr, nc = r + dr, c + dc
                # Stay within a small buffer of the first quadrant to avoid blowup
                if nr >= -2 and nc >= -2 and (nr, nc) not in visited:
                    visited.add((nr, nc))
                    queue.append((nr, nc, moves + 1))
        return -1
```

### Go

```
func minKnightMoves(x int, y int) int {
    // Reflect into first quadrant — the board is 8-fold symmetric
    if x < 0 { x = -x }
    if y < 0 { y = -y }
    type State struct{ r, c, moves int }
    type Pos struct{ r, c int }
    queue := []State{{0, 0, 0}}
    visited := map[Pos]bool{{0, 0}: true}
    dr := []int{1, 2, 2, 1, -1, -2, -2, -1}
    dc := []int{2, 1, -1, -2, -2, -1, 1, 2}
    for len(queue) > 0 {
        cur := queue[0]; queue = queue[1:]
        if cur.r == x && cur.c == y { return cur.moves }
        for d := 0; d < 8; d++ {
            nr, nc := cur.r+dr[d], cur.c+dc[d]
            p := Pos{nr, nc}
            // Stay within a small buffer of the first quadrant to avoid blowup
            if nr >= -2 && nc >= -2 && !visited[p] {
                visited[p] = true
                queue = append(queue, State{nr, nc, cur.moves + 1})
            }
        }
    }
    return -1
}
```

### Rust

```
use std::collections::{VecDeque, HashSet};
impl Solution {
    pub fn min_knight_moves(x: i32, y: i32) -> i32 {
        // Reflect into first quadrant — the board is 8-fold symmetric
        let (x, y) = (x.abs(), y.abs());
        let mut queue = VecDeque::new();
        queue.push_back((0i32, 0i32, 0i32));
        let mut visited = HashSet::new();
        visited.insert((0i32, 0i32));
        let dirs = [(1,2),(2,1),(-1,2),(-2,1),(1,-2),(2,-1),(-1,-2),(-2,-1)];
        while let Some((r, c, moves)) = queue.pop_front() {
            if r == x && c == y { return moves; }
            for (dr, dc) in dirs {
                let (nr, nc) = (r + dr, c + dc);
                // Stay within a small buffer of the first quadrant to avoid blowup
                if nr >= -2 && nc >= -2 && !visited.contains(&(nr, nc)) {
                    visited.insert((nr, nc));
                    queue.push_back((nr, nc, moves + 1));
                }
            }
        }
        -1
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `x = 2, y = 1`
One knight move $(0,0) \to (2,1)$ reaches the target directly. BFS finds it in level 1. Answer: **1**.

### 2. Slightly Complex
**Input:** `x = 5, y = 5`
Optimal path uses 4 moves, e.g. $(0,0)\to(2,1)\to(4,2)\to(3,4)\to(5,5)$. BFS explores level by level and returns 4 at the first hit.

### 3. Edge Case: Time Factor
**Input:** `x = 300, y = 300`
Symmetry reduction folds the board into the first quadrant; BFS visits roughly $300 \times 300 = 90{,}000$ cells — all processed in $O(|x| \cdot |y|)$ time.

### 4. Edge Case: Space Factor
**Input:** `x = 300, y = 0`
A straight-line target along an axis requires many intermediate diagonal moves; the visited set and queue together hold $O(|x| \cdot |y|)$ entries in the worst case.

### 5. Almost-Impossible but Plausible
**Input:** `x = 0, y = 0`
Start equals destination; BFS dequeues $(0,0,0)$ immediately and returns **0** without enqueuing any neighbor.
