# LeetCode #1205: Google Question

https://leetcode.com/problems/google-question/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Sort and Two-Pointer ★** | $O(n \log n)$ | $O(1)$ extra |

---

## Understanding the Methods

### Brute Force
For each pair of jobs, check whether they overlap by comparing start/end times with every other job, counting conflicts with a nested loop. $O(n^2)$ comparisons, impractical for large inputs.

### Optimal: Sort and Two-Pointer ★
Sort jobs by start time; maintain a min-heap (or sorted end-time pointer) tracking the earliest-ending active room. For each new job, if the earliest-ending room finishes before the new job starts, reuse it; otherwise open a new room. Sorting dominates at $O(n \log n)$; the two-pointer sweep over sorted end-times keeps space to $O(1)$ extra beyond the sort.

**Constraints:**
* $1 \leq \text{intervals.length} \leq 10^4$
* $\text{intervals}[i].\text{length} == 2$
* $0 \leq \text{start}_i < \text{end}_i \leq 10^6$


## Solutions

### C#

```
public class Solution {
    public int MinMeetingRooms(int[][] intervals) {
        int n = intervals.Length;
        int[] starts = new int[n], ends = new int[n];
        for (int i = 0; i < n; i++) {
            starts[i] = intervals[i][0];
            ends[i]   = intervals[i][1];
        }
        // Sorting starts and ends independently lets us greedily match rooms
        Array.Sort(starts);
        Array.Sort(ends);
        int rooms = 0, ep = 0;
        for (int i = 0; i < n; i++) {
            // A room is freed only when its meeting ends before the next one starts
            if (starts[i] < ends[ep]) rooms++;
            else ep++;
        }
        return rooms;
    }
}
```

### Python

```
class Solution:
    def min_meeting_rooms(self, intervals: list[list[int]]) -> int:
        n = len(intervals)
        starts = sorted(i[0] for i in intervals)
        ends   = sorted(i[1] for i in intervals)
        rooms = ep = 0
        for i in range(n):
            # A room is freed only when its meeting ends before the next one starts
            if starts[i] < ends[ep]:
                rooms += 1
            else:
                ep += 1
        return rooms
```

### Go

```
import "sort"

func minMeetingRooms(intervals [][]int) int {
    n := len(intervals)
    starts, ends := make([]int, n), make([]int, n)
    for i, iv := range intervals {
        starts[i], ends[i] = iv[0], iv[1]
    }
    // Sorting starts and ends independently lets us greedily match rooms
    sort.Ints(starts)
    sort.Ints(ends)
    rooms, ep := 0, 0
    for i := 0; i < n; i++ {
        // A room is freed only when its meeting ends before the next one starts
        if starts[i] < ends[ep] {
            rooms++
        } else {
            ep++
        }
    }
    return rooms
}
```

### Rust

```
impl Solution {
    pub fn min_meeting_rooms(intervals: Vec<Vec<i32>>) -> i32 {
        let mut starts: Vec<i32> = intervals.iter().map(|iv| iv[0]).collect();
        let mut ends:   Vec<i32> = intervals.iter().map(|iv| iv[1]).collect();
        // Sorting starts and ends independently lets us greedily match rooms
        starts.sort_unstable();
        ends.sort_unstable();
        let (mut rooms, mut ep) = (0i32, 0usize);
        for i in 0..starts.len() {
            // A room is freed only when its meeting ends before the next one starts
            if starts[i] < ends[ep] {
                rooms += 1;
            } else {
                ep += 1;
            }
        }
        rooms
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `intervals = [[0,30],[5,10],[15,20]]`
Starts: `[0,5,15]`, ends: `[10,20,30]`. At `s=0`: `0 < 10`, open room 1. At `s=5`: `5 < 10`, open room 2. At `s=15`: `15 >= 10`, reuse — `ep` advances. Answer: **2 rooms**.

### 2. Slightly Complex
**Input:** `intervals = [[7,10],[2,4]]`
Starts: `[2,7]`, ends: `[4,10]`. At `s=2`: room 1. At `s=7`: `7 >= 4` — reuse, `ep=1`. Answer: **1 room**. Non-sorted input handled seamlessly.

### 3. Edge Case: Time Factor
**Input:** $n = 10{,}000$ intervals all starting at $t=0$
Every meeting overlaps with every other; no room is ever freed. The pointer `ep` never advances, so `rooms` increments $n$ times. Full $O(n)$ sweep required after the $O(n \log n)$ sort.

### 4. Edge Case: Space Factor
**Input:** $n = 10{,}000$ back-to-back non-overlapping intervals
The `ep` pointer races ahead in lockstep with `i`; `rooms` never exceeds 1. Only the two sorted arrays (in-place) plus two scalars needed — $O(1)$ extra space.

### 5. Almost-Impossible but Plausible
**Input:** `intervals = [[0,1000000],[1,999999],[2,999998]]` (nearly full overlap with one-unit offsets)
All three start before any ends; all three rooms open. The tight boundary (`start < end`, strict inequality) means `[0,5]` and `[5,10]` do **not** overlap — a near-miss that's easy to get wrong with `<=` vs `<`.
