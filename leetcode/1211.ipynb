# LeetCode #1211: Divide Chocolate

https://leetcode.com/problems/divide-chocolate/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^n)$ | $O(n)$ |
| **Optimal: Binary Search on Answer ★** | $O(n \log(\text{sum}))$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Try every possible way to split the array into exactly $k+1$ pieces and track the maximum minimum piece. Exponential in $n$ — impractical for $n > 20$.

### Optimal: Binary Search on Answer ★
Binary search on the minimum sweetness value $m$ you keep. For each candidate $m$, greedily scan left-to-right and cut whenever the accumulated sweetness hits $m$ — if you can make at least $k+1$ cuts, then $m$ is achievable. The largest achievable $m$ is the answer.

**Why this is better than Brute Force:** Instead of enumerating partitions, we reduce the search to $O(\log \text{sum})$ feasibility checks, each $O(n)$.

**Constraints:**
* $1 \le k < n \le 10^4$
* $1 \le sweetness[i] \le 10^5$

## Solutions

### C#

```
public class Solution {
    public int MaximizeSweetness(int[] sweetness, int k) {
        // Binary search between the smallest possible piece and the equal-split average
        int lo = sweetness.Min(), hi = sweetness.Sum() / (k + 1);
        while (lo < hi) {
            int mid = (lo + hi + 1) / 2;
            // Check: can we cut at least k+1 pieces each with sweetness >= mid?
            if (CanDivide(sweetness, k + 1, mid)) lo = mid;
            else hi = mid - 1;
        }
        return lo;
    }
    private bool CanDivide(int[] sweetness, int pieces, int minSweet) {
        int count = 0, cur = 0;
        foreach (int s in sweetness) {
            cur += s;
            // Cut here because the current piece meets the minimum threshold
            if (cur >= minSweet) { count++; cur = 0; }
        }
        return count >= pieces;
    }
}
```

### Python

```
class Solution:
    def maximize_sweetness(self, sweetness: list[int], k: int) -> int:
        def can_divide(min_sweet: int) -> bool:
            count = cur = 0
            for s in sweetness:
                cur += s
                # Cut here because the current piece meets the minimum threshold
                if cur >= min_sweet:
                    count += 1
                    cur = 0
            return count >= k + 1
        # Binary search between the smallest possible piece and the equal-split average
        lo, hi = min(sweetness), sum(sweetness) // (k + 1)
        while lo < hi:
            mid = (lo + hi + 1) // 2
            if can_divide(mid):
                lo = mid
            else:
                hi = mid - 1
        return lo
```

### Go

```
func maximizeSweetness(sweetness []int, k int) int {
    lo, hi, total := sweetness[0], 0, 0
    for _, s := range sweetness {
        if s < lo { lo = s }
        total += s
    }
    hi = total / (k + 1)
    canDivide := func(minSweet int) bool {
        count, cur := 0, 0
        for _, s := range sweetness {
            cur += s
            // Cut here because the current piece meets the minimum threshold
            if cur >= minSweet { count++; cur = 0 }
        }
        return count >= k+1
    }
    // Binary search between the smallest possible piece and the equal-split average
    for lo < hi {
        mid := (lo + hi + 1) / 2
        if canDivide(mid) { lo = mid } else { hi = mid - 1 }
    }
    return lo
}
```

### Rust

```
impl Solution {
    pub fn maximize_sweetness(sweetness: Vec<i32>, k: i32) -> i32 {
        let (mut lo, total) = (*sweetness.iter().min().unwrap(),
                               sweetness.iter().sum::<i32>());
        let mut hi = total / (k + 1);
        let can_divide = |min_sweet: i32| -> bool {
            let (mut count, mut cur) = (0i32, 0i32);
            for &s in &sweetness {
                cur += s;
                // Cut here because the current piece meets the minimum threshold
                if cur >= min_sweet { count += 1; cur = 0; }
            }
            count >= k + 1
        };
        // Binary search between the smallest possible piece and the equal-split average
        while lo < hi {
            let mid = (lo + hi + 1) / 2;
            if can_divide(mid) { lo = mid; } else { hi = mid - 1; }
        }
        lo
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `sweetness = [1,2,3,4,5,6,7,8,9], k = 5`
Divide into 6 pieces. Binary search converges to $\text{mid} = 6$: greedy cutting gives pieces $[1..5]=15 \ge 6$ — yes, feasible. Answer: **6**.

### 2. Slightly Complex
**Input:** `sweetness = [5,6,7,8,9,1,2,3,4], k = 8`
9 pieces for 9 elements means each piece is exactly one element; the minimum element value is 1. Binary search immediately confirms $lo = hi = 1$. Answer: **1**.

### 3. Edge Case: Time Factor
**Input:** `sweetness` has $n = 10^4$ elements all equal to $10^5$, $k = 1$.
Binary search runs $\approx \log_2(5 \times 10^8) \approx 29$ iterations, each sweeping $10^4$ elements — total $\approx 290{,}000$ operations, well within limits.

### 4. Edge Case: Space Factor
**Input:** Any input with $n = 10^4$.
No auxiliary array is allocated; only a running sum and a counter are maintained, keeping space at $O(1)$ regardless of input size.

### 5. Almost-Impossible but Plausible
**Input:** `sweetness = [1], k = 0`
Only one element, zero friends — you keep the whole bar. Binary search: $lo = hi = 1$. Answer: **1**.
