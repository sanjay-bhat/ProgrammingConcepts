# LeetCode #1216: Valid Palindrome III

https://leetcode.com/problems/valid-palindrome-iii/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^n)$ | $O(n)$ |
| **Optimal: LCS-based DP ★** | $O(n^2)$ | $O(n^2)$ |

---

## Understanding the Methods

### Brute Force
Enumerate all $2^n$ subsequences of the string and check whether any palindromic subsequence of length $\geq n - k$ exists. Completely infeasible for strings of length 1000.

### Optimal: LCS-based DP ★
A string can be made a palindrome by deleting at most $k$ characters iff its Longest Common Subsequence with its reverse has length $\geq n - k$ (each character not in the LCS must be deleted). Compute LCS in $O(n^2)$ time and $O(n^2)$ space.

**Constraints:**
* $1 \leq s.length \leq 1000$
* $s$ consists of lowercase English letters
* $1 \leq k \leq s.length$

## Solutions

### C#

```
public class Solution {
    public bool IsValidPalindrome(string s, int k) {
        int n = s.Length;
        string rev = new string(s.ToCharArray());
        char[] arr = rev.ToCharArray(); System.Array.Reverse(arr); rev = new string(arr);
        // LCS of s and its reverse = length of longest palindromic subsequence
        var dp = new int[n + 1, n + 1];
        for (int i = 1; i <= n; i++)
            for (int j = 1; j <= n; j++)
                if (s[i-1] == rev[j-1])
                    // Matching chars extend the palindromic subsequence
                    dp[i, j] = dp[i-1, j-1] + 1;
                else
                    dp[i, j] = Math.Max(dp[i-1, j], dp[i, j-1]);
        // Characters outside the LCS must be deleted; check if count fits within k
        return n - dp[n, n] <= k;
    }
}
```

### Python

```
class Solution:
    def is_valid_palindrome(self, s: str, k: int) -> bool:
        n = len(s)
        rev = s[::-1]
        # LCS of s and its reverse = length of longest palindromic subsequence
        dp = [[0] * (n + 1) for _ in range(n + 1)]
        for i in range(1, n + 1):
            for j in range(1, n + 1):
                if s[i-1] == rev[j-1]:
                    # Matching chars extend the palindromic subsequence
                    dp[i][j] = dp[i-1][j-1] + 1
                else:
                    dp[i][j] = max(dp[i-1][j], dp[i][j-1])
        # Characters outside the LCS must be deleted; check if count fits within k
        return n - dp[n][n] <= k
```

### Go

```
func isValidPalindrome(s string, k int) bool {
    n := len(s)
    runes := []byte(s)
    for i, j := 0, n-1; i < j; i, j = i+1, j-1 { runes[i], runes[j] = runes[j], runes[i] }
    rev := string(runes)
    // LCS of s and its reverse = length of longest palindromic subsequence
    dp := make([][]int, n+1)
    for i := range dp { dp[i] = make([]int, n+1) }
    for i := 1; i <= n; i++ {
        for j := 1; j <= n; j++ {
            if s[i-1] == rev[j-1] {
                // Matching chars extend the palindromic subsequence
                dp[i][j] = dp[i-1][j-1] + 1
            } else if dp[i-1][j] > dp[i][j-1] {
                dp[i][j] = dp[i-1][j]
            } else {
                dp[i][j] = dp[i][j-1]
            }
        }
    }
    // Characters outside the LCS must be deleted; check if count fits within k
    return n - dp[n][n] <= k
}
```

### Rust

```
impl Solution {
    pub fn is_valid_palindrome(s: String, k: i32) -> bool {
        let n = s.len();
        let sb: Vec<u8> = s.bytes().collect();
        let rev: Vec<u8> = sb.iter().rev().cloned().collect();
        // LCS of s and its reverse = length of longest palindromic subsequence
        let mut dp = vec![vec![0i32; n + 1]; n + 1];
        for i in 1..=n {
            for j in 1..=n {
                if sb[i-1] == rev[j-1] {
                    // Matching chars extend the palindromic subsequence
                    dp[i][j] = dp[i-1][j-1] + 1;
                } else {
                    dp[i][j] = dp[i-1][j].max(dp[i][j-1]);
                }
            }
        }
        // Characters outside the LCS must be deleted; check if count fits within k
        (n as i32) - dp[n][n] <= k
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `s = "abcdeca"`, `k = 2`
Reverse is `"acedcba"`. LCS = `"acdca"` or `"acea"`, length 4 (or `"aceca"` = 5). Deleting 2 chars suffices. $n - LCS = 7 - 5 = 2 \leq k$. Answer: **true**.

### 2. Slightly Complex
**Input:** `s = "abbababa"`, `k = 1`
Longest palindromic subsequence has length 7 (`"abababa"`). Deletions needed: $8 - 7 = 1 \leq 1$. Answer: **true**.

### 3. Edge Case: Time Factor
**Input:** `s` of length 1000 with all distinct characters.
LCS of `s` and its reverse is at most $\lfloor n/2 \rfloor = 500$. The full $1000 \times 1000 = 10^6$ DP cells are computed regardless, exercising the worst-case time.

### 4. Edge Case: Space Factor
**Input:** `s` of length 1000.
The DP table uses $(1001)^2 \approx 10^6$ integers — $\approx 4$ MB. This is the maximum $O(n^2)$ space footprint at the constraint ceiling.

### 5. Almost-Impossible but Plausible
**Input:** `s = "aaa...a"` (1000 `a`s), `k = 0`.
The string is already a palindrome ($LCS = 1000$). Deletions needed: $1000 - 1000 = 0 \leq 0$. The DP fills all cells with the correct value of $\min(i,j)$ along the path. Answer: **true**.
