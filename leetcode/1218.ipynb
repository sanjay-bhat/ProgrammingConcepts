# LeetCode #1218: Longest Arithmetic Subsequence of Given Difference

https://leetcode.com/problems/longest-arithmetic-subsequence-of-given-difference/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: DP with HashMap ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For every pair $(i, j)$ where $j > i$ and `arr[j] - arr[i] == difference`, extend the subsequence ending at $i$. This nested scan is $O(n^2)$.

### Optimal: DP with HashMap ★
Maintain `dp[x]` = length of the longest arithmetic subsequence with the given difference ending at value `x`. For each element `v`, `dp[v] = dp[v - difference] + 1` in $O(1)$ using a hash map. A single left-to-right pass gives $O(n)$.

**Constraints:**
* $1 \leq arr.length \leq 10^5$
* $-10^4 \leq arr[i], difference \leq 10^4$

## Solutions

### C#

```
using System.Collections.Generic;
public class Solution {
    public int LongestSubsequence(int[] arr, int difference) {
        // dp[v] = longest arithmetic subsequence with fixed difference ending at value v
        var dp = new Dictionary<int, int>();
        int best = 1;
        foreach (int v in arr) {
            // Extend the chain that ends at v - difference (or start a new chain of length 1)
            int prev = dp.ContainsKey(v - difference) ? dp[v - difference] : 0;
            dp[v] = prev + 1;
            if (dp[v] > best) best = dp[v];
        }
        return best;
    }
}
```

### Python

```
class Solution:
    def longest_subsequence(self, arr: list[int], difference: int) -> int:
        # dp[v] = longest arithmetic subsequence with fixed difference ending at value v
        dp: dict[int, int] = {}
        best = 1
        for v in arr:
            # Extend the chain that ends at v - difference (or start a new chain of length 1)
            dp[v] = dp.get(v - difference, 0) + 1
            if dp[v] > best:
                best = dp[v]
        return best
```

### Go

### Rust

```
use std::collections::HashMap;
impl Solution {
    pub fn longest_subsequence(arr: Vec<i32>, difference: i32) -> i32 {
        // dp[v] = longest arithmetic subsequence with fixed difference ending at value v
        let mut dp: HashMap<i32, i32> = HashMap::new();
        let mut best = 1i32;
        for v in arr {
            // Extend the chain that ends at v - difference (or start a new chain of length 1)
            let prev = *dp.get(&(v - difference)).unwrap_or(&0);
            dp.insert(v, prev + 1);
            if prev + 1 > best { best = prev + 1; }
        }
        best
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [1, 2, 3, 4]`, `difference = 1`
Each element extends the previous: $dp[1]=1, dp[2]=2, dp[3]=3, dp[4]=4$. The entire array is one arithmetic subsequence. Answer: **4**.

### 2. Slightly Complex
**Input:** `arr = [1, 3, 5, 7]`, `difference = 1`
No two adjacent values differ by exactly 1. Each element starts a new chain of length 1. Answer: **1**.

### 3. Edge Case: Time Factor
**Input:** `arr` of length $10^5$ where all elements form one chain (e.g., $1, 2, 3, \ldots, 10^5$ with `difference = 1`).
Every element hits the HashMap lookup and insert once — $O(n)$ operations total with no hash collisions in practice.

### 4. Edge Case: Space Factor
**Input:** `arr = [1, 3, 5, 7, \ldots]$ ($10^5$ distinct odd values), `difference = 3`.
Each value maps to a distinct key in `dp` with no predecessor. All $10^5$ entries are inserted but chain lengths stay at 1. HashMap reaches $O(n)$ size.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [10000, -10000, 10000, -10000]`, `difference = -20000`.
$dp[10000] = 1$, $dp[-10000] = dp[10000 + (-20000)] + 1 = dp[-10000... wait: -10000 - (-20000) = 10000$, so $dp[-10000] = dp[10000] + 1 = 2$, then $dp[10000] = dp[10000 - (-20000)] + 1 = dp[30000] + 1 = 1$ (no 30000), so $dp[10000]$ gets overwritten to $1$, then $dp[-10000] = dp[10000]+1 = 2$. Best chain: **2**.
