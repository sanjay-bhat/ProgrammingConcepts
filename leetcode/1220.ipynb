# LeetCode #1220: Count Vowels Permutation

https://leetcode.com/problems/count-vowels-permutation/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(5^n)$ | $O(n)$ |
| **Optimal: Transition-Matrix DP ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Recursively enumerate all vowel strings of length `n` obeying the adjacency rules, counting valid ones. The branching factor of up to 4 makes this exponential and unusable beyond tiny inputs.

### Optimal: Transition-Matrix DP ★
Track five counts — one per vowel — representing how many valid strings of the current length end with each vowel. At each step, propagate counts according to the fixed adjacency rules (e.g., 'a' → 'e'; 'e' → 'a','i'; etc.). Only the previous row is ever needed, giving $O(1)$ space.

**Why this is better than Brute Force:** We never enumerate strings; we only accumulate counts using 5 integer additions per step, reducing runtime from exponential to linear.

**Constraints:**
* $1 \leq n \leq 2 \times 10^4$
* Answer modulo $10^9 + 7$

## Solutions

### C#

```
public class Solution {
    public int CountVowelPermutation(int n) {
        const long MOD = 1_000_000_007;
        // Counts of valid strings of current length ending in a, e, i, o, u
        long a = 1, e = 1, i = 1, o = 1, u = 1;

        for (int step = 1; step < n; step++) {
            // Each vowel's new count = sum of counts of vowels that can precede it
            // Rules (what can come BEFORE each vowel):
            //   a  <- e, i, u
            //   e  <- a, i
            //   i  <- e, o
            //   o  <- i
            //   u  <- i, o
            long na = (e + i + u) % MOD;
            long ne = (a + i) % MOD;
            long ni = (e + o) % MOD;
            long no = i % MOD;
            long nu = (i + o) % MOD;
            a = na; e = ne; i = ni; o = no; u = nu;
        }
        return (int)((a + e + i + o + u) % MOD);
    }
}
```

### Python

```
class Solution:
    def count_vowel_permutation(self, n: int) -> int:
        MOD = 10**9 + 7
        # Counts of valid strings of current length ending in a, e, i, o, u
        a = e = i = o = u = 1

        for _ in range(n - 1):
            # Each vowel's new count = sum of counts of vowels that can precede it
            # Rules (what can come BEFORE each vowel):
            #   a  <- e, i, u
            #   e  <- a, i
            #   i  <- e, o
            #   o  <- i
            #   u  <- i, o
            a, e, i, o, u = (
                (e + i + u) % MOD,
                (a + i) % MOD,
                (e + o) % MOD,
                i % MOD,
                (i + o) % MOD,
            )
        return (a + e + i + o + u) % MOD
```

### Go

```
func countVowelPermutation(n int) int {
    const MOD = 1_000_000_007
    // Counts of valid strings of current length ending in a, e, i, o, u
    a, e, i, o, u := 1, 1, 1, 1, 1

    for step := 1; step < n; step++ {
        // Each vowel's new count = sum of counts of vowels that can precede it
        // Rules (what can come BEFORE each vowel):
        //   a  <- e, i, u
        //   e  <- a, i
        //   i  <- e, o
        //   o  <- i
        //   u  <- i, o
        na := (e + i + u) % MOD
        ne := (a + i) % MOD
        ni := (e + o) % MOD
        no := i % MOD
        nu := (i + o) % MOD
        a, e, i, o, u = na, ne, ni, no, nu
    }
    return (a + e + i + o + u) % MOD
}
```

### Rust

```
impl Solution {
    pub fn count_vowel_permutation(n: i32) -> i32 {
        const MOD: u64 = 1_000_000_007;
        // Counts of valid strings of current length ending in a, e, i, o, u
        let (mut a, mut e, mut i, mut o, mut u) = (1u64, 1, 1, 1, 1);

        for _ in 1..n {
            // Each vowel's new count = sum of counts of vowels that can precede it
            // Rules (what can come BEFORE each vowel):
            //   a  <- e, i, u
            //   e  <- a, i
            //   i  <- e, o
            //   o  <- i
            //   u  <- i, o
            let (na, ne, ni, no, nu) = (
                (e + i + u) % MOD,
                (a + i) % MOD,
                (e + o) % MOD,
                i % MOD,
                (i + o) % MOD,
            );
            (a, e, i, o, u) = (na, ne, ni, no, nu);
        }
        ((a + e + i + o + u) % MOD) as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n = 1`
Every single vowel is a valid string of length 1. Count = $5$.

### 2. Slightly Complex
**Input:** `n = 2`
Propagate once: $a=3, e=2, i=2, o=1, u=2$. Sum = **10** valid two-vowel strings.

### 3. Edge Case: Time Factor
**Input:** `n = 20000`
All 19 999 propagation steps must execute; no early exit is possible. Each step is 5 additions modulo $10^9+7$, totalling exactly $O(n)$ work.

### 4. Edge Case: Space Factor
**Input:** `n = 20000`
Only 5 integer variables are maintained regardless of input size — $O(1)$ space. No array or table grows with `n`.

### 5. Almost-Impossible but Plausible
**Input:** `n = 19999`
The accumulated counts can approach $(10^9+7)^5$ before modular reduction, but since each variable stays $< 10^9+7 \approx 10^9$, intermediate sums fit comfortably in a 64-bit integer ($\leq 3 \times 10^9 < 2^{32}$), making overflow a non-issue.
