# LeetCode #1232: Check If It Is a Straight Line

https://leetcode.com/problems/check-if-it-is-a-straight-line/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n)$ | $O(1)$ |
| **Optimal: Cross Product ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Compute the slope between the first two points as a fraction and compare each subsequent pair. Division and floating-point comparisons risk precision errors.

### Optimal: Cross Product ★
Fix vectors `dx = x1-x0, dy = y1-y0` from the first two points. For every subsequent point `(xi, yi)`, check that the cross product `(xi-x0)*dy == (yi-y0)*dx` equals zero using integer arithmetic — no division, no floating-point rounding.

**Constraints:**
* `2 <= coordinates.length <= 1000`
* `coordinates[i].length == 2`
* `-10^4 <= coordinates[i][0], coordinates[i][1] <= 10^4`
* All coordinates are distinct.

## Solutions

### C#

```
public class Solution {
    public bool CheckStraightLine(int[][] coordinates) {
        // Reference direction vector from first two points
        int dx = coordinates[1][0] - coordinates[0][0];
        int dy = coordinates[1][1] - coordinates[0][1];

        for (int i = 2; i < coordinates.Length; i++) {
            int ex = coordinates[i][0] - coordinates[0][0];
            int ey = coordinates[i][1] - coordinates[0][1];
            // Cross product must be zero for collinearity — avoids division entirely
            if (ex * dy != ey * dx) return false;
        }
        return true;
    }
}
```

### Python

```
class Solution:
    def checkStraightLine(self, coordinates: list[list[int]]) -> bool:
        # Reference direction from the first two points
        dx = coordinates[1][0] - coordinates[0][0]
        dy = coordinates[1][1] - coordinates[0][1]

        for x, y in coordinates[2:]:
            ex = x - coordinates[0][0]
            ey = y - coordinates[0][1]
            # Cross product zero ↔ vectors are parallel ↔ points are collinear
            if ex * dy != ey * dx:
                return False
        return True
```

### Go

```
func checkStraightLine(coordinates [][]int) bool {
    // Direction vector anchored at the first point
    dx := coordinates[1][0] - coordinates[0][0]
    dy := coordinates[1][1] - coordinates[0][1]

    for _, p := range coordinates[2:] {
        ex := p[0] - coordinates[0][0]
        ey := p[1] - coordinates[0][1]
        // Cross product = 0 confirms collinearity without floating-point division
        if ex*dy != ey*dx {
            return false
        }
    }
    return true
}
```

### Rust

```
impl Solution {
    pub fn check_straight_line(coordinates: Vec<Vec<i32>>) -> bool {
        // Reference direction from point 0 to point 1
        let dx = coordinates[1][0] - coordinates[0][0];
        let dy = coordinates[1][1] - coordinates[0][1];

        for p in &coordinates[2..] {
            let ex = p[0] - coordinates[0][0];
            let ey = p[1] - coordinates[0][1];
            // Integer cross product avoids division and floating-point error
            if ex * dy != ey * dx {
                return false;
            }
        }
        true
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `coordinates = [[1,2],[2,3],[3,4],[4,5]]`
`dx=1, dy=1`. For `(3,4)`: cross = $(3-1)\cdot1 - (4-2)\cdot1 = 0$. All zero — returns **true**.

### 2. Slightly Complex
**Input:** `coordinates = [[1,1],[2,2],[3,4]]`
`dx=1, dy=1`. For `(3,4)`: $(3-1)\cdot1 - (4-1)\cdot1 = 2-3 = -1 \neq 0$ — returns **false**.

### 3. Edge Case: Time Factor
**Input:** 1000 collinear points along `y = 3x`.
The loop runs 998 cross-product checks in $O(n)$, each an integer multiply — no slowdown from large $n$.

### 4. Edge Case: Space Factor
**Input:** Any 1000-point input.
Only `dx, dy, ex, ey` are stored — constant $O(1)$ extra space regardless of input size.

### 5. Almost-Impossible but Plausible
**Input:** `coordinates = [[0,0],[10000,10000],[-10000,-10000]]`
`dx=10000, dy=10000`. For `(-10000,-10000)`: $(-10000-0)\cdot10000 = -10^8$; $(-10000-0)\cdot10000 = -10^8$. Equal — **true**. Values stay within 32-bit range.
