# LeetCode #1243: Array Transformation

https://leetcode.com/problems/array-transformation/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Simulation with Early Exit ★** | $O(n^2)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Always run a fixed number of iterations (e.g. $n$ rounds), even if the array has already stabilised. Wastes time on no-op passes.

### Optimal: Simulation with Early Exit ★
Apply the local-min/max rule each round (increment a local minimum, decrement a local maximum), but track a `changed` flag. Stop as soon as a complete pass produces no changes — the array has reached a fixed point.

**Why this is better than Brute Force:** Identical worst-case, but exits immediately when the array stabilises, which often happens in far fewer than $n$ rounds.

**Constraints:**
* $1 \leq n \leq 100$
* $1 \leq \text{arr}[i] \leq 100$
* First and last elements are never modified


## Solutions

### C#

```
public class Solution {
    public IList<int> TransformArray(int[] arr) {
        bool changed = true;
        while (changed) {
            changed = false;
            // Work on a copy so this round's decisions don't affect each other
            int[] next = (int[])arr.Clone();
            for (int i = 1; i < arr.Length - 1; i++) {
                if (arr[i] < arr[i - 1] && arr[i] < arr[i + 1]) {
                    next[i]++;   // local minimum nudges upward
                    changed = true;
                } else if (arr[i] > arr[i - 1] && arr[i] > arr[i + 1]) {
                    next[i]--;   // local maximum nudges downward
                    changed = true;
                }
            }
            arr = next;
        }
        return arr;
    }
}
```

### Python

```
class Solution:
    def transformArray(self, arr: list[int]) -> list[int]:
        changed = True
        while changed:
            changed = False
            # Snapshot the current round so updates don't cascade within one pass
            nxt = arr[:]
            for i in range(1, len(arr) - 1):
                if arr[i] < arr[i - 1] and arr[i] < arr[i + 1]:
                    nxt[i] += 1   # local minimum rises toward its neighbors
                    changed = True
                elif arr[i] > arr[i - 1] and arr[i] > arr[i + 1]:
                    nxt[i] -= 1   # local maximum falls toward its neighbors
                    changed = True
            arr = nxt
        return arr
```

### Rust

```
impl Solution {
    pub fn transform_array(arr: Vec<i32>) -> Vec<i32> {
        let mut arr = arr;
        let mut changed = true;
        while changed {
            changed = false;
            // Snapshot the array so each round reads pre-update values
            let prev = arr.clone();
            for i in 1..arr.len() - 1 {
                if prev[i] < prev[i - 1] && prev[i] < prev[i + 1] {
                    arr[i] += 1; // local minimum steps up toward its neighbors
                    changed = true;
                } else if prev[i] > prev[i - 1] && prev[i] > prev[i + 1] {
                    arr[i] -= 1; // local maximum steps down toward its neighbors
                    changed = true;
                }
            }
        }
        arr
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [6, 2, 3, 4]`
Round 1: index 1 is local min (2 < 6, 2 < 3) → becomes 3. Array: `[6,3,3,4]`. Round 2: index 1 is equal on left (3==3), not a strict local min — no change. Stabilised → return `[6,3,3,4]`.

### 2. Slightly Complex
**Input:** `arr = [1, 6, 3, 4, 3, 5]`
Round 1: index 1 is local max (6>1,6>3) → 5; index 3 is local max (4>3,4>3) → 3. Array: `[1,5,3,3,3,5]`. Round 2: index 1 is still local max → 4. Continues converging. Eventually stabilises at `[1,4,4,4,4,5]` (values level toward the non-extremes).

### 3. Edge Case: Time Factor
**Input:** `arr = [1, 100, 1, 100, ..., 1]` (alternating extremes, $n = 100$)
Every interior element oscillates by 1 each round. Worst case approaches $\sim 50$ rounds before all interior values flatten. This is the maximum-iteration scenario confirming $O(n^2)$ total work.

### 4. Edge Case: Space Factor
**Input:** `arr = [5]` or `arr = [5, 5]` ($n = 1$ or $n = 2$)
No interior indices to process (loop from 1 to $n-2$ is empty). `changed` stays false immediately; the while-loop exits after one iteration. Returns input unchanged. One clone per pass — $O(n)$ space with $n \leq 2$.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [50, 50, 50, ..., 50]` (all identical, $n = 100$)
No element is a strict local min or max (all equal to neighbours). `changed` stays false on the very first pass; exits immediately without modification. Confirms the algorithm doesn't trigger on ties.
