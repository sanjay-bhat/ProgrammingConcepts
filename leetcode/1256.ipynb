# LeetCode #1256: Encode Number

https://leetcode.com/problems/encode-number/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n)$ | $O(n)$ |
| **Optimal: Bit Manipulation ★** | $O(\log n)$ | $O(\log n)$ |

---

## Understanding the Methods

### Brute Force
Simulate the encoding by iterating from 0 upward, building strings according to the pattern rule (0 maps to "", 1 maps to "0", 2 maps to "1", ...). Stop when the target index is reached. Linear in $n$.

### Optimal: Bit Manipulation ★
The encoding maps $n$ to the binary representation of $n+1$ with the leading 1 stripped. For example, $n=5$: $5+1=6=110_2$, remove leading 1 → "10". Compute by converting $n+1$ to binary and dropping the first character — $O(\log n)$ and no simulation loop needed.

**Constraints:**
* $0 \leq num \leq 10^9$

## Solutions

### C#

```
public class Solution {
    public string Encode(int num) {
        // n+1 in binary, drop the leading '1' — that's the encoded string
        return Convert.ToString(num + 1, 2).Substring(1);
    }
}
```

### Python

```
class Solution:
    def encode(self, num: int) -> str:
        # n+1 in binary, drop the leading '1' — that's the encoded string
        return bin(num + 1)[3:]  # bin() returns '0b...', skip '0b1' prefix
```

### Go

```
import "strconv"
func encode(num int) string {
    // n+1 in binary, drop the leading '1' — that's the encoded string
    s := strconv.FormatInt(int64(num+1), 2)
    return s[1:]
}
```

### Rust

```
impl Solution {
    pub fn encode(num: i32) -> String {
        // n+1 in binary, drop the leading '1' — that's the encoded string
        let s = format!("{:b}", num + 1);
        s[1..].to_string()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `num = 5`
$5 + 1 = 6 = 110_2$. Drop leading 1 → **"10"**. The encoding table confirms: 0→"", 1→"0", 2→"1", 3→"00", 4→"01", 5→"10". Answer: **"10"**.

### 2. Slightly Complex
**Input:** `num = 23`
$23 + 1 = 24 = 11000_2$. Drop leading 1 → **"1000"**. Four-bit output for a number in the range $[16, 31]$.

### 3. Edge Case: Time Factor
**Input:** `num = 10^9`.
$10^9 + 1 \approx 2^{30}$; its binary representation has 30 bits. Stripping the leading 1 yields a 29-character string. `strconv.FormatInt` runs in $O(\log n) = O(30)$ steps.

### 4. Edge Case: Space Factor
**Input:** `num = 0`.
$0 + 1 = 1 = 1_2$. Drop leading 1 → **""** (empty string). Minimum possible output — zero characters allocated.

### 5. Almost-Impossible but Plausible
**Input:** `num = 2^{30} - 1 = 1{,}073{,}741{,}823` (just under $10^9$ constraint).
$2^{30} = 10...0_2$ (31 bits). Drop leading 1 → 30 zeros: **"000...0"** (30 chars). This is the maximum output length within the constraint, and the result looks deceptively like an all-zero encoding despite the large input.
