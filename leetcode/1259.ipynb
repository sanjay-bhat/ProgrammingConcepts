# LeetCode #1259: Handshakes That Don't Cross

https://leetcode.com/problems/handshakes-that-dont-cross/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (Recursion)** | $O(4^{n/2})$ | $O(n)$ |
| **Optimal: Catalan DP ★** | $O(n^2)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force (Recursion)
Enumerate all perfect matchings by recursively pairing person 0 with each possible even-indexed person $2k$, then recursing on the two groups. This visits $C_{n/2}$ configurations — the Catalan number, which grows as $4^{n/2}$ — without memoization.

### Optimal: Catalan DP ★
The number of non-crossing handshakes among $2n$ people arranged in a circle equals the $n$-th Catalan number $C_n$. Use the DP recurrence $C_n = \sum_{k=0}^{n-1} C_k \cdot C_{n-1-k}$ with $C_0 = 1$. Fill the table bottom-up in $O(n^2)$ time and $O(n)$ space. All results are taken modulo $10^9 + 7$.

**Constraints:**
* `num_people` is even
* $2 \leq num\_people \leq 1000$

## Solutions

### C#

```
public class Solution {
    const int MOD = 1_000_000_007;
    public int NumberOfWays(int numPeople) {
        int n = numPeople / 2;
        // catalan[i] = number of non-crossing perfect matchings for 2i people
        var catalan = new long[n + 1];
        catalan[0] = 1; // base case: zero people, one empty matching
        for (int i = 1; i <= n; i++)
            for (int k = 0; k < i; k++)
                // Person 0 pairs with person 2k+1; left group has k pairs, right has i-1-k
                catalan[i] = (catalan[i] + catalan[k] * catalan[i - 1 - k]) % MOD;
        return (int)catalan[n];
    }
}
```

### Python

```
class Solution:
    def number_of_ways(self, num_people: int) -> int:
        MOD = 10**9 + 7
        n = num_people // 2
        # catalan[i] = number of non-crossing perfect matchings for 2i people
        catalan = [0] * (n + 1)
        catalan[0] = 1  # base case: zero people, one empty matching
        for i in range(1, n + 1):
            for k in range(i):
                # Person 0 pairs with person 2k+1; left group has k pairs, right has i-1-k
                catalan[i] = (catalan[i] + catalan[k] * catalan[i - 1 - k]) % MOD
        return catalan[n]
```

### Go

### Rust

```
impl Solution {
    pub fn number_of_ways(num_people: i32) -> i32 {
        const MOD: i64 = 1_000_000_007;
        let n = (num_people / 2) as usize;
        // catalan[i] = number of non-crossing perfect matchings for 2i people
        let mut catalan = vec![0i64; n + 1];
        catalan[0] = 1; // base case: zero people, one empty matching
        for i in 1..=n {
            for k in 0..i {
                // Person 0 pairs with person 2k+1; left group has k pairs, right has i-1-k
                catalan[i] = (catalan[i] + catalan[k] * catalan[i - 1 - k]) % MOD;
            }
        }
        catalan[n] as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `num_people = 4`
$n = 2$. $C_2 = C_0 \cdot C_1 + C_1 \cdot C_0 = 1 \cdot 1 + 1 \cdot 1 = 2$. Two non-crossing matchings: (0,1)(2,3) and (0,3)(1,2). Answer: **2**.

### 2. Slightly Complex
**Input:** `num_people = 6`
$n = 3$. $C_3 = C_0 C_2 + C_1 C_1 + C_2 C_0 = 2 + 1 + 2 = 5$. Answer: **5**.

### 3. Edge Case: Time Factor
**Input:** `num_people = 1000` ($n = 500$).
The DP fills $\sum_{i=1}^{500} i = 125{,}250$ cells. Each is one multiplication and one modular reduction — roughly $125{,}000$ operations, well within $O(n^2)$.

### 4. Edge Case: Space Factor
**Input:** `num_people = 1000`.
The `catalan` array holds exactly $501$ 64-bit integers — minimal $O(n)$ space, independent of the intermediate products (which never need to be stored simultaneously).

### 5. Almost-Impossible but Plausible
**Input:** `num_people = 2` ($n = 1$).
$C_1 = C_0 \cdot C_0 = 1$. Only one handshake possible between the two people, and it trivially doesn't cross itself. The DP loop runs exactly once. Answer: **1**.
