# LeetCode #1262: Greatest Sum Divisible by Three

https://leetcode.com/problems/greatest-sum-divisible-by-three/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^n)$ | $O(n)$ |
| **Optimal: Remainder-State DP ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Enumerate every subset, compute its sum, keep the maximum divisible by 3. Exponential — useless beyond ~25 elements.

### Optimal: Remainder-State DP ★
Maintain three running values: `dp[r]` = maximum sum achievable so far whose remainder modulo 3 equals `r` (r = 0, 1, 2). For each new number, attempt to extend each existing state and update whichever remainder bucket improves. `dp[0]` is the answer.

**Why this is better than Brute Force:** We collapse all subset possibilities into 3 scalars, processing each element exactly once in constant work.

**Constraints:**
* $1 \leq$ `nums.length` $\leq 4 \times 10^4$
* $1 \leq$ `nums[i]` $\leq 10^4$

## Solutions

### C#

```
public class Solution {
    public int MaxSumDivThree(int[] nums) {
        // dp[r] = largest achievable sum with remainder r modulo 3
        int[] dp = new int[3]; // dp[0]=0, dp[1]=-inf, dp[2]=-inf (use 0; adjust below)

        foreach (int num in nums) {
            // Snapshot to avoid using updated values in the same pass
            int[] next = (int[])dp.Clone();
            for (int r = 0; r < 3; r++) {
                // Extending state r by num changes remainder to (r + num%3) % 3
                int newR = (r + num % 3) % 3;
                // Keep the largest sum for each remainder bucket
                next[newR] = Math.Max(next[newR], dp[r] + num);
            }
            dp = next;
        }
        return dp[0]; // Maximum sum divisible by 3
    }
}
```

### Python

```
from typing import List

class Solution:
    def max_sum_div_three(self, nums: List[int]) -> int:
        # dp[r] = largest achievable sum with remainder r modulo 3
        dp = [0, 0, 0]

        for num in nums:
            # Snapshot to avoid using updated values in the same pass
            next_dp = dp[:]
            for r in range(3):
                # Extending state r by num changes remainder to (r + num % 3) % 3
                new_r = (r + num % 3) % 3
                # Keep the largest sum for each remainder bucket
                next_dp[new_r] = max(next_dp[new_r], dp[r] + num)
            dp = next_dp

        return dp[0]  # Maximum sum divisible by 3
```

### Go

```
func maxSumDivThree(nums []int) int {
    // dp[r] = largest achievable sum with remainder r modulo 3
    dp := [3]int{}

    for _, num := range nums {
        // Snapshot to avoid using updated values in the same pass
        next := dp
        for r := 0; r < 3; r++ {
            // Extending state r by num changes remainder to (r + num%3) % 3
            newR := (r + num%3) % 3
            // Keep the largest sum for each remainder bucket
            if dp[r]+num > next[newR] {
                next[newR] = dp[r] + num
            }
        }
        dp = next
    }
    return dp[0] // Maximum sum divisible by 3
}
```

### Rust

```
impl Solution {
    pub fn max_sum_div_three(nums: Vec<i32>) -> i32 {
        // dp[r] = largest achievable sum with remainder r modulo 3
        let mut dp = [0i32; 3];

        for num in nums {
            // Snapshot to avoid using updated values in the same pass
            let mut next = dp;
            for r in 0..3 {
                // Extending state r by num changes remainder to (r + num%3) % 3
                let new_r = ((r as i32 + num % 3) % 3) as usize;
                // Keep the largest sum for each remainder bucket
                next[new_r] = next[new_r].max(dp[r] + num);
            }
            dp = next;
        }
        dp[0] // Maximum sum divisible by 3
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [3, 6, 5, 1, 8]`
Take 3+6+5+1+8=23 (remainder 2). Skip 8: 3+6+5+1=15 ✓. Or skip 5: 3+6+1+8=18 ✓. Best: **18**.

### 2. Slightly Complex
**Input:** `nums = [4]`
Only one element; 4 % 3 = 1, not divisible. dp stays [0,4,0]. Answer: **0** (empty subset).

### 3. Edge Case: Time Factor
**Input:** `nums = [10000] * 40000`
All elements have the same value and remainder. The DP runs 40 000 iterations, each doing 3 comparisons — worst case for throughput, not for code paths.

### 4. Edge Case: Space Factor
**Input:** `nums` of length 40 000
Only 3 integers are maintained regardless of input size — $O(1)$ space. The snapshot array never grows.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [1, 1]`
Both have remainder 1. Sum = 2 (remainder 2). Neither alone nor together is divisible by 3. Answer: **0**. The DP correctly returns dp[0]=0 because no non-empty subset sums to a multiple of 3.
