# LeetCode #1266: Minimum Time Visiting All Points

https://leetcode.com/problems/minimum-time-visiting-all-points/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n)$ | $O(1)$ |
| **Optimal: Chebyshev Distance ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Simulate diagonal then horizontal/vertical steps between consecutive points, counting each step individually. Same complexity but more bookkeeping.

### Optimal: Chebyshev Distance ★
In a grid where diagonal moves cost 1 second (same as horizontal/vertical), the time to go from point A to point B is `max(|Δx|, |Δy|)` — the Chebyshev distance. Sum this over all consecutive pairs for the answer in a single pass.

**Constraints:**
* `1 <= points.length <= 100`
* `points[i].length == 2`
* `-1000 <= points[i][0], points[i][1] <= 1000`

## Solutions

### C#

```
public class Solution {
    public int MinTimeToVisitAllPoints(int[][] points) {
        int total = 0;
        for (int i = 1; i < points.Length; i++) {
            int dx = Math.Abs(points[i][0] - points[i-1][0]);
            int dy = Math.Abs(points[i][1] - points[i-1][1]);
            // Chebyshev distance: diagonals handle both axes simultaneously
            total += Math.Max(dx, dy);
        }
        return total;
    }
}
```

### Python

```
class Solution:
    def minTimeToVisitAllPoints(self, points: list[list[int]]) -> int:
        total = 0
        for i in range(1, len(points)):
            dx = abs(points[i][0] - points[i-1][0])
            dy = abs(points[i][1] - points[i-1][1])
            # Chebyshev distance: diagonal moves eat both x and y offset at once
            total += max(dx, dy)
        return total
```

### Go

```
func minTimeToVisitAllPoints(points [][]int) int {
    total := 0
    for i := 1; i < len(points); i++ {
        dx := points[i][0] - points[i-1][0]
        if dx < 0 { dx = -dx }
        dy := points[i][1] - points[i-1][1]
        if dy < 0 { dy = -dy }
        // Chebyshev distance: max of absolute differences
        if dx > dy {
            total += dx
        } else {
            total += dy
        }
    }
    return total
}
```

### Rust

```
impl Solution {
    pub fn min_time_to_visit_all_points(points: Vec<Vec<i32>>) -> i32 {
        let mut total = 0;
        for i in 1..points.len() {
            let dx = (points[i][0] - points[i-1][0]).abs();
            let dy = (points[i][1] - points[i-1][1]).abs();
            // Chebyshev distance: diagonal moves cover both axes simultaneously
            total += dx.max(dy);
        }
        total
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `points = [[1,1],[3,4],[-1,0]]`
Leg 1: $|2|,|3| \rightarrow \max(2,3)=3$. Leg 2: $|4|,|4| \rightarrow 4$. Total = **7**.

### 2. Slightly Complex
**Input:** `points = [[3,2],[-2,2]]`
$dx=5, dy=0 \rightarrow \max=5$. Walk 5 steps left. Total = **5**.

### 3. Edge Case: Time Factor
**Input:** 100 points all at `(-1000,-1000)` and `(1000,1000)` alternating.
99 legs each with $dx=dy=2000 \rightarrow \max=2000$. Answer $= 99 \times 2000 = 198000$, computed in $O(n)$.

### 4. Edge Case: Space Factor
**Input:** Any 100 points.
Two integer variables `dx, dy` per iteration — $O(1)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `points = [[0,0],[1000,0],[1000,1000],[0,1000],[0,0]]`
Legs: 1000, 1000, 1000, 1000. Total = **4000**. Starting and ending at origin forces full perimeter traversal.
