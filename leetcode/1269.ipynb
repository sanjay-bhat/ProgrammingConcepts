# LeetCode #1269: Number of Ways to Stay in the Same Place After Some Steps

https://leetcode.com/problems/number-of-ways-to-stay-in-the-same-place-after-some-steps/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (Recursion)** | $O(3^{steps})$ | $O(steps)$ |
| **Optimal: DP with Rolling Array ★** | $O(steps \times \min(steps, arrLen))$ | $O(\min(steps, arrLen))$ |

---

## Understanding the Methods

### Brute Force (Recursion)
At each position, try all three moves (left, stay, right) recursively, counting sequences that end at index 0 after exactly `steps` moves. With no memoization, $O(3^{steps})$ paths are explored.

### Optimal: DP with Rolling Array ★
Let `dp[i]` = number of ways to be at index $i$ after the current number of steps. Update with a rolling array over `steps` iterations. The reachable index range is bounded by $\min(steps, arrLen - 1)$, keeping both time and space proportional to this smaller value.

**Constraints:**
* $1 \leq steps \leq 500$
* $1 \leq arrLen \leq 10^6$

## Solutions

### C#

```
public class Solution {
    const int MOD = 1_000_000_007;
    public int NumWays(int steps, int arrLen) {
        // Reachable positions are bounded by steps (can't go further in steps moves)
        int maxPos = Math.Min(steps / 2 + 1, arrLen);
        var dp = new long[maxPos];
        dp[0] = 1; // start at index 0
        for (int s = 0; s < steps; s++) {
            var next = new long[maxPos];
            for (int i = 0; i < maxPos; i++) {
                if (dp[i] == 0) continue;
                // Stay at i
                next[i] = (next[i] + dp[i]) % MOD;
                // Move right
                if (i + 1 < maxPos) next[i + 1] = (next[i + 1] + dp[i]) % MOD;
                // Move left
                if (i > 0) next[i - 1] = (next[i - 1] + dp[i]) % MOD;
            }
            dp = next;
        }
        return (int)dp[0];
    }
}
```

### Python

```
class Solution:
    def num_ways(self, steps: int, arr_len: int) -> int:
        MOD = 10**9 + 7
        # Reachable positions are bounded by steps (can't go further in steps moves)
        max_pos = min(steps // 2 + 1, arr_len)
        dp = [0] * max_pos
        dp[0] = 1  # start at index 0
        for _ in range(steps):
            nxt = [0] * max_pos
            for i in range(max_pos):
                if not dp[i]: continue
                # Stay at i
                nxt[i] = (nxt[i] + dp[i]) % MOD
                # Move right
                if i + 1 < max_pos: nxt[i + 1] = (nxt[i + 1] + dp[i]) % MOD
                # Move left
                if i > 0: nxt[i - 1] = (nxt[i - 1] + dp[i]) % MOD
            dp = nxt
        return dp[0]
```

### Rust

```
impl Solution {
    pub fn num_ways(steps: i32, arr_len: i32) -> i32 {
        const MOD: i64 = 1_000_000_007;
        // Reachable positions are bounded by steps (can't go further in steps moves)
        let max_pos = ((steps / 2 + 1) as usize).min(arr_len as usize);
        let mut dp = vec![0i64; max_pos];
        dp[0] = 1; // start at index 0
        for _ in 0..steps {
            let mut next = vec![0i64; max_pos];
            for i in 0..max_pos {
                if dp[i] == 0 { continue; }
                // Stay at i
                next[i] = (next[i] + dp[i]) % MOD;
                // Move right
                if i + 1 < max_pos { next[i+1] = (next[i+1] + dp[i]) % MOD; }
                // Move left
                if i > 0 { next[i-1] = (next[i-1] + dp[i]) % MOD; }
            }
            dp = next;
        }
        dp[0] as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `steps = 3`, `arrLen = 2`
From index 0, explore all 3-step paths on an array of length 2. Only positions 0 and 1 are reachable. The DP finds 4 valid paths back to 0: stay-stay-stay, right-left-stay, stay-right-left, right-stay-left. Answer: **4**.

### 2. Slightly Complex
**Input:** `steps = 2`, `arrLen = 4`
Bounded to $\min(2, 4) = 2$ reachable positions (0 and 1). Only three 2-step sequences return to 0: stay-stay, right-left, left... wait, can't go left from 0. So: stay-stay and right-left. Answer: **2**.

### 3. Edge Case: Time Factor
**Input:** `steps = 500`, `arrLen = 10^6`.
`maxPos = 500/2 + 1 = 251`. Each of 500 steps processes 251 positions: $500 \times 251 = 125{,}500$ transitions. The $10^6$ array length is irrelevant — the cap dominates.

### 4. Edge Case: Space Factor
**Input:** `steps = 500`, `arrLen = 1`.
`maxPos = 1`. The DP is a single cell: only "stay" is valid, so `dp[0] = 1` at every step. Rolling arrays hold 1 element each. Answer: **1**.

### 5. Almost-Impossible but Plausible
**Input:** `steps = 500`, `arrLen = 251`.
Both bounds give the same `maxPos = 251`. Every reachable position is used, and the DP must propagate through all 251 cells for each of 500 steps. The answer is a huge number modulo $10^9 + 7$, but the algorithm handles it cleanly with modular arithmetic at every addition.
