# LeetCode #1277: Count Square Submatrices with All Ones

https://leetcode.com/problems/count-square-submatrices-with-all-ones/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^4)$ | $O(1)$ |
| **Optimal: DP ★** | $O(m \times n)$ | $O(m \times n)$ |

---

## Understanding the Methods

### Brute Force
For every possible top-left corner and every possible size, scan the entire sub-square to check if it contains all ones. The extra scan per candidate raises the cost to $O(n^4)$.

### Optimal: DP ★
`dp[i][j]` = side length of the largest all-ones square with its bottom-right corner at `(i,j)`. The recurrence is `dp[i][j] = min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1]) + 1` when `matrix[i][j] == 1`. The answer is the sum of all `dp` values, since a cell with value $k$ contributes one square of each side from 1 to $k$.

**Constraints:**
* $1 \leq m, n \leq 300$
* `matrix[i][j]` is `0` or `1`

## Solutions

### C#

```
public class Solution {
    public int CountSquares(int[][] matrix) {
        int m = matrix.Length, n = matrix[0].Length, ans = 0;
        // dp[i][j] = side of largest all-ones square with bottom-right at (i,j)
        var dp = new int[m, n];
        for (int i = 0; i < m; i++) {
            for (int j = 0; j < n; j++) {
                if (matrix[i][j] == 0) continue;
                if (i == 0 || j == 0) {
                    dp[i, j] = 1; // border cells can only anchor 1x1 squares
                } else {
                    // Extend the smallest neighbouring square in all three directions
                    dp[i, j] = Math.Min(dp[i-1, j], Math.Min(dp[i, j-1], dp[i-1, j-1])) + 1;
                }
                ans += dp[i, j]; // each dp value counts squares of all valid sizes
            }
        }
        return ans;
    }
}
```

### Python

```
class Solution:
    def count_squares(self, matrix: list[list[int]]) -> int:
        m, n = len(matrix), len(matrix[0])
        # dp[i][j] = side of largest all-ones square with bottom-right at (i,j)
        dp = [[0]*n for _ in range(m)]
        ans = 0
        for i in range(m):
            for j in range(n):
                if matrix[i][j] == 0: continue
                if i == 0 or j == 0:
                    dp[i][j] = 1  # border cells can only anchor 1x1 squares
                else:
                    # Extend the smallest neighbouring square in all three directions
                    dp[i][j] = min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1]) + 1
                ans += dp[i][j]  # each dp value counts squares of all valid sizes
        return ans
```

### Go

```
func countSquares(matrix [][]int) int {
    m, n, ans := len(matrix), len(matrix[0]), 0
    // dp[i][j] = side of largest all-ones square with bottom-right at (i,j)
    dp := make([][]int, m)
    for i := range dp { dp[i] = make([]int, n) }
    for i := 0; i < m; i++ {
        for j := 0; j < n; j++ {
            if matrix[i][j] == 0 { continue }
            if i == 0 || j == 0 {
                dp[i][j] = 1 // border cells can only anchor 1x1 squares
            } else {
                // Extend the smallest neighbouring square in all three directions
                mn := dp[i-1][j]; if dp[i][j-1] < mn { mn = dp[i][j-1] }
                if dp[i-1][j-1] < mn { mn = dp[i-1][j-1] }
                dp[i][j] = mn + 1
            }
            ans += dp[i][j] // each dp value counts squares of all valid sizes
        }
    }
    return ans
}
```

### Rust

```
impl Solution {
    pub fn count_squares(matrix: Vec<Vec<i32>>) -> i32 {
        let (m, n) = (matrix.len(), matrix[0].len());
        // dp[i][j] = side of largest all-ones square with bottom-right at (i,j)
        let mut dp = vec![vec![0i32; n]; m];
        let mut ans = 0i32;
        for i in 0..m {
            for j in 0..n {
                if matrix[i][j] == 0 { continue; }
                if i == 0 || j == 0 {
                    dp[i][j] = 1; // border cells can only anchor 1x1 squares
                } else {
                    // Extend the smallest neighbouring square in all three directions
                    dp[i][j] = dp[i-1][j].min(dp[i][j-1]).min(dp[i-1][j-1]) + 1;
                }
                ans += dp[i][j]; // each dp value counts squares of all valid sizes
            }
        }
        ans
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `matrix = [[0,1,1,1],[1,1,1,1],[0,1,1,1]]`
DP values at each cell of ones: row 0 → [0,1,1,1]; row 1 → [1,1,2,2]; row 2 → [0,1,2,3]. Sum = $1+1+1+1+1+2+2+1+2+3 = 15$. Answer: **15**.

### 2. Slightly Complex
**Input:** `matrix = [[1,0,1],[1,1,0],[1,1,0]]`
DP: [1,0,1], [1,1,0], [1,2,0]. Sum = $1+1+1+1+2 = 6$. Answer: **6**.

### 3. Edge Case: Time Factor
**Input:** $300 \times 300$ matrix of all ones.
Every cell computes `min(3 values) + 1` once — $90{,}000$ operations. The DP value at `(299,299)` is 300, and the total sum equals $\sum_{k=1}^{300} k \times (300-k+1)^2$, a large number computed exactly.

### 4. Edge Case: Space Factor
**Input:** $300 \times 300$ matrix.
The `dp` array uses $300 \times 300 = 90{,}000$ integers — $O(m \times n)$ at the constraint maximum. An in-place variant would reuse `matrix` itself, but the two-array form is clearer.

### 5. Almost-Impossible but Plausible
**Input:** $300 \times 300$ matrix where only the last row and last column contain ones (an L-shape border).
Every border cell gets `dp = 1`, contributing only $1 \times 1$ squares. The `min` recurrence never exceeds 1 because the inner corners lack top-left neighbours. Total: $300 + 299 = 599$ squares.
