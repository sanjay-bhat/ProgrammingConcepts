# LeetCode #1299: Replace Elements with Greatest Element on Right Side

https://leetcode.com/problems/replace-elements-with-greatest-element-on-right-side/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Reverse Linear Scan ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
For each position $i$, scan every element to its right to find the maximum, then replace `arr[i]` with that maximum. Two nested loops make this $O(n^2)$.

### Optimal: Reverse Linear Scan ★
Scan from right to left, maintaining a running maximum of everything seen so far. At each position, write the current running max into `arr[i]`, then extend the max with the original value before it was overwritten. One pass, no extra space.

**Why this is better than Brute Force:** The running max eliminates the inner scan — each element is visited exactly once, dropping $O(n^2)$ to $O(n)$.

**Constraints:**
* $1 \leq arr.length \leq 10^4$
* $1 \leq arr[i] \leq 10^5$


## Solutions

### C#

```
public class Solution {
    public int[] ReplaceElements(int[] arr) {
        int maxSoFar = -1; // Last element always becomes -1
        // Scan right-to-left so the running max covers all elements to the right
        for (int i = arr.Length - 1; i >= 0; i--) {
            int original = arr[i];           // Save before overwriting
            arr[i] = maxSoFar;               // Replace with max of everything to the right
            maxSoFar = Math.Max(maxSoFar, original); // Extend the rightward max
        }
        return arr;
    }
}
```

### Python

```
class Solution:
    def replaceElements(self, arr: list[int]) -> list[int]:
        max_so_far = -1  # Last element always becomes -1
        # Scan right-to-left so the running max covers all elements to the right
        for i in range(len(arr) - 1, -1, -1):
            original = arr[i]                        # Save before overwriting
            arr[i] = max_so_far                      # Replace with max of everything to the right
            max_so_far = max(max_so_far, original)   # Extend the rightward max
        return arr
```

### Go

```
func replaceElements(arr []int) []int {
    maxSoFar := -1 // Last element always becomes -1
    // Scan right-to-left so the running max covers all elements to the right
    for i := len(arr) - 1; i >= 0; i-- {
        original := arr[i]           // Save before overwriting
        arr[i] = maxSoFar            // Replace with max of everything to the right
        if original > maxSoFar {
            maxSoFar = original      // Extend the rightward max
        }
    }
    return arr
}
```

### Rust

```
impl Solution {
    pub fn replace_elements(mut arr: Vec<i32>) -> Vec<i32> {
        let mut max_so_far = -1i32; // Last element always becomes -1
        // Scan right-to-left so the running max covers all elements to the right
        for i in (0..arr.len()).rev() {
            let original = arr[i];               // Save before overwriting
            arr[i] = max_so_far;                 // Replace with max of everything to the right
            max_so_far = max_so_far.max(original); // Extend the rightward max
        }
        arr
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [17, 18, 5, 4, 6, 1]`
Right-to-left: max starts at -1. Position 5 → -1 (max=1). Position 4 → 1 (max=6). Position 3 → 6 (max=6). Position 2 → 6 (max=6). Position 1 → 6 (max=18). Position 0 → 18. Result: `[18, 6, 6, 6, 1, -1]`.

### 2. Slightly Complex
**Input:** `arr = [400, 3, 100, 50, 200]`
The max encountered going right-to-left is 200, then 200, then 200, then 400. The 400 at position 0 becomes 200 (the max to its right), while 200 becomes -1. Result: `[200, 200, 200, 200, -1]`.

### 3. Edge Case: Time Factor
**Input:** `arr = [1, 2, 3, ..., 10000]` (strictly increasing, $n = 10^4$)
The maximum to the right of each element is always the last element. Every position must be visited exactly once — full $O(n)$ scan with no early exits.

### 4. Edge Case: Space Factor
**Input:** `arr = [5]`
Single-element array; no right neighbors exist. The only element becomes -1. The in-place update uses $O(1)$ extra space regardless of input size.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [100000, 100000, ..., 100000]` (all same value, $n = 10^4$)
All elements are equal to the max. Each position gets 100000 (the running max from the right), except the last which becomes -1. Tests that the algorithm doesn't skip equal-max updates.
