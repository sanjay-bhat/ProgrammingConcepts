# LeetCode #1303: Find the Team Size

https://leetcode.com/problems/find-the-team-size/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Group Count with Hash Map ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each employee, count every other employee sharing the same `team_id` with a nested loop. $O(n^2)$ comparisons, $O(1)$ extra space.

### Optimal: Group Count with Hash Map ★
One pass builds a frequency map from `team_id` to count; a second pass writes each employee's answer using a direct lookup. Two linear scans, $O(n)$ time, $O(n)$ space for the map.

**Constraints:**
* $1 \leq n \leq 1000$ (number of employees)
* Each employee belongs to exactly one team
* `employee_id` values are unique


## Solutions

### C#

```
public class Solution {
    public int[] FindTeamSize(int[] teamId) {
        // Count how many employees belong to each team
        var freq = new Dictionary<int, int>();
        foreach (int t in teamId)
            freq[t] = freq.GetValueOrDefault(t, 0) + 1;
        // Map each employee to their team's total head-count
        int[] result = new int[teamId.Length];
        for (int i = 0; i < teamId.Length; i++)
            result[i] = freq[teamId[i]];
        return result;
    }
}
```

### Python

```
from collections import Counter

class Solution:
    def find_team_size(self, team_id: list[int]) -> list[int]:
        # Count how many employees belong to each team
        freq = Counter(team_id)
        # Map each employee to their team's total head-count
        return [freq[t] for t in team_id]
```

### Go

```
func findTeamSize(teamId []int) []int {
    // Count how many employees belong to each team
    freq := make(map[int]int)
    for _, t := range teamId {
        freq[t]++
    }
    // Map each employee to their team's total head-count
    result := make([]int, len(teamId))
    for i, t := range teamId {
        result[i] = freq[t]
    }
    return result
}
```

### Rust

```
use std::collections::HashMap;

impl Solution {
    pub fn find_team_size(team_id: Vec<i32>) -> Vec<i32> {
        // Count how many employees belong to each team
        let mut freq: HashMap<i32, i32> = HashMap::new();
        for &t in &team_id {
            *freq.entry(t).or_insert(0) += 1;
        }
        // Map each employee to their team's total head-count
        team_id.iter().map(|t| freq[t]).collect()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `teamId = [8,8,3,3,8]`
Frequency map: `{8:3, 3:2}`. Result: `[3,3,2,2,3]`. Each employee correctly sees their full team count via $O(1)$ map lookup.

### 2. Slightly Complex
**Input:** `teamId = [1,2,3,4,5]` (every employee on a different team)
Frequency map: all counts are 1. Result: `[1,1,1,1,1]`. Confirms singleton teams are handled without edge-case branching.

### 3. Edge Case: Time Factor
**Input:** $n = 1000$, all employees on the same team
First pass counts to 1000; second pass writes 1000 to all slots — maximum iterations, still $O(n)$. The hash map has exactly one entry, so lookup is $O(1)$ per step.

### 4. Edge Case: Space Factor
**Input:** $n = 1000$, all employees on distinct teams
The frequency map holds 1000 entries — maximum memory usage of $O(n)$. Result array is also $n$ entries of value 1.

### 5. Almost-Impossible but Plausible
**Input:** `teamId = [1000,1000,1,1000]` (team IDs at integer boundary)
Frequency: `{1000:3, 1:1}`. Result: `[3,3,1,3]`. Large team ID values cause no collision or overflow in hash map implementations — a subtle gotcha if using array-indexed buckets instead.
