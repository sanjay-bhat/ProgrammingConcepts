# LeetCode #1304: Find N Unique Integers Sum up to Zero

https://leetcode.com/problems/find-n-unique-integers-sum-up-to-zero/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Direct Construction ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Generate numbers greedily, checking each candidate against all previously chosen values for uniqueness; adjust the last element to hit the target sum. Uniqueness checking costs $O(n)$ per candidate, making it $O(n^2)$ overall.

### Optimal: Direct Construction ★
Fill the first $n - 1$ slots with $[1, 2, \ldots, n-1]$ — guaranteed distinct — then set the last slot to $-(1 + 2 + \cdots + (n-1)) = -\frac{(n-1)n}{2}$. This negation is always distinct from every positive integer already placed, so uniqueness holds with one pass.

**Constraints:**
* $1 \leq n \leq 1000$
* Any valid answer is accepted.


## Solutions

### C#

```
public class Solution {
    public int[] SumZero(int n) {
        int[] result = new int[n];
        int sum = 0;
        // Fill first n-1 slots with consecutive positive integers
        for (int i = 0; i < n - 1; i++) {
            result[i] = i + 1;
            sum += i + 1;
        }
        // Last element is the negation of all others, forcing the total to zero
        result[n - 1] = -sum;
        return result;
    }
}
```

### Python

```
class Solution:
    def sum_zero(self, n: int) -> list[int]:
        # Fill first n-1 slots with consecutive positive integers
        result = list(range(1, n))
        # Last element is the negation of all others, forcing the total to zero
        result.append(-sum(result))
        return result
```

### Go

```
func sumZero(n int) []int {
    result := make([]int, n)
    sum := 0
    // Fill first n-1 slots with consecutive positive integers
    for i := 0; i < n-1; i++ {
        result[i] = i + 1
        sum += i + 1
    }
    // Last element is the negation of all others, forcing the total to zero
    result[n-1] = -sum
    return result
}
```

### Rust

```
impl Solution {
    pub fn sum_zero(n: i32) -> Vec<i32> {
        // Collect [1, 2, ..., n-1] into the result
        let mut result: Vec<i32> = (1..n).collect();
        // Last element is the negation of all others, forcing the total to zero
        let total: i32 = result.iter().sum();
        result.push(-total);
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n = 5`
Result: `[1, 2, 3, 4, -10]`. The sum of positives is $1+2+3+4=10$, so the last element is $-10$. All five integers are distinct and their total is exactly zero.

### 2. Slightly Complex
**Input:** `n = 3`
Result: `[1, 2, -3]`. Minimal non-trivial case: $1 + 2 - 3 = 0$. Only three slots, two positive and one negative anchor.

### 3. Edge Case: Time Factor
**Input:** `n = 1000`
One pass builds 999 positive integers, then computes their sum $= \frac{999 \times 1000}{2} = 499500$ and places $-499500$ at the end. All 1000 operations are $O(1)$; total is $O(n)$.

### 4. Edge Case: Space Factor
**Input:** `n = 1`
The loop does not execute; `sum = 0`; the single slot gets $-0 = 0$. Result: `[0]`. The only integer whose solo sum is zero — also the minimum possible output.

### 5. Almost-Impossible but Plausible
**Input:** `n = 2`
Result: `[1, -1]`. Last element $= -(1) = -1$. The smallest non-trivial pair that sums to zero; both values are within $[-n, n]$ so the output is both valid and minimal. No overflow possible since $n \leq 1000 \Rightarrow |\text{last}| \leq 499500$.
