# LeetCode #1309: Decompress Run-Length Encoded List

https://leetcode.com/problems/decompress-run-length-encoded-list/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(k)$ | $O(k)$ |
| **Optimal: Single Pass Expansion ★** | $O(k)$ | $O(k)$ |

---

## Understanding the Methods

### Brute Force
Same as the optimal — the problem requires producing every element in the decompressed sequence, so $O(k)$ time and space (where $k$ is the total output length) is unavoidable. Any extra work beyond one pass is overhead.

### Optimal: Single Pass Expansion ★
Iterate through input pairs `[freq, val]` stepping by 2, and for each pair append `val` exactly `freq` times. One loop, no extra bookkeeping, output built directly. The output itself is the only significant allocation.

**Constraints:**
* $2 \leq \text{nums.length} \leq 100$ (always even)
* $1 \leq \text{nums}[i] \leq 100$


## Solutions

### C#

```
public class Solution {
    public int[] DecompressRLElist(int[] nums) {
        var result = new List<int>();
        // Step through consecutive [freq, val] pairs
        for (int i = 0; i < nums.Length; i += 2) {
            int freq = nums[i], val = nums[i + 1];
            // Repeat the value exactly freq times to reconstruct its run
            for (int j = 0; j < freq; j++)
                result.Add(val);
        }
        return result.ToArray();
    }
}
```

### Python

```
class Solution:
    def decompress_rl_elist(self, nums: list[int]) -> list[int]:
        result = []
        # Step through consecutive [freq, val] pairs
        for i in range(0, len(nums), 2):
            # Repeat the value exactly freq times to reconstruct its run
            result.extend([nums[i + 1]] * nums[i])
        return result
```

### Go

```
func decompressRLElist(nums []int) []int {
    var result []int
    // Step through consecutive [freq, val] pairs
    for i := 0; i < len(nums); i += 2 {
        freq, val := nums[i], nums[i+1]
        // Repeat the value exactly freq times to reconstruct its run
        for j := 0; j < freq; j++ {
            result = append(result, val)
        }
    }
    return result
}
```

### Rust

```
impl Solution {
    pub fn decompress_rl_elist(nums: Vec<i32>) -> Vec<i32> {
        let mut result = Vec::new();
        // Step through consecutive [freq, val] pairs using chunks
        for chunk in nums.chunks(2) {
            let (freq, val) = (chunk[0] as usize, chunk[1]);
            // Repeat the value exactly freq times to reconstruct its run
            result.extend(std::iter::repeat(val).take(freq));
        }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [1, 2, 3, 4]`
Pair (1, 2) → append 2 once; pair (3, 4) → append 4 three times. Output: `[2, 4, 4, 4]`.

### 2. Slightly Complex
**Input:** `nums = [1, 1, 2, 3]`
Pair (1, 1) → `[1]`; pair (2, 3) → `[3, 3]`. Output: `[1, 3, 3]`. Mixed frequencies with the same loop, no special handling.

### 3. Edge Case: Time Factor
**Input:** `nums = [100, 7]`
One pair, freq = 100; the inner loop runs 100 times, appending 7 each iteration. Worst-case expansion for a single pair — the entire output time is spent in one frequency run.

### 4. Edge Case: Space Factor
**Input:** `nums = [1, 1, 1, 2, 1, 3, 1, 4]`
Four pairs each with freq = 1; output has exactly 4 elements. Maximum pair count (50 pairs for length-100 input) with minimum frequency — the output is as small as the input.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [10, 10, 10, 10]`
Two pairs each with freq = 10 and val = 10; output is `[10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10, 10]` (twenty 10s). Verifies both high-frequency and high-value pairs are handled identically.
