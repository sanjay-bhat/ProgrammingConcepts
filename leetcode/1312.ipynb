# LeetCode #1312: Minimum Insertion Steps to Make a String Palindrome

https://leetcode.com/problems/minimum-insertion-steps-to-make-a-string-palindrome/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(2^n)$ | $O(n)$ |
| **Optimal: LCS-Based DP ★** | $O(n^2)$ | $O(n^2)$ |

---

## Understanding the Methods

### Brute Force
Recursively try inserting every character at every position, exploring all paths to a palindrome. Exponential branching is impractical beyond small inputs.

### Optimal: LCS-Based DP ★
The minimum insertions needed equals `n - LCS(s, reverse(s))`, where LCS is the longest common subsequence. Characters already in the LCS form the palindrome's "backbone" and need no insertion; the remaining `n - LCS` characters each need one mirror insertion. Standard LCS DP runs in $O(n^2)$ time and space.

**Constraints:**
* `1 <= s.length <= 500`
* `s` consists of lowercase English letters.

## Solutions

### C#

```
public class Solution {
    public int MinInsertions(string s) {
        int n = s.Length;
        string rev = new string(s.Reverse().ToArray());

        // lcs[i][j] = LCS length of s[0..i-1] and rev[0..j-1]
        int[,] lcs = new int[n + 1, n + 1];
        for (int i = 1; i <= n; i++) {
            for (int j = 1; j <= n; j++) {
                if (s[i - 1] == rev[j - 1])
                    // Matching characters extend the common backbone
                    lcs[i, j] = lcs[i - 1, j - 1] + 1;
                else
                    // Take the best of skipping one character from either string
                    lcs[i, j] = Math.Max(lcs[i - 1, j], lcs[i, j - 1]);
            }
        }
        // Characters outside the LCS each need one mirror insertion
        return n - lcs[n, n];
    }
}
```

### Python

```
class Solution:
    def minInsertions(self, s: str) -> int:
        n = len(s)
        rev = s[::-1]
        # lcs[i][j] = LCS of s[:i] and rev[:j]
        lcs = [[0] * (n + 1) for _ in range(n + 1)]

        for i in range(1, n + 1):
            for j in range(1, n + 1):
                if s[i - 1] == rev[j - 1]:
                    # Shared character forms part of the palindrome backbone
                    lcs[i][j] = lcs[i - 1][j - 1] + 1
                else:
                    lcs[i][j] = max(lcs[i - 1][j], lcs[i][j - 1])

        # Every character not in the LCS needs exactly one insertion
        return n - lcs[n][n]
```

### Go

```
func minInsertions(s string) int {
    n := len(s)
    rev := make([]byte, n)
    for i, c := range s { rev[n-1-i] = c }

    // lcs[i][j] = LCS length of s[:i] and rev[:j]
    lcs := make([][]int, n+1)
    for i := range lcs { lcs[i] = make([]int, n+1) }

    for i := 1; i <= n; i++ {
        for j := 1; j <= n; j++ {
            if s[i-1] == rev[j-1] {
                // Characters match — extend the shared backbone
                lcs[i][j] = lcs[i-1][j-1] + 1
            } else if lcs[i-1][j] > lcs[i][j-1] {
                lcs[i][j] = lcs[i-1][j]
            } else {
                lcs[i][j] = lcs[i][j-1]
            }
        }
    }
    // Each non-backbone character costs one insertion
    return n - lcs[n][n]
}
```

### Rust

```
impl Solution {
    pub fn min_insertions(s: String) -> i32 {
        let s: Vec<u8> = s.bytes().collect();
        let n = s.len();
        let rev: Vec<u8> = s.iter().rev().cloned().collect();

        // lcs[i][j] = LCS of s[..i] and rev[..j]
        let mut lcs = vec![vec![0usize; n + 1]; n + 1];
        for i in 1..=n {
            for j in 1..=n {
                if s[i-1] == rev[j-1] {
                    // Matching character contributes to the palindrome backbone
                    lcs[i][j] = lcs[i-1][j-1] + 1;
                } else {
                    lcs[i][j] = lcs[i-1][j].max(lcs[i][j-1]);
                }
            }
        }
        // Non-backbone characters each require one mirror insertion
        (n - lcs[n][n]) as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `s = "zzazz"`
Already a palindrome; `LCS("zzazz", "zzazz") = 5`. Insertions = $5 - 5 = $ **0**.

### 2. Slightly Complex
**Input:** `s = "mbadm"`
`rev = "mdabm"`. LCS = `"mam"` or `"mbm"` = 3. Insertions = $5 - 3 = $ **2** (insert to get "mbdabdm" or similar).

### 3. Edge Case: Time Factor
**Input:** `s` of length 500.
LCS table is $500 \times 500 = 250000$ cells — each filled in $O(1)$. Total $O(n^2)$.

### 4. Edge Case: Space Factor
**Input:** `s` of length 500.
LCS table uses $501 \times 501 \approx 250000$ integers — $O(n^2)$ space; a space-optimised two-row version is possible but the $O(n^2)$ table is clear and fast enough.

### 5. Almost-Impossible but Plausible
**Input:** `s = "abcdefghijklmnopqrstuvwxyz"` (26 distinct chars).
Every character appears once; `LCS(s, rev(s)) = 1` (single shared character). Insertions = $26 - 1 = $ **25**.
