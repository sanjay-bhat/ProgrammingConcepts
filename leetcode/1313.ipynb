# LeetCode #1313: Decompress Array

https://leetcode.com/problems/decompress-array/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(k)$ | $O(k)$ |
| **Optimal: Single Pass Expansion ★** | $O(k)$ | $O(k)$ |

---

## Understanding the Methods

### Brute Force
Materialize the output na\xefvely by iterating through each encoded pair and growing the result array element by element. Since the full output must be produced, $O(k)$ is the lower bound — brute force and optimal share the same Big-O.

### Optimal: Single Pass Expansion ★
Process encoded pairs in one forward pass: for each `[val, freq]` pair, extend the output with `val` repeated `freq` times. Pre-computing the total output size allows a single pre-allocated buffer rather than repeated resizing, keeping constant-factor overhead minimal.

**Constraints:**
* $2 \leq \text{nums.length} \leq 100$ (always even)
* $1 \leq \text{nums}[i] \leq 100$


## Solutions

### C#

```
public class Solution {
    public int[] DecompressArray(int[] nums) {
        // Pre-compute total output size to allocate once instead of resizing
        int total = 0;
        for (int i = 1; i < nums.Length; i += 2) total += nums[i];
        int[] result = new int[total];
        int idx = 0;
        // Step through [val, freq] pairs and write each run into the result
        for (int i = 0; i < nums.Length; i += 2) {
            int val = nums[i], freq = nums[i + 1];
            for (int j = 0; j < freq; j++)
                result[idx++] = val;
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def decompress_array(self, nums: list[int]) -> list[int]:
        result = []
        # Step through [val, freq] pairs and write each run into the result
        for i in range(0, len(nums), 2):
            val, freq = nums[i], nums[i + 1]
            # Extend with val repeated freq times to reconstruct the run
            result.extend([val] * freq)
        return result
```

### Go

```
func decompressArray(nums []int) []int {
    // Pre-compute total output size to allocate once instead of resizing
    total := 0
    for i := 1; i < len(nums); i += 2 {
        total += nums[i]
    }
    result := make([]int, 0, total)
    // Step through [val, freq] pairs and write each run into the result
    for i := 0; i < len(nums); i += 2 {
        val, freq := nums[i], nums[i+1]
        for j := 0; j < freq; j++ {
            result = append(result, val)
        }
    }
    return result
}
```

### Rust

```
impl Solution {
    pub fn decompress_array(nums: Vec<i32>) -> Vec<i32> {
        // Pre-compute total output size to allocate once
        let total: usize = nums.iter().skip(1).step_by(2).map(|&f| f as usize).sum();
        let mut result = Vec::with_capacity(total);
        // Step through [val, freq] pairs and write each run into the result
        for chunk in nums.chunks(2) {
            let (val, freq) = (chunk[0], chunk[1] as usize);
            result.extend(std::iter::repeat(val).take(freq));
        }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [2, 1, 4, 3]`
Pair (val=2, freq=1) → `[2]`; pair (val=4, freq=3) → `[4, 4, 4]`. Output: `[2, 4, 4, 4]`.

### 2. Slightly Complex
**Input:** `nums = [3, 2, 1, 1]`
Pair (val=3, freq=2) → `[3, 3]`; pair (val=1, freq=1) → `[1]`. Output: `[3, 3, 1]`. Demonstrates decreasing values with varying frequencies.

### 3. Edge Case: Time Factor
**Input:** `nums = [7, 100]`
One pair, val=7, freq=100; the inner loop runs 100 times. This is the worst-case single-pair expansion — pre-allocating `total = 100` avoids any reallocation during the 100 writes.

### 4. Edge Case: Space Factor
**Input:** `nums = [1, 1, 2, 1, 3, 1, 4, 1]`
Four pairs each with freq=1; total output size = 4. Maximum pair count with minimum frequency — proves that pre-allocation correctly handles $O(k)$ space even when $k = \frac{n}{2}$.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [100, 10, 100, 10]`
Two pairs, val=100, freq=10 each; output is 20 copies of 100. Both val and freq are at or near their upper bounds simultaneously — verifies the pre-computed `total = 20` buffer and loop correctness under combined stress.
