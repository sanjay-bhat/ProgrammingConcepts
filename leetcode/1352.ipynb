# LeetCode #1352: Product of Array Except Self

https://leetcode.com/problems/product-of-array-except-self/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Division Approach** | $O(n)$ | $O(1)$ extra |
| **Optimal: Prefix-Suffix Pass ★** | $O(n)$ | $O(1)$ extra |

---

## Understanding the Methods

### Brute Force
For each index $i$, multiply every other element — $O(n)$ per element, $O(n^2)$ total.

### Division Approach
Compute the total product, then divide by `nums[i]` for each index. Fails when zeros are present and is explicitly disallowed by the problem.

### Optimal: Prefix-Suffix Pass ★
First pass (left to right): fill the output array with running prefix products — `out[i]` = product of all elements before `i`. Second pass (right to left): multiply each `out[i]` by a running suffix accumulator that tracks the product of all elements after `i`. No extra array beyond the required output; $O(1)$ additional space.

**Why this is better than Division:** It handles zeros naturally and uses no division, satisfying the problem constraint entirely.

**Constraints:**
* $2 \leq \text{nums.length} \leq 10^5$
* $-30 \leq \text{nums}[i] \leq 30$
* The product of any prefix or suffix fits in a 32-bit integer
* No division allowed


## Solutions

### C#

```
public class Solution {
    public int[] ProductExceptSelf(int[] nums) {
        int n = nums.Length;
        int[] out = new int[n];
        // Forward pass: out[i] holds the product of all elements to the left of i
        out[0] = 1;
        for (int i = 1; i < n; i++)
            out[i] = out[i - 1] * nums[i - 1];
        // Backward pass: multiply in the running product of all elements to the right
        int suffix = 1;
        for (int i = n - 1; i >= 0; i--) {
            out[i] *= suffix;
            suffix *= nums[i];
        }
        return out;
    }
}
```

### Python

```
class Solution:
    def product_except_self(self, nums: list[int]) -> list[int]:
        n = len(nums)
        out = [1] * n
        # Forward pass: out[i] holds the product of all elements to the left of i
        for i in range(1, n):
            out[i] = out[i - 1] * nums[i - 1]
        # Backward pass: multiply in the running product of all elements to the right
        suffix = 1
        for i in range(n - 1, -1, -1):
            out[i] *= suffix
            suffix *= nums[i]
        return out
```

### Rust

```
impl Solution {
    pub fn product_except_self(nums: Vec<i32>) -> Vec<i32> {
        let n = nums.len();
        let mut out = vec![1i32; n];
        // Forward pass: out[i] holds the product of all elements to the left of i
        for i in 1..n {
            out[i] = out[i - 1] * nums[i - 1];
        }
        // Backward pass: multiply in the running product of all elements to the right
        let mut suffix = 1i32;
        for i in (0..n).rev() {
            out[i] *= suffix;
            suffix *= nums[i];
        }
        out
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [1,2,3,4]`
Forward pass → `[1,1,2,6]`. Backward with suffix `[24,12,4,1]` applied → `[24,12,8,6]`. Every element is the product of all others — classic result.

### 2. Slightly Complex
**Input:** `nums = [-1,1,0,-3,3]`
One zero at index 2. Forward: `[1,-1,-1,0,0]`. Backward suffix accumulates through zero, so only index 2's slot ends up non-zero after the right pass. Result: `[0,0,9,0,0]`.

### 3. Edge Case: Time Factor
**Input:** `nums` has $n = 10^5$ elements
Two full linear passes — $2n$ operations regardless of values. No inner loop, no recursion; the constant factor is minimal.

### 4. Edge Case: Space Factor
**Input:** `nums = [-30, 30, -30, 30]` (boundary values)
The only extra space is the single `suffix` scalar. No prefix array is stored separately — the output array doubles as the prefix store, satisfying the $O(1)$ extra constraint exactly.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [0, 0]`
Two zeros. Forward: `[1,0]`. Backward suffix through zero: `out[1]=0*1=0`, suffix becomes 0; `out[0]=1*0=0`. Result: `[0,0]`. A naive division approach would produce NaN or a division-by-zero error — prefix-suffix handles it silently.
