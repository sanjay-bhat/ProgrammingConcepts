# LeetCode #1369: Get Maximum in Generated Array

https://leetcode.com/problems/get-maximum-in-generated-array/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n)$ | $O(n)$ |
| **Optimal: Dynamic Programming (Bottom-Up) ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Same as the optimal — the generation rules mandate computing every element from scratch. A recursive top-down implementation without memoization would recompute sub-results and cost $O(n^2)$, but even basic caching brings it back to $O(n)$.

### Optimal: Dynamic Programming (Bottom-Up) ★
Build the array iteratively from index 0 upward, applying the two filling rules: `arr[2i] = arr[i]` and `arr[2i+1] = arr[i] + arr[i+1]`. Because smaller indices are always computed before larger ones, every lookup is already cached. A single scan for the maximum follows. No recursion stack overhead.

**Constraints:**
* $0 \leq n \leq 100$
* $\text{arr}[0] = 0$, $\text{arr}[1] = 1$ (base cases)


## Solutions

### C#

```
public class Solution {
    public int GetMaximumGenerated(int n) {
        if (n == 0) return 0;
        int[] arr = new int[n + 1];
        arr[1] = 1;
        // Build each slot using the two filling rules in index order
        for (int i = 1; 2 * i <= n; i++) {
            arr[2 * i] = arr[i];
            if (2 * i + 1 <= n)
                arr[2 * i + 1] = arr[i] + arr[i + 1];
        }
        // Scan the completed array to find the largest generated value
        int max = 0;
        foreach (int x in arr) if (x > max) max = x;
        return max;
    }
}
```

### Python

```
class Solution:
    def get_maximum_generated(self, n: int) -> int:
        if n == 0:
            return 0
        arr = [0] * (n + 1)
        arr[1] = 1
        # Build each slot using the two filling rules in index order
        for i in range(1, n // 2 + 1):
            arr[2 * i] = arr[i]
            if 2 * i + 1 <= n:
                arr[2 * i + 1] = arr[i] + arr[i + 1]
        return max(arr)
```

### Go

```
func getMaximumGenerated(n int) int {
    if n == 0 {
        return 0
    }
    arr := make([]int, n+1)
    arr[1] = 1
    // Build each slot using the two filling rules in index order
    for i := 1; 2*i <= n; i++ {
        arr[2*i] = arr[i]
        if 2*i+1 <= n {
            arr[2*i+1] = arr[i] + arr[i+1]
        }
    }
    // Scan the completed array to find the largest generated value
    max := 0
    for _, x := range arr {
        if x > max {
            max = x
        }
    }
    return max
}
```

### Rust

```
impl Solution {
    pub fn get_maximum_generated(n: i32) -> i32 {
        if n == 0 { return 0; }
        let n = n as usize;
        let mut arr = vec![0i32; n + 1];
        arr[1] = 1;
        // Build each slot using the two filling rules in index order
        for i in 1..=(n / 2) {
            arr[2 * i] = arr[i];
            if 2 * i + 1 <= n {
                arr[2 * i + 1] = arr[i] + arr[i + 1];
            }
        }
        // Scan the completed array to find the largest generated value
        *arr.iter().max().unwrap()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n = 7`
Array builds to `[0, 1, 1, 2, 1, 3, 2, 3]`. The rule `arr[5] = arr[2] + arr[3] = 1 + 2 = 3` creates the first maximum; `arr[7] = arr[3] + arr[4] = 2 + 1 = 3` ties it. Returns 3.

### 2. Slightly Complex
**Input:** `n = 2`
Array: `[0, 1, 1]`. Rule: `arr[2] = arr[1] = 1`. No odd-index slot within bounds. Maximum is 1 — confirming the base case propagates correctly.

### 3. Edge Case: Time Factor
**Input:** `n = 100`
The loop runs 50 iterations (i from 1 to 50), filling both even and odd indices where valid. All $n + 1 = 101$ slots are touched, then scanned once more for the max — two $O(n)$ passes total.

### 4. Edge Case: Space Factor
**Input:** `n = 0`
Early return 0; no array is allocated at all. The only allocated memory is the return value register — minimum possible space usage.

### 5. Almost-Impossible but Plausible
**Input:** `n = 1`
Array: `[0, 1]`. The loop condition `2*1 <= 1` is false, so the loop body never executes. Maximum over `[0, 1]` is 1. Catches implementations that skip the `n == 0` guard but fail on `n == 1` because `arr[1]` is the only non-zero seed.
