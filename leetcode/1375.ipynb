# LeetCode #1375: Bulb Switcher III

https://leetcode.com/problems/bulb-switcher-iii/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Max-Count Comparison ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Maintain a boolean array of ON/OFF states. After each bulb is turned on, scan all positions from 1 to the current maximum to check whether every bulb in that prefix is on. This scan is $O(n)$ per step, giving $O(n^2)$ total.

### Optimal: Max-Count Comparison ★
Track two values: the maximum position ever turned on (`maxPos`) and the count of bulbs turned on so far (`count`). A "blue" moment occurs exactly when `maxPos == count` — meaning every position from 1 to `maxPos` must be ON (no gaps). One pass, constant extra space.

**Why this is better than Brute Force:** The prefix-completeness check reduces to a single integer comparison instead of a full scan, dropping $O(n^2)$ to $O(n)$.

**Constraints:**
* $n == light.length$
* $1 \leq n \leq 5 \times 10^4$
* $light$ is a permutation of $[1, n]$


## Solutions

### C#

```
public class Solution {
    public int NumTimesAllBlue(int[] light) {
        int maxPos = 0, count = 0, result = 0;
        foreach (int pos in light) {
            count++;                              // One more bulb is now ON
            maxPos = Math.Max(maxPos, pos);       // Extend the rightmost ON position
            // Gap-free prefix: every slot 1..maxPos must be occupied
            if (maxPos == count) result++;
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def numTimesAllBlue(self, light: list[int]) -> int:
        max_pos = count = result = 0
        for pos in light:
            count += 1                              # One more bulb is now ON
            max_pos = max(max_pos, pos)             # Extend the rightmost ON position
            # Gap-free prefix: every slot 1..max_pos must be occupied
            if max_pos == count:
                result += 1
        return result
```

### Go

```
func numTimesAllBlue(light []int) int {
    maxPos, count, result := 0, 0, 0
    for _, pos := range light {
        count++                                // One more bulb is now ON
        if pos > maxPos {
            maxPos = pos                       // Extend the rightmost ON position
        }
        // Gap-free prefix: every slot 1..maxPos must be occupied
        if maxPos == count {
            result++
        }
    }
    return result
}
```

### Rust

```
impl Solution {
    pub fn num_times_all_blue(light: Vec<i32>) -> i32 {
        let mut max_pos = 0i32;
        let mut count = 0i32;
        let mut result = 0i32;
        for pos in light {
            count += 1;                          // One more bulb is now ON
            max_pos = max_pos.max(pos);          // Extend the rightmost ON position
            // Gap-free prefix: every slot 1..max_pos must be occupied
            if max_pos == count {
                result += 1;
            }
        }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `light = [2, 1, 3, 5, 4]`
Step 2: max=2, count=1 — gap at pos 1, no blue. Step 1: max=2, count=2 — prefix [1,2] full, blue! Step 3: max=3, count=3 — prefix [1,2,3] full, blue! Steps 4-5 are also blue. Total: **4** blue moments.

### 2. Slightly Complex
**Input:** `light = [3, 2, 4, 1, 5]`
Steps 1-3 all have `maxPos > count` (gaps exist), so no blue moments. At step 4 (turning on bulb 1), max=4, count=4 — finally a complete prefix. Step 5 also qualifies. Total: **2** blue moments.

### 3. Edge Case: Time Factor
**Input:** `light = [n, n-1, ..., 2, 1]` (reversed order, $n = 5 \times 10^4$)
No blue moment occurs until the very last step when bulb 1 is lit and all $n$ positions are filled. Every step is processed, and the comparison fires only once. Worst case for the algorithm's patience.

### 4. Edge Case: Space Factor
**Input:** `light = [1]`
Single bulb: after turning it on, max=1, count=1 — immediate blue moment. No additional storage needed; extra space is $O(1)$ regardless of $n$.

### 5. Almost-Impossible but Plausible
**Input:** `light = [1, 3, 2, 5, 4]` (alternating gaps)
Blue at step 1 (max=1, count=1). Step 2: max=3, count=2 — gap. Step 3: max=3, count=3 — blue again. Step 4: max=5, count=4 — gap. Step 5: max=5, count=5 — blue. Total **3** blue moments, testing that the counter resets correctly after a gap breaks the streak.
