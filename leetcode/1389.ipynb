# LeetCode #1389: Create Target Array in the Given Order

https://leetcode.com/problems/create-target-array-in-the-given-order/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: List Insert Simulation ★** | $O(n^2)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Manually shift elements in a plain array to make room for each insertion. $O(n)$ per insertion due to shifting, and $n$ insertions gives $O(n^2)$ total. No better asymptotic bound is achievable with a contiguous array.

### Optimal: List Insert Simulation ★
Use the language's built-in resizable list and its `Insert` method to handle shifting automatically. The complexity is the same $O(n^2)$, but the implementation is clean and leverages the optimized built-in rather than manual array manipulation.

**Why this is better than Brute Force:** The built-in insert uses an optimized memory-move under the hood, and the code is far simpler to read and verify.

**Constraints:**
* $1 \leq nums.length \leq 100$
* $1 \leq nums[i] \leq 100$
* $0 \leq index[i] \leq i$


## Solutions

### C#

```
public class Solution {
    public int[] CreateTargetArray(int[] nums, int[] index) {
        List<int> target = new List<int>();
        for (int i = 0; i < nums.Length; i++) {
            // Insert nums[i] at the specified index, shifting existing elements right
            target.Insert(index[i], nums[i]);
        }
        return target.ToArray();
    }
}
```

### Python

```
class Solution:
    def createTargetArray(self, nums: list[int], index: list[int]) -> list[int]:
        target: list[int] = []
        for i, (val, idx) in enumerate(zip(nums, index)):
            # Insert val at the specified index, shifting existing elements right
            target.insert(idx, val)
        return target
```

### Go

```
func createTargetArray(nums []int, index []int) []int {
    target := []int{}
    for i, val := range nums {
        idx := index[i]
        // Insert val at idx by slicing and re-appending to shift elements right
        target = append(target, 0)                         // Grow by one
        copy(target[idx+1:], target[idx:])                 // Shift right
        target[idx] = val                                  // Place the new element
    }
    return target
}
```

### Rust

```
impl Solution {
    pub fn create_target_array(nums: Vec<i32>, index: Vec<i32>) -> Vec<i32> {
        let mut target: Vec<i32> = Vec::new();
        for (val, idx) in nums.into_iter().zip(index.into_iter()) {
            // Insert val at idx, shifting existing elements right
            target.insert(idx as usize, val);
        }
        target
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [0, 1, 2, 3, 4]`, `index = [0, 1, 2, 2, 1]`
Insertions: [0] → [0,1] → [0,1,2] → [0,1,3,2] → [0,4,1,3,2]. Each `insert` at the given index shifts the tail right by one position.

### 2. Slightly Complex
**Input:** `nums = [1, 2, 3, 4, 0]`, `index = [0, 1, 2, 3, 0]`
First four insert at the tail; the last inserts 0 at front (index 0), pushing everything right. Result: `[0, 1, 2, 3, 4]`. The leading-index insertion exercises the shift logic on a full array.

### 3. Edge Case: Time Factor
**Input:** `nums = [1, ..., 100]`, `index = [0, 0, ..., 0]` (all zeros, $n = 100$)
Every element is inserted at position 0, forcing the entire target array to shift right each time. After $n$ insertions the array is reversed and every shift is of maximum length — worst case for $O(n^2)$ work.

### 4. Edge Case: Space Factor
**Input:** `nums = [42]`, `index = [0]`
Single element inserted at position 0. Target grows from empty to `[42]`. Minimum space: $O(1)$ output, no shifting needed.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [100, 1, 100, 1, ...]` (alternating), `index = [0, 0, 0, 0, ...]` (all zeros, $n = 100$)
Each insert at position 0 reverses the accumulated order. The constant index=0 combined with values that exceed typical boundary checks stresses the algorithm's ability to handle fully front-loaded insertions without off-by-one errors.
