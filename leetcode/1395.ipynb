# LeetCode #1395: Count Number of Teams

https://leetcode.com/problems/count-number-of-teams/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^3)$ | $O(1)$ |
| **Optimal: Middle Element Count ★** | $O(n^2)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Triple nested loop: try every triplet $(i, j, k)$ with $i < j < k$ and check if ratings form a strictly increasing or strictly decreasing sequence. $O(n^3)$ and too slow for $n = 1000$.

### Optimal: Middle Element Count ★
Fix each index $j$ as the middle soldier. Count how many elements to the **left** of $j$ are smaller (`leftSmall`) and larger (`leftLarge`). Elements to the right follow by complement: `rightLarge = n - j - 1 - leftSmall_after_j`, etc. Valid ascending teams through $j$: `leftSmall * rightLarge`; descending: `leftLarge * rightSmall`. Sum over all $j$ gives the answer in $O(n^2)$.

**Constraints:**
* $n == \text{rating.length}$
* $3 \leq n \leq 1000$
* $1 \leq \text{rating}[i] \leq 10^5$
* All integers in `rating` are unique


## Solutions

### C#

```
public class Solution {
    public int NumTeams(int[] rating) {
        int n = rating.Length, count = 0;
        for (int j = 1; j < n - 1; j++) {
            int leftSmall = 0, leftLarge = 0;
            // Count how many elements to the left are smaller/larger than rating[j]
            for (int i = 0; i < j; i++) {
                if (rating[i] < rating[j]) leftSmall++;
                else if (rating[i] > rating[j]) leftLarge++;
            }
            // Right counts derive from what's not on the left (all values are distinct)
            int rightLarge = (n - j - 1) - leftSmall; // elements right of j > rating[j]
            int rightSmall = (n - j - 1) - leftLarge;  // elements right of j < rating[j]
            // Ascending triplets through j + descending triplets through j
            count += leftSmall * rightLarge + leftLarge * rightSmall;
        }
        return count;
    }
}
```

### Python

```
class Solution:
    def num_teams(self, rating: list[int]) -> int:
        n = len(rating)
        count = 0
        for j in range(1, n - 1):
            left_small = left_large = 0
            # Count how many elements to the left are smaller/larger than rating[j]
            for i in range(j):
                if rating[i] < rating[j]:   left_small += 1
                elif rating[i] > rating[j]: left_large += 1
            # Right counts derive from what's not on the left (all values are distinct)
            right_large = (n - j - 1) - left_small
            right_small = (n - j - 1) - left_large
            # Ascending triplets through j + descending triplets through j
            count += left_small * right_large + left_large * right_small
        return count
```

### Rust

```
impl Solution {
    pub fn num_teams(rating: Vec<i32>) -> i32 {
        let n = rating.len();
        let mut count = 0i32;
        for j in 1..n - 1 {
            let (mut left_small, mut left_large) = (0i32, 0i32);
            // Count how many elements to the left are smaller/larger than rating[j]
            for i in 0..j {
                if rating[i] < rating[j]      { left_small += 1; }
                else if rating[i] > rating[j] { left_large += 1; }
            }
            // Right counts derive from what's not on the left (all values are distinct)
            let right_large = (n - j - 1) as i32 - left_small;
            let right_small = (n - j - 1) as i32 - left_large;
            // Ascending triplets through j + descending triplets through j
            count += left_small * right_large + left_large * right_small;
        }
        count
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `rating = [2,5,3,4,1]`
Fix `j=1` (rating 5): leftSmall=1, leftLarge=0; rightLarge=2, rightSmall=1. Ascending: $1 \times 2=2$. Fix `j=2` (rating 3): leftSmall=1, leftLarge=1; rightLarge=1, rightSmall=1. Adds $1+1=2$. Total across all `j`: **3**.

### 2. Slightly Complex
**Input:** `rating = [2,1,3]`
Only `j=1`: leftSmall=0, leftLarge=1; rightLarge=1, rightSmall=0. Ascending: 0; descending: $1 \times 0 = 0$. Total: **0** — no valid team.

### 3. Edge Case: Time Factor
**Input:** `rating` of length 1000, all in increasing order
Every middle `j` has leftSmall=j, rightLarge=n−j−1. Inner loop runs up to 1000 times per `j`; $O(n^2) = 10^6$ operations. Maximum work for this approach.

### 4. Edge Case: Space Factor
**Input:** Any `rating` of length 1000
Only six scalar variables (`leftSmall`, `leftLarge`, `rightLarge`, `rightSmall`, `count`, loop indices) — $O(1)$ space regardless of input size or answer magnitude.

### 5. Almost-Impossible but Plausible
**Input:** `rating = [1, 100000, 2]`
Fix `j=1` (100000): leftSmall=1 (only 1 is less), leftLarge=0; rightLarge=0, rightSmall=1. Ascending: $1 \times 0=0$; descending: $0 \times 1=0$. Answer: **0**. Extreme value gap does not trick the boundary arithmetic.
