# LeetCode #1398: Company Finder

https://leetcode.com/problems/company-finder/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n \cdot m)$ | $O(1)$ |
| **Optimal: HashSet Lookup ★** | $O(n + m)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each element in the second list, linearly scan the first list to check if it exists. With $n$ elements in the first list and $m$ in the second, this is $O(n \cdot m)$.

### Optimal: HashSet Lookup ★
Load all elements from the first array into a HashSet ($O(n)$ build, $O(1)$ lookup). Then scan the second array once, collecting elements that hit the set. Total $O(n + m)$ time with $O(n)$ extra space for the set.

**Why this is better than Brute Force:** The HashSet replaces the inner linear scan with a constant-time membership test, reducing $O(n \cdot m)$ to $O(n + m)$.

**Constraints:**
* $1 \leq companies1.length, companies2.length \leq 1000$
* $1 \leq companies1[i], companies2[j] \leq 10^6$


## Solutions

### C#

```
public class Solution {
    public IList<int> FindCompanies(int[] companies1, int[] companies2) {
        // Build a set of all companies in the first list for O(1) lookup
        HashSet<int> set1 = new HashSet<int>(companies1);
        List<int> result = new List<int>();
        // Collect companies from the second list that are also in the first
        foreach (int c in companies2) {
            if (set1.Contains(c)) result.Add(c);
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def findCompanies(self, companies1: list[int], companies2: list[int]) -> list[int]:
        # Build a set of all companies in the first list for O(1) lookup
        set1 = set(companies1)
        # Collect companies from the second list that are also in the first
        return [c for c in companies2 if c in set1]
```

### Go

```
func findCompanies(companies1 []int, companies2 []int) []int {
    // Build a set of all companies in the first list for O(1) lookup
    set1 := make(map[int]bool, len(companies1))
    for _, c := range companies1 {
        set1[c] = true
    }
    result := []int{}
    // Collect companies from the second list that are also in the first
    for _, c := range companies2 {
        if set1[c] {
            result = append(result, c)
        }
    }
    return result
}
```

### Rust

```
use std::collections::HashSet;

impl Solution {
    pub fn find_companies(companies1: Vec<i32>, companies2: Vec<i32>) -> Vec<i32> {
        // Build a set of all companies in the first list for O(1) lookup
        let set1: HashSet<i32> = companies1.into_iter().collect();
        // Collect companies from the second list that are also in the first
        companies2.into_iter().filter(|c| set1.contains(c)).collect()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `companies1 = [3, 1, 5]`, `companies2 = [5, 2, 3]`
Set from companies1: {1, 3, 5}. Scan companies2: 5 hits, 2 misses, 3 hits. Result: `[5, 3]` (order of companies2).

### 2. Slightly Complex
**Input:** `companies1 = [10, 20, 30, 40]`, `companies2 = [25, 30, 40, 15, 10]`
Set from companies1: {10, 20, 30, 40}. Companies2 scan yields 30, 40, 10 — three matches out of five. Result: `[30, 40, 10]`.

### 3. Edge Case: Time Factor
**Input:** `companies1 = [1..1000]`, `companies2 = [1..1000]` (both lists full, $n = m = 1000$)
All 1000 lookups hit the set. Both build and scan loops run to completion — maximum $O(n + m)$ work with maximum output size.

### 4. Edge Case: Space Factor
**Input:** `companies1 = [1..1000]`, `companies2 = [2000, 2001, ..., 3000]` (no overlap)
The HashSet is fully populated with 1000 entries (maximum space), but every lookup misses. Result is empty. Tests maximum set size with zero output.

### 5. Almost-Impossible but Plausible
**Input:** `companies1 = [1000000]`, `companies2 = [1000000, 1000000, 1000000]` (max ID, repeated)
The single-element set contains $10^6$. All three lookups hit. Result: `[1000000, 1000000, 1000000]`. Tests that duplicates in companies2 are preserved and that the maximum company ID value doesn't cause hash collision issues.
