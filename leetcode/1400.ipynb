# LeetCode #1400: Construct K Palindrome Strings

https://leetcode.com/problems/construct-k-palindrome-strings/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | Impractical (try all k-partitions) | $O(n)$ |
| **Optimal: Odd-Frequency Count ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Try every way to partition the string's characters into `k` groups and check whether each group can form a palindrome. The partition count is enormous — completely impractical.

### Optimal: Odd-Frequency Count ★
A string can form a palindrome iff at most one character has an odd frequency (the center). Count how many characters have odd frequency (`odd`). We need at least `odd` palindromes (one for each odd-frequency character) and at most `n` (one palindrome per character). Return `odd <= k <= n`.

**Why this is better than Brute Force:** We never construct any palindrome — a single frequency pass and two comparisons are sufficient.

**Constraints:**
* $1 \leq$ `s.length` $\leq 10^5$
* $1 \leq k \leq 10^5$
* `s` contains lowercase English letters only

## Solutions

### C#

```
public class Solution {
    public bool CanConstruct(string s, int k) {
        // Count characters with odd frequency — each needs its own palindrome center
        int odd = 0;
        var freq = new int[26];
        foreach (char c in s) freq[c - 'a']++;
        foreach (int f in freq) odd += f % 2;

        // Need at least 'odd' palindromes, and at most s.Length (one char each)
        return odd <= k && k <= s.Length;
    }
}
```

### Python

```
from collections import Counter

class Solution:
    def can_construct(self, s: str, k: int) -> bool:
        # Count characters with odd frequency — each needs its own palindrome center
        freq = Counter(s)
        odd = sum(1 for f in freq.values() if f % 2 == 1)

        # Need at least 'odd' palindromes, and at most len(s) (one char each)
        return odd <= k <= len(s)
```

### Go

```
func canConstruct(s string, k int) bool {
    // Count characters with odd frequency — each needs its own palindrome center
    var freq [26]int
    for _, c := range s {
        freq[c-'a']++
    }
    odd := 0
    for _, f := range freq {
        if f%2 == 1 {
            odd++
        }
    }
    // Need at least 'odd' palindromes, and at most len(s) (one char each)
    return odd <= k && k <= len(s)
}
```

### Rust

```
impl Solution {
    pub fn can_construct(s: String, k: i32) -> bool {
        // Count characters with odd frequency — each needs its own palindrome center
        let mut freq = [0u32; 26];
        for b in s.bytes() {
            freq[(b - b'a') as usize] += 1;
        }
        let odd = freq.iter().filter(|&&f| f % 2 == 1).count() as i32;

        // Need at least 'odd' palindromes, and at most s.len() (one char each)
        odd <= k && k <= s.len() as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `s = "annabelle", k = 2`
Frequencies: a=2, n=2, b=1, e=2, l=2 — odd count=1. We need 1 ≤ 2 ≤ 9. **true**.

### 2. Slightly Complex
**Input:** `s = "leetcode", k = 3`
Frequencies: l=1,e=2,t=1,c=1,o=1,d=1 — odd count=5. Need 5 ≤ 3? **false**.

### 3. Edge Case: Time Factor
**Input:** `s` of length $10^5$, `k = 1`
Frequency counting scans all $10^5$ characters — the full $O(n)$ scan. Then check `odd <= 1`.

### 4. Edge Case: Space Factor
**Input:** any `s`, any `k`
Only a 26-element frequency array and one integer are used — $O(1)$ extra space regardless of `s.length`.

### 5. Almost-Impossible but Plausible
**Input:** `s = "a", k = 2`
One character, k=2. We'd need 2 palindromes but only have 1 character — impossible since `k > len(s)`. **false**. Note that `odd=1 <= k=2` passes the first check, but `k=2 > len(s)=1` catches it.
