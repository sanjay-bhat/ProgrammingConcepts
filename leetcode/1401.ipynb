# LeetCode #1401: Circle and Rectangle Overlapping

https://leetcode.com/problems/circle-and-rectangle-overlapping/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(r^2)$ | $O(1)$ |
| **Optimal: Clamped Distance ★** | $O(1)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Rasterise the circle's bounding box and test each pixel for membership in both circle and rectangle. Not practical for large radii.

### Optimal: Clamped Distance ★
Find the point inside the rectangle that is closest to the circle's centre by clamping each coordinate: `nearestX = clamp(cx, x1, x2)`, `nearestY = clamp(cy, y1, y2)`. If the squared distance from the centre to that nearest point is $\leq r^2$, the shapes overlap. Pure $O(1)$ arithmetic, no square-root needed.

**Constraints:**
* `1 <= radius <= 2000`
* `-10^4 <= x1 < x2 <= 10^4`
* `-10^4 <= y1 < y2 <= 10^4`
* `-10^4 <= cx, cy <= 10^4`

## Solutions

### C#

```
public class Solution {
    public bool CheckOverlap(int radius, int xCenter, int yCenter,
                             int x1, int y1, int x2, int y2) {
        // Closest point in the rectangle to the circle's centre
        int nearX = Math.Clamp(xCenter, x1, x2);
        int nearY = Math.Clamp(yCenter, y1, y2);

        // Overlap exists when squared distance ≤ r² (avoids floating-point sqrt)
        long dx = xCenter - nearX;
        long dy = yCenter - nearY;
        return dx * dx + dy * dy <= (long)radius * radius;
    }
}
```

### Python

```
class Solution:
    def checkOverlap(self, radius: int, xCenter: int, yCenter: int,
                     x1: int, y1: int, x2: int, y2: int) -> bool:
        # Clamp centre coordinates to rectangle bounds
        near_x = max(x1, min(xCenter, x2))
        near_y = max(y1, min(yCenter, y2))

        # Compare squared distance to avoid floating-point square root
        dx = xCenter - near_x
        dy = yCenter - near_y
        return dx * dx + dy * dy <= radius * radius
```

### Go

```
func checkOverlap(radius int, xCenter int, yCenter int,
                  x1 int, y1 int, x2 int, y2 int) bool {
    // Clamp the circle's centre to the rectangle to find the nearest point
    nearX := clamp(xCenter, x1, x2)
    nearY := clamp(yCenter, y1, y2)

    // Squared distance comparison avoids square root
    dx := xCenter - nearX
    dy := yCenter - nearY
    return dx*dx+dy*dy <= radius*radius
}

func clamp(v, lo, hi int) int {
    if v < lo { return lo }
    if v > hi { return hi }
    return v
}
```

### Rust

```
impl Solution {
    pub fn check_overlap(radius: i32, x_center: i32, y_center: i32,
                         x1: i32, y1: i32, x2: i32, y2: i32) -> bool {
        // Nearest point on the rectangle to the circle's centre
        let near_x = x_center.clamp(x1, x2);
        let near_y = y_center.clamp(y1, y2);

        // Use squared distance to skip floating-point square root
        let dx = (x_center - near_x) as i64;
        let dy = (y_center - near_y) as i64;
        dx * dx + dy * dy <= (radius as i64) * (radius as i64)
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `radius=1, xCenter=0, yCenter=0, x1=1, y1=-1, x2=3, y2=1`
Nearest point: `(1, 0)`. Distance$^2 = 1 \leq 1$. Returns **true** (circle just touches the edge).

### 2. Slightly Complex
**Input:** `radius=1, xCenter=1, yCenter=1, x1=-3, y1=-3, x2=3, y2=3`
Centre is inside the rectangle; clamped point equals centre, distance = 0. Returns **true**.

### 3. Edge Case: Time Factor
**Input:** `radius=2000, xCenter=0, yCenter=0, x1=-10000, y1=-10000, x2=10000, y2=10000`
Always $O(1)$ — no loops regardless of radius magnitude.

### 4. Edge Case: Space Factor
**Input:** Any valid input.
Only five variables (`nearX, nearY, dx, dy, r²`) — $O(1)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `radius=2000, xCenter=-10000, yCenter=-10000, x1=9998, y1=9998, x2=10000, y2=10000`
Clamped nearest point: `(9998, 9998)`. $dx = -10000-9998 = -19998$, $dy$ likewise. $dx^2+dy^2 \approx 8 \times 10^8 \gg 4 \times 10^6 = r^2$. Returns **false**.
