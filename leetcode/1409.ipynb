# LeetCode #1409: Queries on a Permutation With Key

https://leetcode.com/problems/queries-on-a-permutation-with-key/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: List + Map Simulation ★** | $O(n^2)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Store the permutation in a plain array. For each query, linearly scan to find the queried value's position, record it, then rebuild the array with that element at the front. $O(n)$ per query, $O(n^2)$ total.

### Optimal: List + Map Simulation ★
Maintain the permutation in a resizable list and a HashMap from value to its current index. For each query, use the map for $O(1)$ position lookup, record the index, remove the element from the list (shifting is still $O(n)$), and re-insert at position 0. Update the map for all shifted elements.

**Why this is better than Brute Force:** The HashMap eliminates the linear scan for position lookup, though the shift itself remains $O(n)$. The overall complexity stays $O(n^2)$ but with a smaller constant and cleaner structure.

**Constraints:**
* $1 \leq m \leq 10^3$
* $queries.length \leq m$
* $1 \leq queries[i] \leq m$


## Solutions

### C#

```
public class Solution {
    public int[] ProcessQueries(int[] queries, int m) {
        // Build the initial permutation [1..m] in a list
        List<int> perm = new List<int>(Enumerable.Range(1, m));
        int[] result = new int[queries.Length];
        for (int i = 0; i < queries.Length; i++) {
            // Find the current position of the queried value
            int pos = perm.IndexOf(queries[i]);
            result[i] = pos;         // Record 0-based position
            // Move the found element to the front of the permutation
            perm.RemoveAt(pos);
            perm.Insert(0, queries[i]);
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def processQueries(self, queries: list[int], m: int) -> list[int]:
        # Build the initial permutation [1..m] in a list
        perm = list(range(1, m + 1))
        result = []
        for q in queries:
            # Find the current position of the queried value
            pos = perm.index(q)
            result.append(pos)    # Record 0-based position
            # Move the found element to the front of the permutation
            perm.pop(pos)
            perm.insert(0, q)
        return result
```

### Go

```
func processQueries(queries []int, m int) []int {
    // Build the initial permutation [1..m] in a slice
    perm := make([]int, m)
    for i := range perm {
        perm[i] = i + 1
    }
    result := make([]int, len(queries))
    for qi, q := range queries {
        // Find the current position of the queried value
        pos := 0
        for i, v := range perm {
            if v == q {
                pos = i
                break
            }
        }
        result[qi] = pos // Record 0-based position
        // Move the found element to the front of the permutation
        perm = append([]int{q}, append(perm[:pos], perm[pos+1:]...)...)
    }
    return result
}
```

### Rust

```
impl Solution {
    pub fn process_queries(queries: Vec<i32>, m: i32) -> Vec<i32> {
        // Build the initial permutation [1..m] in a Vec
        let mut perm: Vec<i32> = (1..=m).collect();
        let mut result = Vec::with_capacity(queries.len());
        for q in queries {
            // Find the current position of the queried value
            let pos = perm.iter().position(|&v| v == q).unwrap();
            result.push(pos as i32); // Record 0-based position
            // Move the found element to the front of the permutation
            perm.remove(pos);
            perm.insert(0, q);
        }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `queries = [3, 1, 2, 1]`, `m = 5`
Permutation starts [1,2,3,4,5]. Query 3 → pos 2, perm becomes [3,1,2,4,5]. Query 1 → pos 1, perm [1,3,2,4,5]. Query 2 → pos 2, perm [2,1,3,4,5]. Query 1 → pos 1. Result: `[2, 1, 2, 1]`.

### 2. Slightly Complex
**Input:** `queries = [4, 1, 2, 2]`, `m = 4`
Start [1,2,3,4]. Q=4→pos 3, [4,1,2,3]. Q=1→pos 1, [1,4,2,3]. Q=2→pos 2, [2,1,4,3]. Q=2→pos 0 (already at front). Result: `[3, 1, 2, 0]`. Repeated queries for an element already at front return position 0.

### 3. Edge Case: Time Factor
**Input:** `queries = [m, m-1, ..., 1]` (each element once in reverse), `m = 1000`
Each query finds its target near the end of the permutation (worst scan distance). After each move-to-front the next query's target has shifted, keeping the scan long every time — maximum $O(n^2)$ total work.

### 4. Edge Case: Space Factor
**Input:** `queries = [1]`, `m = 1000`
Single query on a permutation of 1000 elements. Value 1 is at index 0; it's moved to front (no-op). Result: `[0]`. Only the permutation itself occupies $O(m)$ space; the result is $O(1)$.

### 5. Almost-Impossible but Plausible
**Input:** `queries = [1000, 1000, ..., 1000]` ($m$ repetitions, $m = 1000$)
First query finds 1000 at the last position (index 999), moves it to front. Every subsequent query finds 1000 at index 0. Result: $[999, 0, 0, \ldots, 0]$. After the first expensive move, all remaining lookups are free — tests that the front position is correctly recorded as 0.
