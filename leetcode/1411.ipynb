# LeetCode #1411: Number of Ways to Paint N × 3 Grid

https://leetcode.com/problems/number-of-ways-to-paint-n-3-grid/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(3^{3n})$ | $O(n)$ |
| **Optimal: Row-Pattern DP ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Enumerate all valid 3-colorings of each cell row by row, checking adjacency constraints. The number of combinations explodes exponentially.

### Optimal: Row-Pattern DP ★
Observe that valid 3-column rows (no adjacent same color) fall into exactly two structural types: ABA (center differs from both ends) and ABC (all three different). Count how each type transitions to the next row: ABA→ABA gives 3 new patterns, ABA→ABC gives 5, ABC→ABA gives 2, ABC→ABC gives 2. Track only the counts of each type across rows — $O(n)$ time, $O(1)$ space.

**Constraints:**
* $1 \leq n \leq 5000$

## Solutions

### C#

```
public class Solution {
    public int NumOfWays(int n) {
        const int MOD = 1_000_000_007;
        // ABA-type: center color differs from both ends (e.g. [1,2,1])
        // ABC-type: all three colors distinct (e.g. [1,2,3])
        long aba = 6, abc = 6; // row 1: 6 ABA patterns, 6 ABC patterns

        for (int i = 1; i < n; i++) {
            // Each ABA spawns 3 new ABA and 5 new ABC patterns in the next row
            // Each ABC spawns 2 new ABA and 2 new ABC patterns
            long newAba = aba * 3 + abc * 2;
            long newAbc = aba * 5 + abc * 2;  // wait, let me correct
            // Actually: ABA->ABA=3, ABA->ABC=5, ABC->ABA=2, ABC->ABC=2 (corrected below)
            newAba = aba * 3 + abc * 2;
            newAbc = aba * 2 + abc * 2;
            aba = newAba % MOD;
            abc = newAbc % MOD;
        }
        return (int)((aba + abc) % MOD);
    }
}
```

### Python

```
class Solution:
    def num_of_ways(self, n: int) -> int:
        MOD = 10**9 + 7
        # Row 1: 6 ABA-type patterns (center differs from ends), 6 ABC-type (all distinct)
        aba, abc = 6, 6

        for _ in range(n - 1):
            # Transition counts derived by enumerating all valid adjacent-row combinations
            # ABA -> new_ABA: 3, ABA -> new_ABC: 2
            # ABC -> new_ABA: 2, ABC -> new_ABC: 2
            aba, abc = (aba * 3 + abc * 2) % MOD, (aba * 2 + abc * 2) % MOD

        return (aba + abc) % MOD
```

### Go

### Rust

```
impl Solution {
    pub fn num_of_ways(n: i32) -> i32 {
        const MOD: i64 = 1_000_000_007;
        // Row 1: 6 ABA patterns, 6 ABC patterns
        let (mut aba, mut abc) = (6i64, 6i64);

        for _ in 1..n {
            // Transition: ABA->ABA=3, ABA->ABC=2; ABC->ABA=2, ABC->ABC=2
            let (new_aba, new_abc) = (
                (aba * 3 + abc * 2) % MOD,
                (aba * 2 + abc * 2) % MOD,
            );
            aba = new_aba;
            abc = new_abc;
        }
        ((aba + abc) % MOD) as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `n = 1`
There is only one row of 3 cells. Valid 3-colorings: $6$ ABA patterns ($+ 6$ ABC patterns) $= 12$ total ways.

### 2. Slightly Complex
**Input:** `n = 2`
After one transition: ABA count becomes $6 \times 3 + 6 \times 2 = 30$, ABC becomes $6 \times 2 + 6 \times 2 = 24$. Total: $54$ ways.

### 3. Edge Case: Time Factor
**Input:** `n = 5000`
The loop runs exactly $4999$ times performing constant-time modular arithmetic. Total: $O(n)$ steps with no exponential blowup — the result stays bounded by $10^9 + 6$.

### 4. Edge Case: Space Factor
**Input:** `n = 5000`
Only two integers (`aba`, `abc`) are maintained throughout — $O(1)$ space regardless of $n$.

### 5. Almost-Impossible but Plausible
**Input:** `n = 5000`
The answer modulo $10^9 + 7$ is a specific non-trivial value. Without the modular arithmetic, the true count would have thousands of digits — the modular reduction keeps each value in the range $[0, 10^9 + 6]$.
