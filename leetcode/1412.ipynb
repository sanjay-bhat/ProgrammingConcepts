# LeetCode #1412: Shortest Distance to Target String in a Circular Array

https://leetcode.com/problems/shortest-distance-to-target-string-in-a-circular-array/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Single Linear Scan ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
For each occurrence of the target string, compute both the clockwise and counterclockwise distance from `startIndex` via a linear search. Re-scanning for every occurrence is $O(n^2)$ in the worst case.

### Optimal: Single Linear Scan ★
In one pass, check every index where `words[i] == target`. Clockwise distance is `(i - startIndex + n) % n`; counterclockwise is `n - clockwise`. Track the running minimum of `min(clockwise, counterclockwise)`. No occurrence means return $-1$.

**Why this is better than Brute Force:** Collapses two nested traversals into one linear scan with two arithmetic operations per match.

**Constraints:**
* $1 \leq n \leq 100$
* $1 \leq \text{words}[i]\text{.length} \leq 100$
* $0 \leq \text{startIndex} < n$
* Returns $-1$ if target does not appear in `words`


## Solutions

### C#

```
public class Solution {
    public int ShortestDistance(string[] words, string target, int startIndex) {
        int n = words.Length;
        int minDist = int.MaxValue;

        for (int i = 0; i < n; i++) {
            if (words[i] != target) continue;

            // Clockwise steps from startIndex to i
            int cw = (i - startIndex + n) % n;
            // Counterclockwise steps is the complement around the circle
            int ccw = n - cw;
            minDist = Math.Min(minDist, Math.Min(cw, ccw));
        }

        return minDist == int.MaxValue ? -1 : minDist;
    }
}
```

### Python

```
class Solution:
    def shortestDistance(self, words: list[str], target: str, startIndex: int) -> int:
        n = len(words)
        min_dist = float('inf')

        for i, word in enumerate(words):
            if word != target:
                continue

            # Clockwise hops from startIndex to i, wrapping around if needed
            cw  = (i - startIndex + n) % n
            # Counterclockwise hops is the circle's complement of cw
            ccw = n - cw
            min_dist = min(min_dist, cw, ccw)

        return -1 if min_dist == float('inf') else min_dist
```

### Go

```
func shortestDistance(words []string, target string, startIndex int) int {
    n := len(words)
    minDist := n + 1 // sentinel larger than any real distance

    for i, w := range words {
        if w != target {
            continue
        }
        // Clockwise distance from startIndex to i
        cw  := (i - startIndex + n) % n
        // Counterclockwise is the remainder of the full circle
        ccw := n - cw
        d   := cw
        if ccw < d {
            d = ccw
        }
        if d < minDist {
            minDist = d
        }
    }

    if minDist == n+1 {
        return -1
    }
    return minDist
}
```

### Rust

```
impl Solution {
    pub fn shortest_distance(words: Vec<String>, target: String, start_index: i32) -> i32 {
        let n = words.len() as i32;
        let mut min_dist = i32::MAX;

        for (i, word) in words.iter().enumerate() {
            if word != &target { continue; }

            // Clockwise steps around the circular array
            let cw  = (i as i32 - start_index + n) % n;
            // Counterclockwise steps complement the clockwise distance
            let ccw = n - cw;
            min_dist = min_dist.min(cw.min(ccw));
        }

        if min_dist == i32::MAX { -1 } else { min_dist }
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `words = ["hello","i","am","happy","best"]`, `target = "happy"`, `startIndex = 3`
Target "happy" is at index 3. Clockwise: $(3-3+5)\%5 = 0$. Distance = **0** (already there).

### 2. Slightly Complex
**Input:** `words = ["a","b","leetcode"]`, `target = "leetcode"`, `startIndex = 0`
Target at index 2. Clockwise: $(2-0+3)\%3 = 2$. Counterclockwise: $3-2=1$. Minimum = **1**.

### 3. Edge Case: Time Factor
**Input:** $n = 100$, target appears at every index
Linear scan visits all 100 indices, computing two arithmetic operations each. Returns 0 (startIndex matches). Maximum iteration count with trivial per-step work confirms $O(n)$.

### 4. Edge Case: Space Factor
**Input:** `words = ["x","y","z"]`, `target = "w"`, `startIndex = 0`
Target never found. Loop runs $n=3$ iterations with no match. `minDist` stays at sentinel. Return **-1**. Only `minDist` and loop counter allocated — $O(1)$ space.

### 5. Almost-Impossible but Plausible
**Input:** `words = ["a","a","a"]`, `target = "a"`, `startIndex = 1`
Matches at indices 0, 1, 2. Distances: index 0 → cw=2, ccw=1, min=1; index 1 → cw=0; index 2 → cw=1, ccw=2, min=1. Overall minimum = **0**. Multiple occurrences are handled correctly — the scan tracks the global minimum across all matches.
