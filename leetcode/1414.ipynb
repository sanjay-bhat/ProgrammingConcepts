# LeetCode #1414: Find the Minimum of an Integer Array

https://leetcode.com/problems/find-the-minimum-of-an-integer-array/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n \log n)$ | $O(1)$ |
| **Optimal: Single Pass ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Sort the array and return the first element. Sorting costs $O(n \log n)$ and rearranges the data unnecessarily when only the minimum is needed.

### Optimal: Single Pass ★
Initialize the minimum to the first element, then scan once from left to right, updating the minimum whenever a smaller value is found. One pass, one variable, $O(n)$ time.

**Constraints:**
* $1 \leq \text{arr.length} \leq 10^5$
* $-10^9 \leq \text{arr}[i] \leq 10^9$


## Solutions

### C#

```
public class Solution {
    public int FindMinimum(int[] arr) {
        // Initialize to the first element so all subsequent comparisons are valid
        int min = arr[0];
        // Update the running minimum whenever a strictly smaller value is found
        for (int i = 1; i < arr.Length; i++)
            if (arr[i] < min) min = arr[i];
        return min;
    }
}
```

### Python

```
class Solution:
    def find_minimum(self, arr: list[int]) -> int:
        # Python's built-in min performs a single linear scan internally
        return min(arr)
```

### Go

```
func findMinimum(arr []int) int {
    // Initialize to the first element so all subsequent comparisons are valid
    min := arr[0]
    // Update the running minimum whenever a strictly smaller value is found
    for _, x := range arr[1:] {
        if x < min {
            min = x
        }
    }
    return min
}
```

### Rust

```
impl Solution {
    pub fn find_minimum(arr: Vec<i32>) -> i32 {
        // Iterator min performs a single linear scan and returns the smallest value
        *arr.iter().min().unwrap()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [3, 1, 4, 1, 5, 9, 2, 6]`
Running min updates at 1 (index 1) and stays there through all remaining elements. Returns 1 after a full pass.

### 2. Slightly Complex
**Input:** `arr = [-5, -3, -8, -1, -4]`
All negatives; min initializes to $-5$, updates to $-8$ at index 2, then remains there. Returns $-8$ — demonstrates handling of negative values without special cases.

### 3. Edge Case: Time Factor
**Input:** `arr = [8, 7, 6, 5, 4, 3, 2, 1]`
Strictly descending; min updates at every single step from 8 down to 1. Forces all $n - 1$ comparisons — worst case for time.

### 4. Edge Case: Space Factor
**Input:** `arr = [42]`
Single element; loop body never executes. Returns 42 using only one extra variable, confirming $O(1)$ space regardless of array content.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [-1000000000, -999999999, -1000000000]`
Two elements tie at $-10^9$; the first occurrence sets the minimum and the second comparison ($= \text{min}$, not $< \text{min}$) is skipped. Returns $-10^9$ correctly — the boundary of the 32-bit signed range without overflow.
