# LeetCode #1415: The k-th Lexicographical String of All Happy Strings of Length n

https://leetcode.com/problems/the-k-th-lexicographical-string-of-all-happy-strings-of-length-n/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (generate all)** | $O(3 \cdot 2^{n-1} \cdot n)$ | $O(3 \cdot 2^{n-1} \cdot n)$ |
| **Optimal: DFS / Mathematical Skip ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force (generate all)
Generate every happy string of length $n$ via DFS, store them all, sort, and return the $k$-th. Uses $O(3 \cdot 2^{n-1})$ space — there are exactly $3 \cdot 2^{n-1}$ happy strings of length $n$ (3 choices for position 0, then 2 for each subsequent position).

### Optimal: DFS / Mathematical Skip ★
Use the group-size formula: at position $i$, each of the valid remaining characters leads a subtree of $2^{n-1-i}$ strings. Skip entire subtrees by comparing $k$ to the subtree size, advancing $k$ and moving to the next character without generating anything. This directly constructs the answer in $O(n)$ character decisions with $O(n)$ stack space.

**Constraints:**
* 1 <= n <= 10
* 1 <= k <= 100

## Solutions
### C#

```
// DFS with mathematical skip: jump over subtrees of size 2^(n-1-pos)
public class Solution {
    public string GetHappyString(int n, int k) {
        var result = new System.Text.StringBuilder();
        char prev = ' ';

        for (int pos = 0; pos < n; pos++) {
            // Each remaining character leads 2^(n-1-pos) strings in this subtree
            int subtreeSize = 1 << (n - 1 - pos);
            bool placed = false;

            foreach (char c in new[] { 'a', 'b', 'c' }) {
                // Skip character if it matches the previous (no consecutive duplicates)
                if (c == prev) continue;
                if (k <= subtreeSize) {
                    // k falls within this character's subtree — pick it
                    result.Append(c);
                    prev = c;
                    placed = true;
                    break;
                }
                // Skip this subtree entirely: decrement k by its size
                k -= subtreeSize;
            }

            if (!placed) return ""; // k exceeds total happy strings
        }

        return result.ToString();
    }
}
```

### Python

```
# DFS with mathematical skip: jump over subtrees of size 2^(n-1-pos)
class Solution:
    def getHappyString(self, n: int, k: int) -> str:
        result = []
        prev = ''

        for pos in range(n):
            # Each valid character at this position heads 2^(n-1-pos) strings
            subtree_size = 1 << (n - 1 - pos)
            placed = False

            for c in 'abc':
                # Skip if it would create consecutive duplicates
                if c == prev:
                    continue
                if k <= subtree_size:
                    # k falls in this subtree — commit to this character
                    result.append(c)
                    prev = c
                    placed = True
                    break
                # Skip the entire subtree and reduce k accordingly
                k -= subtree_size

            if not placed:
                return ''  # k is larger than the total count

        return ''.join(result)
```

### Go

```
// DFS with mathematical skip: jump over subtrees of size 2^(n-1-pos)
package main

func getHappyString(n int, k int) string {
    result := make([]byte, 0, n)
    prev := byte(0)

    for pos := 0; pos < n; pos++ {
        // Each valid character leads 2^(n-1-pos) strings in its subtree
        subtreeSize := 1 << (n - 1 - pos)
        placed := false

        for _, c := range []byte{'a', 'b', 'c'} {
            // Consecutive duplicates are forbidden in happy strings
            if c == prev {
                continue
            }
            if k <= subtreeSize {
                // k falls within this subtree — select this character
                result = append(result, c)
                prev = c
                placed = true
                break
            }
            // Skip the entire subtree
            k -= subtreeSize
        }

        if !placed {
            return "" // k exceeds the number of valid happy strings
        }
    }

    return string(result)
}
```

### Rust

```
// DFS with mathematical skip: jump over subtrees of size 2^(n-1-pos)
impl Solution {
    pub fn get_happy_string(n: i32, mut k: i32) -> String {
        let n = n as usize;
        let mut result = Vec::with_capacity(n);
        let mut prev = b' ';

        for pos in 0..n {
            // Each valid character heads 2^(n-1-pos) strings in its subtree
            let subtree_size = 1i32 << (n - 1 - pos);
            let mut placed = false;

            for &c in b"abc" {
                // Skip consecutive duplicates
                if c == prev {
                    continue;
                }
                if k <= subtree_size {
                    // k lands in this subtree — commit to this character
                    result.push(c);
                    prev = c;
                    placed = true;
                    break;
                }
                // Skip the whole subtree
                k -= subtree_size;
            }

            if !placed {
                return String::new(); // k exceeds total happy strings
            }
        }

        String::from_utf8(result).unwrap()
    }
}
```

## Example Scenarios

There are exactly $3 \cdot 2^{n-1}$ happy strings of length $n$ — three choices at position 0, then two at each later position since a character may not repeat its predecessor. At position $i$ every still-valid character heads a subtree of exactly $2^{\,n-1-i}$ strings, so comparing $k$ against that size lets whole subtrees be skipped without generating anything.

### 1. Common Case — Length 1, the Last String

**Input:** `n = 1, k = 3`

The three happy strings are `"a"`, `"b"`, `"c"`, each its own subtree of size $2^0 = 1$. $k = 3 > 1$ skips `"a"` leaving $k = 2$; $k = 2 > 1$ skips `"b"` leaving $k = 1$; $k = 1 \le 1$ commits to `"c"`. **Answer: `"c"`.**

### 2. Slightly Complex — Three Levels of Skipping

**Input:** `n = 3, k = 9`

$3 \cdot 2^2 = 12$ strings exist. At position 0 each character heads $2^2 = 4$ of them: $k=9 > 4$ skips `a` ($k \to 5$), $k=5 > 4$ skips `b` ($k \to 1$), and $k = 1 \le 4$ commits to `c`. At position 1 the subtrees are size 2 and `a` is the first legal character, so $k = 1 \le 2$ takes it immediately. At position 2 the subtrees are size 1 and `b` is first legal, so it is taken. **Answer: `"cab"`** — the 9th of `aba, abc, aca, acb, bab, bac, bca, bcb, cab, cac, cba, cbc`.

### 3. Edge Case: Time Factor — Longest String, First Position

**Input:** `n = 10, k = 1`

With $k = 1$ the first legal character always wins immediately, producing the lexicographic minimum `"ababababab"`. The point is the cost: $3 \cdot 2^9 = 1536$ happy strings exist, yet only **10** character decisions are made and nothing is enumerated, stored or sorted. The brute force would build and sort all 1536.

### 4. Edge Case: Space Factor — k Beyond the Total Count

**Input:** `n = 1, k = 4`

Only 3 happy strings of length 1 exist, so $k = 4$ is out of range. The inner loop skips all three characters without ever satisfying $k \le \text{subtree size}$, `placed` stays false, and the function returns the empty string `""`. No partial result is retained — the single `result` list is at most $n$ characters at any moment, so space is $O(n)$ regardless of how large $k$ is.

### 5. Almost-Impossible but Plausible — Both Parameters at Their Maximum

**Input:** `n = 10, k = 100`

The largest $n$ and the largest $k$ the constraints allow. Since $3 \cdot 2^9 = 1536 \ge 100$ an answer exists, and the subtree arithmetic narrows $k$ from 100 to a single leaf in exactly 10 steps: **`"abacbabacb"`**. Note how close the two extremes are — $k = 100$ is still only the 100th of 1536, so the answer stays deep in the `a` branch; reaching the `c` branch at $n = 10$ would need $k > 1024$, which the constraint $k \le 100$ never permits.
