# LeetCode #1426: Counting Elements

https://leetcode.com/problems/counting-elements/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: HashSet Membership ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each element `x`, scan the entire array to check whether `x + 1` exists anywhere. Two nested loops give $O(n^2)$.

### Optimal: HashSet Membership ★
Insert all elements into a HashSet in one pass. Then in a second pass, count every element `x` for which `x + 1` is found in the set. Two linear passes, $O(n)$ total time, $O(n)$ space for the set.

**Why this is better than Brute Force:** The HashSet converts the inner membership scan from $O(n)$ to $O(1)$, cutting total time from $O(n^2)$ to $O(n)$.

**Constraints:**
* $1 \leq arr.length \leq 1000$
* $0 \leq arr[i] \leq 1000$


## Solutions

### C#

```
public class Solution {
    public int CountElements(int[] arr) {
        // Build a set so we can check x+1 membership in O(1)
        HashSet<int> seen = new HashSet<int>(arr);
        int count = 0;
        // Count each x whose successor x+1 also appears in the array
        foreach (int x in arr) {
            if (seen.Contains(x + 1)) count++;
        }
        return count;
    }
}
```

### Python

```
class Solution:
    def countElements(self, arr: list[int]) -> int:
        # Build a set so we can check x+1 membership in O(1)
        seen = set(arr)
        # Count each x whose successor x+1 also appears in the array
        return sum(1 for x in arr if x + 1 in seen)
```

### Go

```
func countElements(arr []int) int {
    // Build a set so we can check x+1 membership in O(1)
    seen := make(map[int]bool, len(arr))
    for _, x := range arr {
        seen[x] = true
    }
    count := 0
    // Count each x whose successor x+1 also appears in the array
    for _, x := range arr {
        if seen[x+1] {
            count++
        }
    }
    return count
}
```

### Rust

```
use std::collections::HashSet;

impl Solution {
    pub fn count_elements(arr: Vec<i32>) -> i32 {
        // Build a set so we can check x+1 membership in O(1)
        let seen: HashSet<i32> = arr.iter().cloned().collect();
        // Count each x whose successor x+1 also appears in the array
        arr.iter().filter(|&&x| seen.contains(&(x + 1))).count() as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `arr = [1, 2, 3]`
Set: {1, 2, 3}. Check: 1 (2 in set ✓), 2 (3 in set ✓), 3 (4 not in set ✗). Count = **2**.

### 2. Slightly Complex
**Input:** `arr = [1, 1, 2, 2]`
Set: {1, 2}. Both 1s pass (2 in set), both 2s fail (3 not in set). Count = **2**. Duplicates each count separately but share the same set lookup.

### 3. Edge Case: Time Factor
**Input:** `arr = [0, 1, 2, ..., 1000]` ($n = 1001$)
Every element except 1000 has a successor in the set. Both the build pass and the count pass iterate the full 1001 elements — maximum $O(n)$ work.

### 4. Edge Case: Space Factor
**Input:** `arr = [0, 2, 4, 6, ..., 1000]` (all even, $n = 501$)
Set contains 501 distinct values; none has an odd successor in the set. Result = **0**. The HashSet reaches near-maximum size while contributing zero to the count.

### 5. Almost-Impossible but Plausible
**Input:** `arr = [999, 1000, 999, 1000, ...]` (alternating 999/1000, $n = 1000$)
Set: {999, 1000}. Each 999 counts (1000 in set); each 1000 does not (1001 not in set). Half of 1000 elements count, giving **500**. Tests that many duplicates all pass the successor check without double-spending.
