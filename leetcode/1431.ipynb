# LeetCode #1431: Kids With the Greatest Number of Candies

https://leetcode.com/problems/kids-with-the-greatest-number-of-candies/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Single Pass ★** | $O(n)$ | $O(1)$ extra |

---

## Understanding the Methods

### Brute Force
For each kid, re-scan all other kids to find the current maximum and compare. $O(n)$ per kid, $O(n^2)$ overall.

### Optimal: Single Pass ★
Find the global maximum in one linear scan. Then in a second pass, each kid's answer is simply `candies[i] + extraCandies >= max`. Two $O(n)$ passes, $O(1)$ extra space beyond the output boolean array.

**Constraints:**
* $2 \leq \text{candies.length} \leq 100$
* $1 \leq \text{candies}[i] \leq 100$
* $1 \leq \text{extraCandies} \leq 50$


## Solutions

### C#

```
public class Solution {
    public IList<bool> KidsWithCandies(int[] candies, int extraCandies) {
        // Find the current global maximum to serve as the greatness threshold
        int max = candies[0];
        foreach (int c in candies) if (c > max) max = c;

        // Each kid qualifies if their total meets or beats the current maximum
        var result = new bool[candies.Length];
        for (int i = 0; i < candies.Length; i++)
            result[i] = candies[i] + extraCandies >= max;
        return result;
    }
}
```

### Python

```
class Solution:
    def kids_with_candies(self, candies: list[int], extra_candies: int) -> list[bool]:
        # Find the current global maximum to serve as the greatness threshold
        max_candies = max(candies)
        # Each kid qualifies if their total meets or beats the current maximum
        return [c + extra_candies >= max_candies for c in candies]
```

### Go

```
func kidsWithCandies(candies []int, extraCandies int) []bool {
    // Find the current global maximum to serve as the greatness threshold
    maxC := candies[0]
    for _, c := range candies {
        if c > maxC {
            maxC = c
        }
    }
    // Each kid qualifies if their total meets or beats the current maximum
    result := make([]bool, len(candies))
    for i, c := range candies {
        result[i] = c+extraCandies >= maxC
    }
    return result
}
```

### Rust

```
impl Solution {
    pub fn kids_with_candies(candies: Vec<i32>, extra_candies: i32) -> Vec<bool> {
        // Find the current global maximum to serve as the greatness threshold
        let max_c = *candies.iter().max().unwrap();
        // Each kid qualifies if their total meets or beats the current maximum
        candies.iter().map(|&c| c + extra_candies >= max_c).collect()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `candies = [2, 3, 5, 1, 3]`, `extraCandies = 3`
Max = 5. Totals: 5, 6, 8, 4, 6. Kids 0, 1, 2, 4 reach $\geq 5$; kid 3 ($1 + 3 = 4 < 5$) does not. Result: `[true, true, true, false, true]`.

### 2. Slightly Complex
**Input:** `candies = [4, 2, 1, 1, 2]`, `extraCandies = 1`
Max = 4. Only kid 0 ($4 + 1 = 5 \geq 4$) qualifies. Result: `[true, false, false, false, false]`. Extra candy is too small to lift any other kid.

### 3. Edge Case: Time Factor
**Input:** `candies = [1, 1, 1, ..., 1]` (100 ones), `extraCandies = 1`
Max = 1; every kid has $1 + 1 = 2 \geq 1$. Both passes run the full $n = 100$ iterations. All kids qualify: `[true, true, ..., true]`.

### 4. Edge Case: Space Factor
**Input:** `candies = [100, 1]`, `extraCandies = 1`
Max = 100. The boolean output has exactly 2 elements. Kid 0: $100 + 1 = 101 \geq 100$ → true. Kid 1: $1 + 1 = 2 < 100$ → false. Minimum output allocation for a two-kid array.

### 5. Almost-Impossible but Plausible
**Input:** `candies = [50, 50]`, `extraCandies = 50`
Max = 50. Both kids reach $100 \geq 50$ → `[true, true]`. Tests the tie-at-max case: the threshold is met via equality, not strict excess. No kid is left out even when starting equal.
