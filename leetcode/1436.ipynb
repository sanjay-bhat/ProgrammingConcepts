# LeetCode #1436: Destination City

https://leetcode.com/problems/destination-city/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Source Hash Set ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each destination city, check whether it appears as a source in any other path. If not, it is the final destination. Requires $O(n^2)$ comparisons.

### Optimal: Source Hash Set ★
Collect all source cities in a set in one pass, then scan destinations in a second pass. The first destination not found in the source set is the answer.

**Constraints:**
* $1 \le paths.length \le 100$
* Each city name is a non-empty string of at most 10 characters

## Solutions

### C#

```
using System.Collections.Generic;

public class Solution {
    public string DestCity(IList<IList<string>> paths) {
        // A source city always has an outgoing edge — collect them all
        var sources = new HashSet<string>();
        foreach (var p in paths) sources.Add(p[0]);

        // The destination city with no outgoing edge is the answer
        foreach (var p in paths)
            if (!sources.Contains(p[1])) return p[1];

        return "";
    }
}
```

### Python

```
class Solution:
    def destCity(self, paths: list[list[str]]) -> str:
        # Build a set of all cities that have an outgoing route
        sources = {p[0] for p in paths}
        # The unique destination that never appears as a source is the endpoint
        for _, dest in paths:
            if dest not in sources:
                return dest
        return ""  # unreachable per problem guarantees
```

### Go

```
func destCity(paths [][]string) string {
	// Record every city that has at least one outgoing path
	sources := map[string]bool{}
	for _, p := range paths {
		sources[p[0]] = true
	}
	// Find the destination that has no outgoing path
	for _, p := range paths {
		if !sources[p[1]] {
			return p[1]
		}
	}
	return ""
}
```

### Rust

```
use std::collections::HashSet;

impl Solution {
    pub fn dest_city(paths: Vec<Vec<String>>) -> String {
        // All cities that appear as a source (have an outgoing edge)
        let sources: HashSet<&str> = paths.iter().map(|p| p[0].as_str()).collect();
        // The destination absent from sources has no outgoing flight
        for p in &paths {
            if !sources.contains(p[1].as_str()) {
                return p[1].clone();
            }
        }
        String::new()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `paths=[["London","New York"],["New York","Lima"],["Lima","Sao Paulo"]]`
Sources = {London, New York, Lima}. "Sao Paulo" is a destination not in sources. Answer: **"Sao Paulo"**.

### 2. Slightly Complex
**Input:** `paths=[["B","C"],["D","B"],["C","A"]]`
Sources = {B, D, C}. Destination "A" never appears as a source. Answer: **"A"**.

### 3. Edge Case: Time Factor
**Input:** 100 paths forming a long chain $A_0 \to A_1 \to \cdots \to A_{100}$.
Hash set construction is $O(n)$; the single scan of destinations is $O(n)$. Total: $O(n)$ even for the maximum input.

### 4. Edge Case: Space Factor
**Input:** 100 paths, each with a unique source city name of length 10.
The source set holds at most 100 strings — $O(n)$ space regardless of city name length.

### 5. Almost-Impossible but Plausible
**Input:** `paths=[["X","Y"]]`
Exactly one path. Source = {X}. Destination "Y" is not in sources. Answer: **"Y"** in $O(1)$.
