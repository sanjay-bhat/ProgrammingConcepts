# LeetCode #1437: Check If All 1's Are at Least Length K Places Away

https://leetcode.com/problems/check-if-all-1s-are-at-least-length-k-places-away/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Gap Tracking ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
For each pair of 1's, compute their distance and check it is $\geq k$. $O(n^2)$ pairs in the worst case.

### Optimal: Gap Tracking ★
Single left-to-right scan. Track the index of the most recently seen 1 (`prev`). When a new 1 is found, compute `currentIndex - prev`. If this gap is $< k$, return `false`. Otherwise update `prev`. One pass, constant space.

**Constraints:**
* $1 \leq \text{nums.length} \leq 10^5$
* $0 \leq \text{nums}[i] \leq 1$
* $0 \leq k \leq \text{nums.length}$


## Solutions

### C#

```
public class Solution {
    public bool KLengthApart(int[] nums, int k) {
        int prev = -1; // Sentinel: no 1 seen yet
        for (int i = 0; i < nums.Length; i++) {
            if (nums[i] != 1) continue;
            // Distance from the previous 1 must cover at least k zeros
            if (prev != -1 && i - prev - 1 < k) return false;
            prev = i;
        }
        return true;
    }
}
```

### Python

```
class Solution:
    def k_length_apart(self, nums: list[int], k: int) -> bool:
        prev = -1  # Sentinel: no 1 seen yet
        for i, v in enumerate(nums):
            if v != 1:
                continue
            # Distance from the previous 1 must cover at least k zeros
            if prev != -1 and i - prev - 1 < k:
                return False
            prev = i
        return True
```

### Go

```
func kLengthApart(nums []int, k int) bool {
    prev := -1 // Sentinel: no 1 seen yet
    for i, v := range nums {
        if v != 1 {
            continue
        }
        // Distance from the previous 1 must cover at least k zeros
        if prev != -1 && i-prev-1 < k {
            return false
        }
        prev = i
    }
    return true
}
```

### Rust

```
impl Solution {
    pub fn k_length_apart(nums: Vec<i32>, k: i32) -> bool {
        let mut prev: i32 = -1; // Sentinel: no 1 seen yet
        for (i, &v) in nums.iter().enumerate() {
            if v != 1 { continue; }
            // Distance from the previous 1 must cover at least k zeros
            if prev != -1 && i as i32 - prev - 1 < k {
                return false;
            }
            prev = i as i32;
        }
        true
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [1,0,0,0,1,1,0,1]`, `k = 2`
Gaps between consecutive 1's: $4-0-1=3 \geq 2$ ✓, $5-4-1=0 < 2$ ✗. Returns `false` immediately on the second gap — no need to scan further.

### 2. Slightly Complex
**Input:** `nums = [1,0,0,1,0,1]`, `k = 1`
Gaps: $3-0-1=2 \geq 1$ ✓, $5-3-1=1 \geq 1$ ✓. All pairs satisfy $k=1$, returns `true`.

### 3. Edge Case: Time Factor
**Input:** `nums = [1,0,0,...,0,1]` (1 at each end, $n = 10^5$)
`prev` is set at index 0. The scan visits all $n-1$ zeros before finding the second 1 at $n-1$. Gap $= n-2 \geq k$ — valid. Full array traversal is needed; $O(n)$ is the tightest possible bound.

### 4. Edge Case: Space Factor
**Input:** `nums = [0,0,0,...,0]` (no 1's at all, $n = 10^5$)
`prev` stays at $-1$ throughout. The condition is vacuously true; only two integer scalars (`prev`, loop index) are ever allocated.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [1,0,1]`, `k = 2`
Gap between 1's: $2-0-1 = 1 < 2$. Returns `false`. Off-by-one trap: the gap formula counts **zeros** between the two 1's (`i - prev - 1`), not the raw index difference (`i - prev`). Confusing the two gives a wrong answer on tight inputs like this one.
