# LeetCode #1450: Number of Students Doing Homework at a Given Time

https://leetcode.com/problems/number-of-students-doing-homework-at-a-given-time/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n)$ | $O(1)$ |
| **Optimal: Single Pass ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
Same as optimal — every student's interval must be checked against `queryTime` at least once. No sub-linear algorithm exists for this comparison.

### Optimal: Single Pass ★
Iterate through all students in one pass. For each student, check whether `startTime[i] <= queryTime <= endTime[i]`. Increment a counter on a hit. One loop, two comparisons, zero extra allocation.

**Constraints:**
* $n = \text{startTime.length} = \text{endTime.length}$
* $1 \leq n \leq 100$
* $1 \leq \text{startTime}[i] \leq \text{endTime}[i] \leq 1000$
* $1 \leq \text{queryTime} \leq 1000$


## Solutions

### C#

```
public class Solution {
    public int BusyStudent(int[] startTime, int[] endTime, int queryTime) {
        int count = 0;
        // Count each student whose homework window contains the query moment
        for (int i = 0; i < startTime.Length; i++)
            if (startTime[i] <= queryTime && queryTime <= endTime[i])
                count++;
        return count;
    }
}
```

### Python

```
class Solution:
    def busy_student(self, start_time: list[int], end_time: list[int], query_time: int) -> int:
        # Count each student whose homework window contains the query moment
        return sum(s <= query_time <= e for s, e in zip(start_time, end_time))
```

### Go

```
func busyStudent(startTime []int, endTime []int, queryTime int) int {
    count := 0
    // Count each student whose homework window contains the query moment
    for i := range startTime {
        if startTime[i] <= queryTime && queryTime <= endTime[i] {
            count++
        }
    }
    return count
}
```

### Rust

```
impl Solution {
    pub fn busy_student(start_time: Vec<i32>, end_time: Vec<i32>, query_time: i32) -> i32 {
        // Count each student whose homework window contains the query moment
        start_time.iter()
            .zip(end_time.iter())
            .filter(|(&s, &e)| s <= query_time && query_time <= e)
            .count() as i32
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `startTime = [1, 2, 3]`, `endTime = [3, 2, 7]`, `queryTime = 4`
Student 0: $1 \leq 4 \leq 3$ — no. Student 1: $2 \leq 4 \leq 2$ — no. Student 2: $3 \leq 4 \leq 7$ — yes. Returns 1.

### 2. Slightly Complex
**Input:** `startTime = [4]`, `endTime = [4]`, `queryTime = 4`
Single student whose window is exactly one time unit. $4 \leq 4 \leq 4$ — both boundary checks pass. Returns 1 — confirms inclusive boundary semantics.

### 3. Edge Case: Time Factor
**Input:** `startTime = [1, 1, 1, ..., 1]` (100 ones), `endTime = [1000, 1000, ..., 1000]`, `queryTime = 500`
All 100 students qualify. Every iteration increments the counter — all $n$ comparisons succeed, exercising the loop fully.

### 4. Edge Case: Space Factor
**Input:** `startTime = [100]`, `endTime = [200]`, `queryTime = 50`
Query is before the student's window; $50 < 100$. Returns 0. One student, zero count — minimum allocation, verifies the early-exit (query before start) branch.

### 5. Almost-Impossible but Plausible
**Input:** `startTime = [1, 500, 999]`, `endTime = [499, 998, 1000]`, `queryTime = 1000`
Only student 2 is still working at $t = 1000$: $999 \leq 1000 \leq 1000$. Students 0 and 1 have finished. Returns 1 — tests the maximum `queryTime` boundary alongside varied window widths.
