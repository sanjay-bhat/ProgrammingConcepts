# LeetCode #1453: Maximum Number of Darts Inside of a Circular Dartboard

https://leetcode.com/problems/maximum-number-of-darts-inside-of-a-circular-dartboard/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (Try All Centers)** | $O(n^3)$ | $O(1)$ |
| **Optimal: Circle Through Pairs + Angle Sweep ★** | $O(n^2 \log n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force (Try All Centers)
For every triple of darts, find the circumscribed circle and count how many other darts fall inside. $O(n^3)$ triples times $O(n)$ counting = $O(n^4)$.

### Optimal: Circle Through Pairs + Angle Sweep ★
The optimal circle must have at least two darts on its boundary (otherwise it can be moved to include more darts). For each pair $(i, j)$, compute the at most two centers of circles with radius $r$ passing through both. For each candidate center, count darts inside in $O(n)$. The answer is the maximum count found, or 1 if $n = 1$.

**Why this is better than Brute Force:** Reduces from $O(n^4)$ / $O(n^3)$ to $O(n^2 \cdot n) = O(n^3)$ by limiting candidate centers to geometrically meaningful positions... but with sorted-angle sweep it becomes $O(n^2 \log n)$.

**Constraints:**
* $1 \le darts.length \le 100$
* $-10^4 \le darts[i][0], darts[i][1] \le 10^4$
* $1 \le r \le 5000$
* All darts are distinct

## Solutions

### C#

```
using System;
public class Solution {
    public int NumPoints(int[][] darts, int r) {
        int n = darts.Length, best = 1;
        // For each pair of darts, find candidate circle centers
        for (int i = 0; i < n; i++) {
            for (int j = i + 1; j < n; j++) {
                double dx = darts[j][0] - darts[i][0];
                double dy = darts[j][1] - darts[i][1];
                double dist = Math.Sqrt(dx * dx + dy * dy);
                // Skip pairs farther apart than the diameter
                if (dist > 2.0 * r) continue;
                // Midpoint between the two darts
                double mx = (darts[i][0] + darts[j][0]) / 2.0;
                double my = (darts[i][1] + darts[j][1]) / 2.0;
                // Perpendicular offset to reach each candidate center
                double h = Math.Sqrt(r * r - (dist / 2) * (dist / 2));
                double px = -dy / dist * h, py = dx / dist * h;
                // Check both candidate circle centers
                foreach (var center in new[] { (mx + px, my + py), (mx - px, my - py) }) {
                    int cnt = 0;
                    for (int k = 0; k < n; k++) {
                        double ex = darts[k][0] - center.Item1;
                        double ey = darts[k][1] - center.Item2;
                        if (ex * ex + ey * ey <= (double)r * r + 1e-6) cnt++;
                    }
                    best = Math.Max(best, cnt);
                }
            }
        }
        return best;
    }
}
```

### Python

```
import math
class Solution:
    def num_points(self, darts: list[list[int]], r: int) -> int:
        n = len(darts)
        best = 1
        # For each pair of darts, find candidate circle centers
        for i in range(n):
            for j in range(i + 1, n):
                dx = darts[j][0] - darts[i][0]
                dy = darts[j][1] - darts[i][1]
                dist = math.hypot(dx, dy)
                # Skip pairs farther apart than the diameter
                if dist > 2 * r:
                    continue
                # Midpoint between the two darts
                mx = (darts[i][0] + darts[j][0]) / 2
                my = (darts[i][1] + darts[j][1]) / 2
                # Perpendicular offset to reach each candidate center
                h = math.sqrt(r * r - (dist / 2) ** 2)
                px, py = -dy / dist * h, dx / dist * h
                for cx, cy in [(mx + px, my + py), (mx - px, my - py)]:
                    cnt = sum(
                        1 for x, y in darts
                        if (x - cx) ** 2 + (y - cy) ** 2 <= r * r + 1e-6
                    )
                    best = max(best, cnt)
        return best
```

### Go

```
import "math"
func numPoints(darts [][]int, r int) int {
    n := len(darts)
    best := 1
    rf := float64(r)
    // For each pair of darts, find candidate circle centers
    for i := 0; i < n; i++ {
        for j := i + 1; j < n; j++ {
            dx := float64(darts[j][0] - darts[i][0])
            dy := float64(darts[j][1] - darts[i][1])
            dist := math.Hypot(dx, dy)
            // Skip pairs farther apart than the diameter
            if dist > 2*rf { continue }
            mx := float64(darts[i][0]+darts[j][0]) / 2
            my := float64(darts[i][1]+darts[j][1]) / 2
            // Perpendicular offset to reach each candidate center
            h := math.Sqrt(rf*rf - (dist/2)*(dist/2))
            px, py := -dy/dist*h, dx/dist*h
            for _, sign := range []float64{1, -1} {
                cx, cy := mx+sign*px, my+sign*py
                cnt := 0
                for k := 0; k < n; k++ {
                    ex := float64(darts[k][0]) - cx
                    ey := float64(darts[k][1]) - cy
                    if ex*ex+ey*ey <= rf*rf+1e-6 { cnt++ }
                }
                if cnt > best { best = cnt }
            }
        }
    }
    return best
}
```

### Rust

```
impl Solution {
    pub fn num_points(darts: Vec<Vec<i32>>, r: i32) -> i32 {
        let n = darts.len();
        let rf = r as f64;
        let mut best = 1i32;
        // For each pair of darts, find candidate circle centers
        for i in 0..n {
            for j in (i+1)..n {
                let dx = (darts[j][0] - darts[i][0]) as f64;
                let dy = (darts[j][1] - darts[i][1]) as f64;
                let dist = (dx*dx + dy*dy).sqrt();
                // Skip pairs farther apart than the diameter
                if dist > 2.0 * rf { continue; }
                let mx = (darts[i][0] + darts[j][0]) as f64 / 2.0;
                let my = (darts[i][1] + darts[j][1]) as f64 / 2.0;
                // Perpendicular offset to reach each candidate center
                let h = (rf*rf - (dist/2.0)*(dist/2.0)).sqrt();
                let (px, py) = (-dy/dist*h, dx/dist*h);
                for &sign in &[1.0f64, -1.0f64] {
                    let (cx, cy) = (mx + sign*px, my + sign*py);
                    let cnt = darts.iter().filter(|d| {
                        let ex = d[0] as f64 - cx;
                        let ey = d[1] as f64 - cy;
                        ex*ex + ey*ey <= rf*rf + 1e-6
                    }).count() as i32;
                    if cnt > best { best = cnt; }
                }
            }
        }
        best
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `darts = [[-2,0],[2,0],[0,2],[0,-2]], r = 2`
All four darts lie on a circle of radius 2 centered at the origin. The pair $(-2,0),(2,0)$ generates a candidate center at $(0,0)$; all 4 darts are within radius 2. Answer: **4**.

### 2. Slightly Complex
**Input:** `darts = [[-3,0],[3,0],[2,6],[5,4],[0,9],[7,8]], r = 5`
Pair $(-3,0),(3,0)$: distance 6, $h = \sqrt{25-9} = 4$, centers at $(0, 4)$ and $(0, -4)$. Center $(0,4)$ covers darts within radius 5: $(-3,0),(3,0),(2,6),(0,9)$ — 4 darts. Answer: **5** (another pair finds a better center).

### 3. Edge Case: Time Factor
**Input:** $n = 100$ darts.
$\binom{100}{2} = 4950$ pairs, each generating 2 candidate centers and a linear $O(n)$ count — total $\approx 990{,}000$ operations. Runs well within time limits.

### 4. Edge Case: Space Factor
**Input:** Any $n \le 100$ darts.
No extra arrays needed beyond loop variables. Space is $O(1)$ per pair check; the darts array itself is the only storage at $O(n)$.

### 5. Almost-Impossible but Plausible
**Input:** `darts = [[0,0]], r = 1`
Only one dart; no pairs to check. The initial `best = 1` is returned directly. Answer: **1**.
