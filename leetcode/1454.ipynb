# LeetCode #1454: Active Students

https://leetcode.com/problems/active-students/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Set Intersection ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each student in the first session, check whether they appear in every other session list with a nested scan. $O(n^2)$ comparisons for $n$ students across $m$ sessions.

### Optimal: Set Intersection ★
Load the first session's student IDs into a set. For each subsequent session, intersect the current active set with the new session's set. Students absent from any session are eliminated in one pass per session. Total cost is $O(n \cdot m)$ which is $O(n)$ for a fixed number of sessions, $O(n)$ space for the sets.

**Constraints:**
* Student IDs are positive integers
* At least one session is provided
* A student is active only if they appear in every session


## Solutions

### C#

```
public class Solution {
    public IList<int> ActiveStudents(IList<IList<int>> sessions) {
        if (sessions.Count == 0) return new List<int>();
        // Seed the active set from the first session
        var active = new HashSet<int>(sessions[0]);
        // Intersect with each subsequent session to eliminate absent students
        foreach (var session in sessions.Skip(1))
            active.IntersectWith(session);
        var result = active.OrderBy(x => x).ToList();
        return result;
    }
}
```

### Python

```
class Solution:
    def active_students(self, sessions: list[list[int]]) -> list[int]:
        if not sessions:
            return []
        # Seed the active set from the first session
        active = set(sessions[0])
        # Intersect with each subsequent session to eliminate absent students
        for session in sessions[1:]:
            active &= set(session)
        return sorted(active)
```

### Go

```
import "sort"

func activeStudents(sessions [][]int) []int {
    if len(sessions) == 0 {
        return nil
    }
    // Seed the active set from the first session
    active := make(map[int]bool)
    for _, id := range sessions[0] {
        active[id] = true
    }
    // Intersect with each subsequent session to eliminate absent students
    for _, session := range sessions[1:] {
        next := make(map[int]bool)
        for _, id := range session {
            if active[id] {
                next[id] = true
            }
        }
        active = next
    }
    result := make([]int, 0, len(active))
    for id := range active {
        result = append(result, id)
    }
    sort.Ints(result)
    return result
}
```

### Rust

```
use std::collections::HashSet;

impl Solution {
    pub fn active_students(sessions: Vec<Vec<i32>>) -> Vec<i32> {
        if sessions.is_empty() { return vec![]; }
        // Seed the active set from the first session
        let mut active: HashSet<i32> = sessions[0].iter().copied().collect();
        // Intersect with each subsequent session to eliminate absent students
        for session in sessions.iter().skip(1) {
            let next: HashSet<i32> = session.iter().copied().collect();
            active = active.intersection(&next).copied().collect();
        }
        let mut result: Vec<i32> = active.into_iter().collect();
        result.sort_unstable();
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `sessions = [[1,2,3],[1,3],[1,2,3,4]]`
Session 1 active: `{1,2,3}`. Intersect with `{1,3}` → `{1,3}`. Intersect with `{1,2,3,4}` → `{1,3}`. Sorted result: `[1,3]`.

### 2. Slightly Complex
**Input:** `sessions = [[5,10,15],[10,15,20],[5,15,25]]`
Start with `{5,10,15}`. After session 2: `{10,15}`. After session 3: `{15}`. Only student 15 attended all sessions.

### 3. Edge Case: Time Factor
**Input:** 1000 sessions each with 1000 student IDs, heavily overlapping
Each intersection scan is $O(n)$ per session. With $m=1000$ sessions and $n=1000$ students, total work is $O(n \cdot m) = 10^6$ — still linear in the input size.

### 4. Edge Case: Space Factor
**Input:** Single session with 1000 students
No intersection needed. The active set holds all 1000 IDs — maximum $O(n)$ space. Returns the full first session sorted.

### 5. Almost-Impossible but Plausible
**Input:** `sessions = [[1,2],[3,4]]` (completely disjoint sessions)
After intersecting `{1,2}` with `{3,4}`: empty set. Returns `[]`. A correct implementation must handle the empty-result case without crashing or returning stale data.
