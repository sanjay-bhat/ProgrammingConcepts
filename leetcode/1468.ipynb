# LeetCode #1468: Find All Numbers Disappeared in an Array

https://leetcode.com/problems/find-all-numbers-disappeared-in-an-array/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: HashSet Scan ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each number in $[1, n]$, linearly scan the array to check whether it appears. $O(n)$ per check, $O(n^2)$ total. No extra space needed beyond the output.

### Optimal: HashSet Scan ★
Insert all elements of the array into a HashSet in one pass. Then iterate over $[1, n]$ and collect every number not found in the set. Two linear passes, $O(n)$ time, $O(n)$ space for the set.

**Why this is better than Brute Force:** The HashSet converts each membership check from $O(n)$ to $O(1)$, reducing total time from $O(n^2)$ to $O(n)$.

**Constraints:**
* $n == nums.length$
* $1 \leq n \leq 10^5$
* $1 \leq nums[i] \leq n$


## Solutions

### C#

```
public class Solution {
    public IList<int> FindDisappearedNumbers(int[] nums) {
        // Build a set of all present numbers for O(1) membership check
        HashSet<int> present = new HashSet<int>(nums);
        List<int> result = new List<int>();
        // Any number in [1..n] absent from the set has disappeared
        for (int i = 1; i <= nums.Length; i++) {
            if (!present.Contains(i)) result.Add(i);
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def findDisappearedNumbers(self, nums: list[int]) -> list[int]:
        # Build a set of all present numbers for O(1) membership check
        present = set(nums)
        # Any number in [1..n] absent from the set has disappeared
        return [i for i in range(1, len(nums) + 1) if i not in present]
```

### Go

```
func findDisappearedNumbers(nums []int) []int {
    // Build a set of all present numbers for O(1) membership check
    present := make(map[int]bool, len(nums))
    for _, v := range nums {
        present[v] = true
    }
    result := []int{}
    // Any number in [1..n] absent from the set has disappeared
    for i := 1; i <= len(nums); i++ {
        if !present[i] {
            result = append(result, i)
        }
    }
    return result
}
```

### Rust

```
use std::collections::HashSet;

impl Solution {
    pub fn find_disappeared_numbers(nums: Vec<i32>) -> Vec<i32> {
        let n = nums.len() as i32;
        // Build a set of all present numbers for O(1) membership check
        let present: HashSet<i32> = nums.into_iter().collect();
        // Any number in [1..n] absent from the set has disappeared
        (1..=n).filter(|i| !present.contains(i)).collect()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [4, 3, 2, 7, 8, 2, 3, 1]` ($n = 8$)
Present: {1,2,3,4,7,8}. Scan 1..8: 5 and 6 are absent. Result: `[5, 6]`.

### 2. Slightly Complex
**Input:** `nums = [1, 1]` ($n = 2$)
Both elements are 1; 2 never appears. Present: {1}. Missing: {2}. Result: `[2]`. Duplicates cause the set to contain only one distinct entry, making the gap obvious.

### 3. Edge Case: Time Factor
**Input:** `nums = [2, 2, 2, ..., 2]` ($n = 10^5$, all twos)
Present: {2}. The scan loop checks all $10^5$ values in $[1, n]$; every number except 2 is missing. Maximum output size ($n - 1$ elements) and maximum scan length — worst case for both passes.

### 4. Edge Case: Space Factor
**Input:** `nums = [1, 2, 3, ..., n]` (no missing numbers, $n = 10^5$)
All values present; the HashSet holds $n$ entries — maximum space. The output list is empty. Tests the set at full capacity with zero output.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [n, n, n, ..., n]` (all equal to $n$, $n = 10^5$)
Present: {$n$}. Every number $1$ through $n-1$ is missing — the output list has $n - 1 \approx 10^5$ entries. Tests the boundary where the only present value is the maximum and the result is nearly as large as the input.
