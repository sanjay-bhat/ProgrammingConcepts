# LeetCode #1469: Find All Numbers that Disappeared

https://leetcode.com/problems/find-all-numbers-that-disappeared/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: HashSet with Count ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
For each number in $[1, n]$, scan the array to count its occurrences. Collect numbers with zero occurrences. $O(n)$ per check, $O(n^2)$ total.

### Optimal: HashSet with Count ★
Use a HashSet (or frequency map) to record every distinct value in one pass. Then scan $[1, n]$ and emit numbers absent from the set. Two passes, $O(n)$ time and $O(n)$ space — unlike the in-place negation trick, this handles repeated duplicates cleanly without modifying the input.

**Why this is better than Brute Force:** Constant-time membership checks replace the $O(n)$ inner scan, bringing total work from $O(n^2)$ to $O(n)$.

**Constraints:**
* $n == nums.length$
* $1 \leq n \leq 10^5$
* $1 \leq nums[i] \leq n$
* Numbers may appear more than twice


## Solutions

### C#

```
public class Solution {
    public IList<int> FindDisappearedNumbers(int[] nums) {
        // Record every distinct value present in the array
        HashSet<int> present = new HashSet<int>(nums);
        List<int> missing = new List<int>();
        // Emit each number in [1..n] that was never seen
        for (int i = 1; i <= nums.Length; i++) {
            if (!present.Contains(i)) missing.Add(i);
        }
        return missing;
    }
}
```

### Python

```
class Solution:
    def findDisappearedNumbers(self, nums: list[int]) -> list[int]:
        # Record every distinct value present in the array
        present = set(nums)
        # Emit each number in [1..n] that was never seen
        return [i for i in range(1, len(nums) + 1) if i not in present]
```

### Go

```
func findDisappearedNumbers(nums []int) []int {
    // Record every distinct value present in the array
    present := make(map[int]bool, len(nums))
    for _, v := range nums {
        present[v] = true
    }
    result := []int{}
    // Emit each number in [1..n] that was never seen
    for i := 1; i <= len(nums); i++ {
        if !present[i] {
            result = append(result, i)
        }
    }
    return result
}
```

### Rust

```
use std::collections::HashSet;

impl Solution {
    pub fn find_disappeared_numbers(nums: Vec<i32>) -> Vec<i32> {
        let n = nums.len() as i32;
        // Record every distinct value present in the array
        let present: HashSet<i32> = nums.into_iter().collect();
        // Emit each number in [1..n] that was never seen
        (1..=n).filter(|i| !present.contains(i)).collect()
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [4, 3, 2, 7, 8, 2, 3, 1]` ($n = 8$)
Present: {1,2,3,4,7,8}. Numbers 5 and 6 are missing from [1..8]. Result: `[5, 6]`.

### 2. Slightly Complex
**Input:** `nums = [1, 1, 1, 1]` ($n = 4$)
All four slots hold 1. Present: {1}. Numbers 2, 3, and 4 are missing. Result: `[2, 3, 4]`. A single value repeated many times leaves most of the range uncovered.

### 3. Edge Case: Time Factor
**Input:** `nums = [1, 1, 1, ..., 1]` ($n = 10^5$, all ones)
The set contains only {1}; the output scan checks all $10^5$ values. Maximum scan length with maximum output — both loops run their full $n$ iterations.

### 4. Edge Case: Space Factor
**Input:** `nums = [1, 2, 3, ..., n]` (perfect permutation, $n = 10^5$)
The HashSet holds all $n$ distinct values — peak space usage. Every check hits the set, so the output list is empty. Tests the set at full capacity with zero missing numbers.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [2, 2, 4, 4, 6, 6, ..., n, n]` (even numbers each doubled, $n$ even)
All odd numbers in $[1, n]$ are missing. With $n = 10^4$, the set holds $n/2$ entries and the output holds $n/2$ entries — exactly half full in both directions. Tests the case where missing and present are perfectly balanced.
