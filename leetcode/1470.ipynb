# LeetCode #1470: Shuffle the Array

https://leetcode.com/problems/shuffle-the-array/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n)$ | $O(n)$ |
| **Optimal: Index Interleaving ★** | $O(n)$ | $O(n)$ |

---

## Understanding the Methods

### Brute Force
Same as optimal — the output must contain all $2n$ elements, so $O(n)$ space for the result is unavoidable. Any extra scan beyond one pass is wasted work.

### Optimal: Index Interleaving ★
Allocate one output array of size $2n$. In a single pass over $i \in [0, n)$, place `nums[i]` (from the first half) at `result[2i]` and `nums[n+i]` (from the second half) at `result[2i+1]`. One loop, direct index mapping, no extra bookkeeping.

**Constraints:**
* $1 \leq n \leq 500$
* `nums.length == 2 * n`
* $1 \leq \text{nums}[i] \leq 10^3$


## Solutions

### C#

```
public class Solution {
    public int[] Shuffle(int[] nums, int n) {
        int[] result = new int[2 * n];
        // Interleave: x_i at even positions, y_i at odd positions
        for (int i = 0; i < n; i++) {
            result[2 * i] = nums[i];         // x half
            result[2 * i + 1] = nums[n + i]; // y half
        }
        return result;
    }
}
```

### Python

```
class Solution:
    def shuffle(self, nums: list[int], n: int) -> list[int]:
        result = [0] * (2 * n)
        # Interleave: x_i at even positions, y_i at odd positions
        for i in range(n):
            result[2 * i] = nums[i]       # x half
            result[2 * i + 1] = nums[n + i]  # y half
        return result
```

### Go

```
func shuffle(nums []int, n int) []int {
    result := make([]int, 2*n)
    // Interleave: x_i at even positions, y_i at odd positions
    for i := 0; i < n; i++ {
        result[2*i] = nums[i]       // x half
        result[2*i+1] = nums[n+i]   // y half
    }
    return result
}
```

### Rust

```
impl Solution {
    pub fn shuffle(nums: Vec<i32>, n: i32) -> Vec<i32> {
        let n = n as usize;
        let mut result = vec![0i32; 2 * n];
        // Interleave: x_i at even positions, y_i at odd positions
        for i in 0..n {
            result[2 * i] = nums[i];         // x half
            result[2 * i + 1] = nums[n + i]; // y half
        }
        result
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [2, 5, 1, 3, 4, 7]`, `n = 3`
First half: [2, 5, 1]; second half: [3, 4, 7]. Interleaved: positions 0,2,4 get 2,5,1 and positions 1,3,5 get 3,4,7. Result: `[2, 3, 5, 4, 1, 7]`.

### 2. Slightly Complex
**Input:** `nums = [1, 2, 3, 4, 4, 3, 2, 1]`, `n = 4`
Both halves are mirror-symmetric; interleaving still works correctly since it uses indices, not values. Result: `[1, 4, 2, 3, 3, 2, 4, 1]`.

### 3. Edge Case: Time Factor
**Input:** `nums = [1, 2, ..., 1000]` ($n = 500$), `n = 500`
Full $n = 500$ iterations; both halves of the loop body are always executed. Maximum loop count — every index pair is visited once.

### 4. Edge Case: Space Factor
**Input:** `nums = [1, 1]`, `n = 1`
Smallest possible input; result array has exactly 2 elements. Loop runs once: `result[0] = 1`, `result[1] = 1`. Returns `[1, 1]`.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [1000, 1000, ..., 1000]` (all 1000s), `n = 500`
All values identical; interleaving still executes correctly — no short-circuit on equality. The output is $2n$ copies of 1000. Tests that the algorithm doesn't assume distinct values.
