# LeetCode #1480: Running Sum of 1d Array

https://leetcode.com/problems/running-sum-of-1d-array/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(n)$ |
| **Optimal: Prefix Sum In-Place ★** | $O(n)$ | $O(1)$ extra |

---

## Understanding the Methods

### Brute Force
For each index $i$, sum all elements from 0 to $i$ using a nested loop. $O(n)$ work per element yields $O(n^2)$ overall — correct but needlessly slow.

### Optimal: Prefix Sum In-Place ★
A single forward pass: `nums[i] += nums[i-1]` propagates the running total into each cell, so every element becomes the cumulative sum up to that point. One pass, in-place modification, $O(1)$ extra space.

**Constraints:**
* $1 \leq \text{nums.length} \leq 1000$
* $-10^6 \leq \text{nums}[i] \leq 10^6$


## Solutions

### C#

```
public class Solution {
    public int[] RunningSum(int[] nums) {
        // Each position absorbs the cumulative total from its left neighbor
        for (int i = 1; i < nums.Length; i++)
            nums[i] += nums[i - 1];
        return nums;
    }
}
```

### Python

```
class Solution:
    def running_sum(self, nums: list[int]) -> list[int]:
        # Each position absorbs the cumulative total from its left neighbor
        for i in range(1, len(nums)):
            nums[i] += nums[i - 1]
        return nums
```

### Go

```
func runningSum(nums []int) []int {
    // Each position absorbs the cumulative total from its left neighbor
    for i := 1; i < len(nums); i++ {
        nums[i] += nums[i-1]
    }
    return nums
}
```

### Rust

```
impl Solution {
    pub fn running_sum(mut nums: Vec<i32>) -> Vec<i32> {
        // Each position absorbs the cumulative total from its left neighbor
        for i in 1..nums.len() {
            nums[i] += nums[i - 1];
        }
        nums
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [1, 2, 3, 4]`
Transforms in-place: $[1, 3, 6, 10]$. Each element accumulates the sum of all previous elements plus itself — the classic prefix sum.

### 2. Slightly Complex
**Input:** `nums = [1, 1, 1, 1, 1]`
Result: $[1, 2, 3, 4, 5]$. Uniform input produces an arithmetic sequence, confirming each cell correctly adds the previous prefix.

### 3. Edge Case: Time Factor
**Input:** `nums = [-1000000, -1000000, ..., -1000000]` (1000 entries)
All negatives; the loop still runs $n - 1 = 999$ iterations with no early exit. Final value is $-10^9$, which fits in a 32-bit signed integer ($-10^9 > -2^{31} \approx -2.1 \times 10^9$). Full scan, no overflow.

### 4. Edge Case: Space Factor
**Input:** `nums = [5]`
Single element; the loop body never executes. Returns `[5]` unchanged. In-place modification means zero extra allocation.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [1000000, -1000000, 1000000, -1000000]`
Alternating extremes; sums oscillate: $[10^6, 0, 10^6, 0]$. No overflow occurs since consecutive cancellation keeps values bounded within $\pm 10^6$. Confirms that in-place updates handle sign-alternating sequences correctly.
