# LeetCode #1491: Average Salary Excluding the Minimum and Maximum Salary

https://leetcode.com/problems/average-salary-excluding-the-minimum-and-maximum-salary/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force (Sort)** | $O(n \log n)$ | $O(1)$ |
| **Optimal: Single-Pass Linear Scan ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force (Sort)
Sort the salary array, sum elements from index 1 to $n-2$, divide by $n-2$. Correct, but $O(n \log n)$ for a problem that only needs min and max.

### Optimal: Single-Pass Linear Scan ★
Track `total`, `minVal`, and `maxVal` in one pass. Subtract both extremes from total, then divide by $n-2$. No sort needed — $O(n)$ time, $O(1)$ space.

**Why this is better than Sort:** Finding min and max is a linear operation; sorting is unnecessarily expensive when only the two extreme values matter.

**Constraints:**
* $3 \leq \text{salary.length} \leq 100$
* $1000 \leq \text{salary}[i] \leq 10^6$
* All integers in `salary` are unique


## Solutions

### C#

```
public class Solution {
    public double Average(int[] salary) {
        int min = int.MaxValue, max = int.MinValue;
        long total = 0;
        // Accumulate total while tracking the extremes to exclude
        foreach (int s in salary) {
            total += s;
            if (s < min) min = s;
            if (s > max) max = s;
        }
        // Divide by the count after removing both outliers
        return (double)(total - min - max) / (salary.Length - 2);
    }
}
```

### Python

```
class Solution:
    def average(self, salary: list[int]) -> float:
        # Accumulate total while tracking the extremes to exclude
        total = sum(salary)
        # Divide by the count after removing both outliers
        return (total - min(salary) - max(salary)) / (len(salary) - 2)
```

### Go

```
func average(salary []int) float64 {
    minVal, maxVal, total := salary[0], salary[0], 0
    // Accumulate total while tracking the extremes to exclude
    for _, s := range salary {
        total += s
        if s < minVal { minVal = s }
        if s > maxVal { maxVal = s }
    }
    // Divide by the count after removing both outliers
    return float64(total-minVal-maxVal) / float64(len(salary)-2)
}
```

### Rust

```
impl Solution {
    pub fn average(salary: Vec<i32>) -> f64 {
        let total: i64 = salary.iter().map(|&s| s as i64).sum();
        let min_val = *salary.iter().min().unwrap() as i64;
        let max_val = *salary.iter().max().unwrap() as i64;
        // Divide by the count after removing both outliers
        (total - min_val - max_val) as f64 / (salary.len() - 2) as f64
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `salary = [4000,3000,1000,2000]`
Total $= 10000$. Min $= 1000$, Max $= 4000$. Average $= (10000 - 1000 - 4000) / 2 = 2500.0$.

### 2. Slightly Complex
**Input:** `salary = [1000,2000,3000]`
Minimal valid input ($n=3$). Excludes 1000 and 3000; average of the sole remaining element $= 2000.0$.

### 3. Edge Case: Time Factor
**Input:** `salary` of length 100, values randomly distributed
All 100 elements visited exactly once. No sort; min/max found in the same pass as the sum — genuinely $O(n)$.

### 4. Edge Case: Space Factor
**Input:** Any `salary` array
Only four scalars: `total`, `min`, `max`, and the loop variable. $O(1)$ space — the linear scan stores nothing proportional to input size.

### 5. Almost-Impossible but Plausible
**Input:** `salary = [1000,1000000,1000000,1000]`
Duplicates at both extremes. The problem guarantees all values are unique, so this cannot actually occur — but implementations that assume `min ≠ max` or use index-based exclusion rather than value-based subtraction would still break. The value-subtraction approach is immune.
