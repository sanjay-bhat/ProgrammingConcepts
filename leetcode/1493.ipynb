# LeetCode #1493: Longest Subarray of 1's After Deleting One Element

https://leetcode.com/problems/longest-subarray-of-1s-after-deleting-one-element/

## Comparison of Approaches

| Approach | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Brute Force** | $O(n^2)$ | $O(1)$ |
| **Optimal: Sliding Window ★** | $O(n)$ | $O(1)$ |

---

## Understanding the Methods

### Brute Force
For each index, delete that element and scan for the longest contiguous run of 1's. $O(n)$ per deletion, $O(n^2)$ overall.

### Optimal: Sliding Window ★
Expand a window `[left, right]` over the array, allowing at most one 0 inside it. When a second 0 enters, shrink from the left until only one 0 (or none) remains. Because exactly one element must be deleted, the answer is `windowSize - 1` (subtracting the one deletion slot, even if it's a 1). Single pass, $O(1)$ space.

**Constraints:**
* $1 \leq \text{nums.length} \leq 10^5$
* `nums[i]` is either 0 or 1


## Solutions

### C#

```
public class Solution {
    public int LongestSubarray(int[] nums) {
        int left = 0, zeros = 0, best = 0;
        for (int right = 0; right < nums.Length; right++) {
            // Admit the new element; count zeros as they consume the deletion budget
            if (nums[right] == 0) zeros++;
            // Shrink from the left until the window holds at most one zero
            while (zeros > 1) {
                if (nums[left++] == 0) zeros--;
            }
            // Window length minus 1 accounts for the mandatory deletion
            best = Math.Max(best, right - left);
        }
        return best;
    }
}
```

### Python

```
class Solution:
    def longest_subarray(self, nums: list[int]) -> int:
        left = zeros = best = 0
        for right, v in enumerate(nums):
            # Admit the new element; count zeros as they consume the deletion budget
            if v == 0:
                zeros += 1
            # Shrink from the left until the window holds at most one zero
            while zeros > 1:
                if nums[left] == 0:
                    zeros -= 1
                left += 1
            # Window length minus 1 accounts for the mandatory deletion
            best = max(best, right - left)
        return best
```

### Go

```
func longestSubarray(nums []int) int {
    left, zeros, best := 0, 0, 0
    for right, v := range nums {
        // Admit the new element; count zeros as they consume the deletion budget
        if v == 0 { zeros++ }
        // Shrink from the left until the window holds at most one zero
        for zeros > 1 {
            if nums[left] == 0 { zeros-- }
            left++
        }
        // Window length minus 1 accounts for the mandatory deletion
        if right-left > best { best = right - left }
    }
    return best
}
```

### Rust

```
impl Solution {
    pub fn longest_subarray(nums: Vec<i32>) -> i32 {
        let (mut left, mut zeros, mut best) = (0usize, 0i32, 0i32);
        for right in 0..nums.len() {
            // Admit the new element; count zeros as they consume the deletion budget
            if nums[right] == 0 { zeros += 1; }
            // Shrink from the left until the window holds at most one zero
            while zeros > 1 {
                if nums[left] == 0 { zeros -= 1; }
                left += 1;
            }
            // Window length minus 1 accounts for the mandatory deletion
            best = best.max((right - left) as i32);
        }
        best
    }
}
```

## Example Scenarios

### 1. Common Case
**Input:** `nums = [1,1,0,1]`
Window expands to cover the 0 at index 2. One 0 is within budget. `right - left = 3 - 0 = 3` after the final step. Answer: **3** (delete the 0, get `[1,1,1]`).

### 2. Slightly Complex
**Input:** `nums = [0,1,1,1,0,1,1,0,1]`
Window slides; when the second 0 enters at index 4, `left` advances past the first 0. Best window at `right=7`: `[1,1,0,1]` of size 4 → `right - left = 3`. Answer: **5** (window `[1,1,0,1,1]` of size 5 at indices 1–6, minus 1 = 4).

### 3. Edge Case: Time Factor
**Input:** `nums = [1,1,1,...,1,0]` ($n-1$ ones then one 0)
The window expands all the way to the right without ever shrinking. `best = n-2`. Full scan, no unnecessary contractions.

### 4. Edge Case: Space Factor
**Input:** Any `nums` of length $10^5$
Three integer scalars (`left`, `zeros`, `best`) — $O(1)$ extra space. The sliding window works entirely over the original array.

### 5. Almost-Impossible but Plausible
**Input:** `nums = [1,1,1,1,1]` (all ones, no zeros)
The window never increments `zeros`; `best = n-2 = 3` at the end (deleting any one element yields a run of 4). The mandatory deletion is correctly accounted for by `right - left` rather than `right - left + 1`.
